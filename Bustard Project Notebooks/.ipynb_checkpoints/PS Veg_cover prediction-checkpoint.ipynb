```python
# 2. Inspect spatial properties
def extract_acquisition_date(filename):
    """
    Extract YYYYMMDD from the beginning of a typical PlanetScope filename.

    Example:
        20250512_103412_42_24f4_3B_AnalyticMS_SR_8b.tif
        -> 2025-05-12
    """
    match = re.search(r"(\d{8})", filename)

    if match is None:
        return None

    try:
        return pd.to_datetime(
            match.group(1),
            format="%Y%m%d"
        ).date()
    except ValueError:
        return None


records = []

for path in sr_files:
    with rasterio.open(path) as src:
        records.append({
            "file": path.name,
            "path": str(path),
            "date": extract_acquisition_date(path.name),
            "bands": src.count,
            "width": src.width,
            "height": src.height,
            "crs": str(src.crs),
            "x_resolution": abs(src.transform.a),
            "y_resolution": abs(src.transform.e),
            "dtype": src.dtypes[0],
            "nodata": src.nodata,
            "left": src.bounds.left,
            "bottom": src.bounds.bottom,
            "right": src.bounds.right,
            "top": src.bounds.top
        })

image_info = pd.DataFrame(records)

image_info.head()
```

```python
def extract_planetscope_metadata(path):
    """
    Extract acquisition date, time, and scene identifiers
    from a typical PlanetScope filename.
    """

    parts = path.stem.split("_")

    date = pd.to_datetime(
        parts[0],
        format="%Y%m%d",
        errors="coerce"
    )

    time_text = parts[1] if len(parts) > 1 else None
    satellite_id = parts[3] if len(parts) > 3 else None

    return {
        "date": date,
        "date_text": date.strftime("%Y-%m-%d") if not pd.isna(date) else "Unknown",
        "time": time_text,
        "satellite": satellite_id
    }


footprint_records = []

for tile_number, path in enumerate(sr_files, start=1):

    metadata = extract_planetscope_metadata(path)

    with rasterio.open(path) as src:

        footprint_records.append({
            "tile_number": tile_number,
            "filename": path.name,
            "path": str(path),
            "date": metadata["date"],
            "date_text": metadata["date_text"],
            "time": metadata["time"],
            "satellite": metadata["satellite"],
            "bands": src.count,
            "crs": str(src.crs),
            "resolution_x": abs(src.transform.a),
            "resolution_y": abs(src.transform.e),
            "geometry": box(
                src.bounds.left,
                src.bounds.bottom,
                src.bounds.right,
                src.bounds.top
            )
        })


footprints = gpd.GeoDataFrame(
    footprint_records,
    geometry="geometry",
    crs="EPSG:32630"
)

footprints[
    [
        "tile_number",
        "date_text",
        "time",
        "satellite",
        "filename"
    ]
]
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.lines import Line2D


def plot_tile_footprints(
    footprints,
    aoi=None,
    label_tiles=True,
    add_basemap=False,
    zoom_to="all",
    figsize=(13, 11)
):
    """
    Plot PlanetScope tile footprints and an optional AOI polygon.

    Parameters
    ----------
    footprints : GeoDataFrame
        PlanetScope tile polygons.

    aoi : GeoDataFrame, optional
        Area-of-interest polygon. It will be reprojected to the
        footprint CRS when necessary.

    label_tiles : bool
        Whether to label each tile with its tile number.

    add_basemap : bool
        Whether to add an OpenStreetMap basemap.

    zoom_to : {"all", "aoi"}
        "all" displays the full PlanetScope tile coverage.
        "aoi" zooms to the aquifer boundary with some padding.

    figsize : tuple
        Figure dimensions.
    """

    if footprints.empty:
        raise ValueError("The footprint GeoDataFrame is empty.")

    fig, ax = plt.subplots(figsize=figsize)

    unique_dates = sorted(
        footprints["date_text"].dropna().unique()
    )

    available_colors = [
        "tab:blue",
        "tab:orange",
        "tab:green",
        "tab:red",
        "tab:purple"
    ]

    date_colors = {
        date_text: available_colors[i % len(available_colors)]
        for i, date_text in enumerate(unique_dates)
    }

    # --------------------------------------------------------
    # Plot PlanetScope footprints
    # --------------------------------------------------------
    for date_text in unique_dates:

        subset = footprints[
            footprints["date_text"] == date_text
        ]

        subset.plot(
            ax=ax,
            facecolor=date_colors[date_text],
            edgecolor=date_colors[date_text],
            alpha=0.25,
            linewidth=1.3,
            zorder=2
        )

    # --------------------------------------------------------
    # Plot AOI
    # --------------------------------------------------------
    aoi_plot = None

    if aoi is not None:

        if aoi.empty:
            raise ValueError("The AOI GeoDataFrame is empty.")

        aoi_plot = aoi.copy()

        if aoi_plot.crs != footprints.crs:
            aoi_plot = aoi_plot.to_crs(footprints.crs)

        # Draw a light transparent fill and a strong boundary
        aoi_plot.plot(
            ax=ax,
            facecolor="none",
            edgecolor="black",
            linewidth=3.0,
            linestyle="--",
            zorder=5
        )

    # --------------------------------------------------------
    # Label individual tiles
    # --------------------------------------------------------
    if label_tiles:

        for _, row in footprints.iterrows():

            representative_point = (
                row.geometry.representative_point()
            )

            ax.text(
                representative_point.x,
                representative_point.y,
                str(row["tile_number"]),
                fontsize=8,
                ha="center",
                va="center",
                zorder=6,
                bbox={
                    "facecolor": "white",
                    "edgecolor": "black",
                    "linewidth": 0.3,
                    "alpha": 0.8,
                    "pad": 1.5
                }
            )

    # --------------------------------------------------------
    # Optional basemap
    # --------------------------------------------------------
    if add_basemap:

        try:
            import contextily as ctx

            ctx.add_basemap(
                ax,
                crs=footprints.crs.to_string(),
                source=ctx.providers.OpenStreetMap.Mapnik,
                zorder=1
            )

        except ImportError:
            print(
                "Contextily is not installed. Install it with:\n"
                "conda install -c conda-forge contextily"
            )

    # --------------------------------------------------------
    # Zoom to AOI if requested
    # --------------------------------------------------------
    if zoom_to == "aoi":

        if aoi_plot is None:
            raise ValueError(
                "zoom_to='aoi' requires an AOI GeoDataFrame."
            )

        xmin, ymin, xmax, ymax = aoi_plot.total_bounds

        x_padding = max((xmax - xmin) * 0.08, 100)
        y_padding = max((ymax - ymin) * 0.08, 100)

        ax.set_xlim(
            xmin - x_padding,
            xmax + x_padding
        )

        ax.set_ylim(
            ymin - y_padding,
            ymax + y_padding
        )

    elif zoom_to != "all":
        raise ValueError(
            "zoom_to must be either 'all' or 'aoi'."
        )

    # --------------------------------------------------------
    # Legend
    # --------------------------------------------------------
    legend_handles = [
        Patch(
            facecolor=date_colors[date_text],
            edgecolor=date_colors[date_text],
            alpha=0.35,
            label=date_text
        )
        for date_text in unique_dates
    ]

    if aoi_plot is not None:
        legend_handles.append(
            Line2D(
                [0],
                [0],
                color="black",
                linewidth=3,
                linestyle="--",
                label="Aquifer boundary"
            )
        )

    ax.legend(
        handles=legend_handles,
        title="Layers",
        loc="best"
    )

    ax.set_title(
        "PlanetScope tile footprints and aquifer boundary"
    )
    ax.set_xlabel("Easting, UTM Zone 30N (m)")
    ax.set_ylabel("Northing, UTM Zone 30N (m)")
    ax.set_aspect("equal")

    plt.tight_layout()
    plt.show()
```

```python
# 5. Memory-safe mosaic using GDAL VRT

from osgeo import gdal

gdal.UseExceptions()

def create_planetscope_mosaic(
    input_files,
    output_tif,
    src_nodata=0,
    output_nodata=0,
    compression="DEFLATE",
    predictor=2,
    overwrite=True
):
    """
    Create a memory-efficient PlanetScope mosaic using GDAL BuildVRT
    followed by GDAL Translate.

    Parameters
    ----------
    input_files : list of Path or str
        PlanetScope raster files to mosaic.

    output_tif : Path or str
        Output mosaic GeoTIFF.

    src_nodata : int or float, optional
        Nodata value in the source images. PlanetScope imagery often
        uses zero outside valid image footprints.

    output_nodata : int or float, optional
        Nodata value assigned to the output mosaic.

    compression : str
        GeoTIFF compression type, for example DEFLATE or LZW.

    predictor : int
        Predictor=2 is appropriate for integer reflectance values.

    overwrite : bool
        Whether to replace existing output files.
    """

    input_files = [str(Path(path)) for path in input_files]
    output_tif = Path(output_tif)
    output_tif.parent.mkdir(parents=True, exist_ok=True)

    if len(input_files) == 0:
        raise ValueError("No input files were provided.")

    vrt_path = output_tif.with_suffix(".vrt")

    if overwrite:
        for path in [output_tif, vrt_path]:
            if path.exists():
                path.unlink()

    print(f"Creating mosaic from {len(input_files)} images...")
    print("VRT:", vrt_path)
    print("Output:", output_tif)

    # Build a virtual raster.
    #
    # GDAL gives files later in the input list higher priority
    # where images overlap.
    vrt_options = gdal.BuildVRTOptions(
        resolution="highest",
        srcNodata=src_nodata,
        VRTNodata=output_nodata,
        resampleAlg="nearest"
    )

    vrt_dataset = gdal.BuildVRT(
        str(vrt_path),
        input_files,
        options=vrt_options
    )

    if vrt_dataset is None:
        raise RuntimeError("GDAL failed to create the VRT.")

    vrt_dataset.FlushCache()
    vrt_dataset = None

    # Convert VRT to a compressed tiled GeoTIFF
    translate_options = gdal.TranslateOptions(
        format="GTiff",
        noData=output_nodata,
        creationOptions=[
            "TILED=YES",
            f"COMPRESS={compression}",
            f"PREDICTOR={predictor}",
            "BIGTIFF=YES",
            "BLOCKXSIZE=512",
            "BLOCKYSIZE=512",
            "NUM_THREADS=ALL_CPUS"
        ]
    )

    output_dataset = gdal.Translate(
        str(output_tif),
        str(vrt_path),
        options=translate_options
    )

    if output_dataset is None:
        raise RuntimeError("GDAL failed to write the output mosaic.")

    output_dataset.FlushCache()
    output_dataset = None

    print("Mosaic completed successfully.")

    return output_tif
```

```python
import numpy as np
import matplotlib.pyplot as plt
from rasterio.enums import Resampling

def percentile_stretch(
    rgb,
    lower_percentile=2,
    upper_percentile=98
):
    """
    Apply an independent percentile stretch to each RGB channel.
    """

    stretched = np.zeros_like(rgb, dtype="float32")

    for band_index in range(3):
        band = rgb[band_index].astype("float32")

        valid = np.isfinite(band) & (band > 0)

        if not np.any(valid):
            continue

        low, high = np.percentile(
            band[valid],
            [lower_percentile, upper_percentile]
        )

        if high <= low:
            continue

        stretched[band_index] = np.clip(
            (band - low) / (high - low),
            0,
            1
        )

    return stretched
```

```python
import numpy as np
import matplotlib.pyplot as plt
import rasterio

from pathlib import Path
from rasterio.enums import Resampling


def plot_planetscope_rgb(
    tif_path,
    rgb_bands=(6, 4, 2),
    max_display_dimension=2500,
    lower_percentile=2,
    upper_percentile=98,
    figsize=(13, 10),
    title=None
):
    """
    Plot a memory-efficient RGB preview of a PlanetScope 8-band raster.

    Parameters
    ----------
    tif_path : str or Path
        Input PlanetScope GeoTIFF.

    rgb_bands : tuple, default=(6, 4, 2)
        PlanetScope 8-band natural-color combination:
        Red=6, Green=4, Blue=2.

    max_display_dimension : int
        Maximum preview width or height in pixels.

    lower_percentile, upper_percentile : float
        Percentile stretch limits.

    figsize : tuple
        Matplotlib figure size.

    title : str, optional
        Custom plot title.
    """

    tif_path = Path(tif_path)

    if not tif_path.exists():
        raise FileNotFoundError(
            f"Raster not found:\n{tif_path}"
        )

    with rasterio.open(tif_path) as src:

        if src.count < max(rgb_bands):
            raise ValueError(
                f"Raster has {src.count} bands, but bands "
                f"{rgb_bands} were requested."
            )

        # Calculate downsampling scale
        scale = min(
            1.0,
            max_display_dimension / max(src.width, src.height)
        )

        display_width = max(
            1,
            int(src.width * scale)
        )

        display_height = max(
            1,
            int(src.height * scale)
        )

        # Read RGB bands at reduced resolution
        rgb = src.read(
            list(rgb_bands),
            out_shape=(
                3,
                display_height,
                display_width
            ),
            resampling=Resampling.bilinear
        ).astype("float32")

        # Read valid-data mask
        valid_mask = src.read_masks(
            rgb_bands[0],
            out_shape=(
                display_height,
                display_width
            ),
            resampling=Resampling.nearest
        ) > 0

        bounds = src.bounds
        crs = src.crs

    # Also exclude zero/nodata pixels
    valid_mask &= np.all(
        np.isfinite(rgb) & (rgb > 0),
        axis=0
    )

    rgb_stretched = np.zeros_like(
        rgb,
        dtype="float32"
    )

    # Stretch each RGB band independently
    for band_index in range(3):

        band = rgb[band_index]
        valid_values = band[valid_mask]

        if valid_values.size == 0:
            continue

        low, high = np.percentile(
            valid_values,
            [lower_percentile, upper_percentile]
        )

        if high <= low:
            continue

        rgb_stretched[band_index] = np.clip(
            (band - low) / (high - low),
            0,
            1
        )

    # Convert from bands × rows × columns to rows × columns × bands
    rgb_display = np.moveaxis(
        rgb_stretched,
        0,
        -1
    )

    # Use transparency for nodata
    alpha = np.where(
        valid_mask,
        1.0,
        0.0
    )

    fig, ax = plt.subplots(
        figsize=figsize
    )

    ax.imshow(
        rgb_display,
        extent=[
            bounds.left,
            bounds.right,
            bounds.bottom,
            bounds.top
        ],
        origin="upper",
        alpha=alpha
    )

    if title is None:
        title = f"PlanetScope RGB mosaic: {tif_path.name}"

    ax.set_title(title)
    ax.set_xlabel("Easting")
    ax.set_ylabel("Northing")
    ax.set_aspect("equal")

    ax.text(
        0.01,
        0.01,
        f"CRS: {crs}",
        transform=ax.transAxes,
        fontsize=9,
        verticalalignment="bottom",
        bbox={
            "facecolor": "white",
            "alpha": 0.7,
            "edgecolor": "none"
        }
    )

    plt.tight_layout()
    plt.show()
```

```python
# RGB pre-view

import numpy as np
from rasterio.enums import Resampling


def read_rgb_preview(
    raster_path,
    rgb_bands=(6, 4, 2),
    max_dimension=600
):
    """
    Read a downsampled RGB preview without loading
    the full-resolution image.
    """

    with rasterio.open(raster_path) as src:

        scale = min(
            1.0,
            max_dimension / max(src.width, src.height)
        )

        preview_width = max(
            1,
            int(src.width * scale)
        )

        preview_height = max(
            1,
            int(src.height * scale)
        )

        rgb = src.read(
            list(rgb_bands),
            out_shape=(
                3,
                preview_height,
                preview_width
            ),
            resampling=Resampling.bilinear
        ).astype("float32")

        mask = src.read_masks(
            rgb_bands[0],
            out_shape=(
                preview_height,
                preview_width
            ),
            resampling=Resampling.nearest
        ) > 0

        bounds = src.bounds

    return rgb, mask, bounds
```

```python
def calculate_global_rgb_stretch(
    raster_files,
    rgb_bands=(6, 4, 2),
    preview_dimension=500,
    lower_percentile=2,
    upper_percentile=98,
    maximum_samples_per_band=100_000,
    random_seed=42
):
    """
    Calculate common RGB percentile limits using samples
    collected from all input images.
    """

    random_generator = np.random.default_rng(
        random_seed
    )

    band_samples = {
        0: [],
        1: [],
        2: []
    }

    for path in raster_files:

        rgb, mask, _ = read_rgb_preview(
            path,
            rgb_bands=rgb_bands,
            max_dimension=preview_dimension
        )

        for band_index in range(3):

            values = rgb[band_index][mask]

            values = values[
                np.isfinite(values) &
                (values > 0)
            ]

            if values.size > maximum_samples_per_band:

                selected_indices = random_generator.choice(
                    values.size,
                    size=maximum_samples_per_band,
                    replace=False
                )

                values = values[selected_indices]

            band_samples[band_index].append(values)

    lower_limits = []
    upper_limits = []

    for band_index in range(3):

        combined_values = np.concatenate(
            band_samples[band_index]
        )

        lower_limits.append(
            np.percentile(
                combined_values,
                lower_percentile
            )
        )

        upper_limits.append(
            np.percentile(
                combined_values,
                upper_percentile
            )
        )

    return (
        np.array(lower_limits, dtype="float32"),
        np.array(upper_limits, dtype="float32")
    )
```

```python
import geopandas as gpd
import rasterio
from rasterio.transform import Affine
import pandas as pd

# -------------------------------
# 1. File paths
# -------------------------------
ra_dir = 'W:/GIS_Data/SatelliteImagery/HighResolution/Morocco/AI_Baten_2025_July_psscene_analytic_8b_sr_udm2/PSScene/'
raster_path = ra_dir + '20250711_PS_bands_si.tif'
shp_dir = 'C:/Users/PangY/OneDrive - Smithsonian Institution/Bustard/01_Data/Reneco/Vegetation Cover and Food Availability Studies/'
points_path = shp_dir + 'Trophic survey_vegetation taxa cover_2016-2023_al_baten_prj.shp'

# -------------------------------
# 2. Load point shapefile
# -------------------------------
points = gpd.read_file(points_path)
print("Points loaded:", len(points))

# -------------------------------
# 3. Open raster and check CRS
# -------------------------------
try:
    with rasterio.open(raster_path) as src:
        raster_crs = src.crs
        raster_transform = src.transform
        band_count = src.count

        print(f"Raster opened. CRS: {raster_crs}, Bands: {band_count}")

        # Ensure CRS alignment
        if points.crs != raster_crs:
            print(f"CRS mismatch detected. Reprojecting points from {points.crs} to {raster_crs}...")
            points = points.to_crs(raster_crs)
            print("Points reprojected.")

        # -------------------------------
        # 4. Extract values for each band
        # -------------------------------
        coords = [(x,y) for x,y in zip(points.geometry.x, points.geometry.y)]
        print("Extracting values for all bands...")

        # Create dictionary to hold extracted band values
        band_data = {f'band_{i+1}': [] for i in range(band_count)}

        for i in range(1, band_count+1):
            values = [val[0] for val in src.sample(coords, indexes=i)]
            band_data[f'band_{i}'] = values
            print(f"Band {i} extraction complete.")

        # -------------------------------
        # 5. Add extracted values to GeoDataFrame
        # -------------------------------
        for band_name, values in band_data.items():
            points[band_name] = values

except rasterio.errors.RasterioIOError as e:
    print(f"Error opening raster file: {e}")
except Exception as e:
    print(f"Unexpected error: {e}")

# -------------------------------
# 6. Save results
# -------------------------------
output_path = "Trophic survey_vegetation taxa cover_2016-2023_al_baten_ravalues.shp"
points.to_file(output_path)
print(f"Saved updated shapefile with extracted raster values → {output_path}")

# Optional: also save as CSV (attributes only, no geometry)
points.drop(columns="geometry").to_csv("Trophic survey_vegetation taxa cover_2016-2023_al_baten_raster_values.csv", index=False)
print("CSV also saved.")
```
