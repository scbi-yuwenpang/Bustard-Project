# Vegetation Dynamics and Remote Sensing Analysis in Moroccan Drylands

## 1. Background

### 1.1 Study Area
- **Region:** North African drylands, Morocco  
- **Working areas:** Eastern Morocco and Tata Region  
- **Ecological focus:**  
  Vegetation dynamics across dryland habitat environments shaped by strong gradients in soil properties, hydrology, and geomorphological processes.

---

## 2. Study Objectives

This notebook focuses on **field-driven exploration and remote sensing-based detection of vegetation dynamics** in dryland ecosystems.

The main objectives are to:

1. **Characterize spatial and temporal vegetation dynamics** across habitats, including seasonal trajectories and interannual variability.
2. **Evaluate the performance of multiple satellite datasets (MODIS, Landsat, Sentinel-2) and spectral indices** in detecting and monitoring vegetation dynamics.
3. Quantify long-term vegetation **trends**.

---

## 3. Remote Sensing Datasets

- **Satellite products:**  
  - Landsat(long-term consistency, moderate resolution, 30m, 16-d)
  - Sentinel-2(high spatial resolution, high sensitivity, 10m, 5-d)
  - MODIS (MOD09A1)(high temporal frequency, ecosystem-scale variability, 500m, 8-d)

- **Temporal resolution:** ~5–16 days  
- **Spatial resolution:** 10–30–500 m  

### 3.1 Preprocessing (Google Earth Engine)
- Cloud and cloud-shadow masking  
- Seasonal composite generation  
- Tree cover removal following Reiner et al. (2023) to focus on herbaceous and shrub vegetation  

---

## 4. Spectral Indices

Spectral indices were computed to capture complementary aspects of vegetation condition:

- **Greenness indices:** NDVI, EVI, SAVI, MSAVI  
- **Disturbance indices:** NBR, NBR2  
- **Wetness indices:** NDMI, NDWI, MDWI  

---

## 5. Analysis Workflow

### 5.1 Representation Capacity: Field–Remote Sensing Linkage
(Objective 1)
- Summarize field-measured vegetation cover (F_alive) across habitats and working areas.
- Characterize seasonal and interannual variability in field vegetation data.
- Evaluate relationships between F_alive and spectral indices using:
    - Scatterplot analysis
    - Correlation analysis
    - Season-specific comparisons
---

### 5.2 Predictive Reconstruction: Random Forest Modeling
(Objective 2)
- Train Random Forest (RF) models to predict F_alive from multi-sensor spectral indices.
- Evaluate model performance (e.g., R², RMSE) and compare predictive accuracy across sensors.
- Assess variable importance to identify key spectral predictors.
- Apply trained RF models to reconstruct continuous vegetation time series across habitats and years.
---

### 5.3 Ecological Inference: Temporal Dynamics and Climate Drivers
(Objective 3)
- Analyze RF-predicted vegetation time series to:
    - Compare seasonal trajectories among habitats
    - Detect long-term trends using Mann–Kendall tests and LOWESS smoothing

- Integrate climate datasets (precipitation, temperature, and monthly anomalies).

- Conduct climate–vegetation coupling analyses:
    - Perform lagged correlation analyses between climate variables and RF-predicted F_alive
    - Quantify habitat-specific response times
    - Identify dominant climatic drivers of vegetation variability

---

```python
# ===============================
# 0. Import libraries
# ===============================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# ===============================
# 1. Read CSV files
# ===============================
# Notes and edits on 3/3/26
# added the harmonized habitat names, from '...01_Data\Reneco\Vegetation Cover and Food Availability Studies\Habitat type Trophic resource.xlsx'
# from the delivered data '...01_Data\Reneco\Vegetation Cover and Food Availability Studies\Habitat names standardised.xlsx'
# there were two columns: habitat_f1;habitat_f2
# added the accurate ssuvry dates in 'date' column, 
# from ...03_Analysis_Code\Bustard R\Bustard project\Trophic survey data analysis_all_2016-2023_date_03032026.csv
# from the delivered data '...01_Data\Reneco\Vegetation Cover and Food Availability Studies\Trophic survey data analysis_2016-2023_dates_03032026.XLS'

# the "Trophic survey data analysis_all_2016-2023_habitat.csv" file was updated via R
veg_df = pd.read_csv("Trophic survey data analysis_all_2016-2023_habitat.csv") #ALL SURVEY RECORDS, ALL INFORMATION WAS INCLUDED
# veg_date = pd.read_csv("Trophic survey data analysis_all_2016-2023_date_03032026.csv") #WITH ACCURATE SURVEY DATE;DOES NEED IT
station_info = pd.read_csv("Trophic_survey_data_114_station_habitat.csv") #STATION WITH HIBITAT AND WORKING_AREA
# veg_df_merged = pd.read_csv("Trophic survey data analysis_all_2016-2023_habitat_wa.csv")

# ===============================
# 2. Merge data frames by 'Station'
# ===============================
# veg_df_merged = veg_df.merge(
#     station_info[['Station', 'Work_area', 'Sub_area']], 
#     on='Station', 
#     how='left'
# )
# veg_df_merged['Year_Season'] = veg_df_merged['Year'].astype(str) + "_" + veg_df_merged['Season']
# Save merged CSV
# veg_df_merged.to_csv("Trophic survey data analysis_all_2016-2023_habitat_wa.csv", index=False)

# ===============================
# 3. Descriptive analysis of station info
# ===============================

# Number of stations per habitat
station_count_habitat = station_info.groupby('Habitat_f2').size().reset_index(name='Station_Count') #'Habitat';'Habitat_f1';"Habitat_f2"
print(station_count_habitat)

# Number of stations per habitat by work_area
station_count_habitat_area = station_info.groupby(['Work_area', 'Habitat_f2']).size().reset_index(name='Station_Count')
station_count_habitat_area = station_count_habitat_area.sort_values(['Work_area', 'Habitat_f2'])
print(station_count_habitat_area)
```

```python
# ---------------------------
# 2. Seasonal Trends of Vegetation Composition
# ---------------------------
def seasonal_trends(df, include_area=False):
    """
    Calculate seasonal mean vegetation values per habitat.
    Optionally include Work_area.
    """
    if include_area:
        group_cols = ['Season', 'Work_area', 'Habitat_f2'] #'Habitat
    else:
        group_cols = ['Season', 'Habitat_f2']
        
    df_grouped = df.groupby(group_cols).agg(
        F_alive_mean=('F_alive', 'mean'),
        F_dead_mean=('F_dead', 'mean'),
        F_ground_mean=('F_ground', 'mean')
    ).reset_index()
    
    return df_grouped
    
# Trends by habitat only
veg_df_merged =veg_df.copy()
trends_by_habitat = seasonal_trends(veg_df_merged, include_area=False)
# Trends by habitat + work_area
trends_by_area_habitat = seasonal_trends(veg_df_merged, include_area=True)

# ===============================
# 5. Prepare data for plotting
# ===============================
# Reorder seasons and habitats
season_order = ["Spring", "Summer", "Fall", "Winter"]
# habitat_order = ["Reg", "Macrochloa steppe", "Salty steppe", "Alluvial spread area", "Gully", "Wadi"] #for 'Habitat'
habitat_order = ["Reg", "Chamaephyte steppe", "Anabasis steppe", "Alfah-grass steppe","Salty steppe", "Spreading area", "Wadi and gullies"] #for 'Habitat_f2'

trends_by_habitat['Season'] = pd.Categorical(trends_by_habitat['Season'], categories=season_order, ordered=True)
trends_by_habitat['Habitat_f2'] = pd.Categorical(trends_by_habitat['Habitat_f2'], categories=habitat_order, ordered=True)

trends_by_area_habitat['Season'] = pd.Categorical(trends_by_area_habitat['Season'], categories=season_order, ordered=True)
trends_by_area_habitat['Habitat_f2'] = pd.Categorical(trends_by_area_habitat['Habitat_f2'], categories=habitat_order, ordered=True)

# Melt data to long format
trends_long = trends_by_habitat.melt(
    id_vars=['Season', 'Habitat_f2'],
    value_vars=['F_alive_mean', 'F_dead_mean', 'F_ground_mean'],
    var_name='Component',
    value_name='Value'
)

# Rename components
trends_long['Component'] = trends_long['Component'].replace({
    'F_alive_mean': 'Alive',
    'F_dead_mean': 'Dead',
    'F_ground_mean': 'Ground'
})
```

```python
# ===============================
# 5. Stacked bar plot: vegetation components
# ===============================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

import os

# Set your working directory
os.chdir(r"C:\Users\PangY\Python-GIS")

# Check the current working directory
# print("Current working directory:", os.getcwd())

# Create 'Plot_F_alive' folder if not exists
outdir = "Plot_F_alive"
os.makedirs(outdir, exist_ok=True)

# Components and colors
components = ['Alive', 'Dead', 'Ground']
colors = {'Alive':'green', 'Dead':'brown', 'Ground':'gray'}

# Stacked bar function
def stacked_bar(data, **kwargs):
    bottom = np.zeros(len(data['Season'].unique()))
    season_index = np.arange(len(data['Season'].unique()))
    season_labels = data['Season'].cat.categories
    
    for comp in components:
        values = data.loc[data['Component']==comp, 'Value'].values
        plt.bar(season_index, values, bottom=bottom, color=colors[comp], label=comp)
        bottom += values
    plt.xticks(season_index, season_labels, rotation=45)

# FacetGrid with smaller subplots
g = sns.FacetGrid(trends_long, col="Habitat_f2", col_wrap=3, height=2, sharey=True) #col="Habitat_f2"/"Habitat"
g.map_dataframe(stacked_bar)
g.set_titles("{col_name}")
g.set_axis_labels("Season", "Average Cover")

# Add a single legend in the top-right
handles, labels = plt.gca().get_legend_handles_labels()
g.fig.legend(handles[:len(components)], labels[:len(components)],
             title="Land Cover Component", loc='lower right', bbox_to_anchor=(0.92, 0.12))

outfile = os.path.join(outdir, "Seasonal_Veg_Stacked_Habitat_f2.png")
plt.tight_layout()
plt.savefig(outfile, dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd

rs_files = [
    'Landsat_FieldStations_TR_26.csv',
    'Sentinel2_FieldStations_TR_26.csv',
    'MODIS_FieldStations_TR_26.csv'
]

SENSOR_NAME = ['Landsat', 'Sentinel2', 'MODIS']

# Loop through each file
for sensor, file in zip(SENSOR_NAME, rs_files):
    df = pd.read_csv(file)

    # Parse date
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df = df.dropna(subset=['date'])
    print(f"{file} Size Before Date Filter:", df.size)
    
    # Filter years
    df = df[(df['date'].dt.year >= 2017) & (df['date'].dt.year <= 2025)]

    # Save output
    out_file = f"{sensor}_FieldStations_TR_26_17-25.csv"
    df.to_csv(out_file, index=False)

    print(f"\nFile processed: {file}")
    print(f"Saved as: {out_file}")
    print("Size after Date Filter:", df.size)
    print("Columns:", df.columns.tolist())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# -------------------------------
# 1. Load Station Metadata
# -------------------------------
# station_info_path = 'Trophic_survey_data_114_station_habitat_workingarea.csv'
# station_info = pd.read_csv(station_info_path)
station_info = pd.read_csv("Trophic_survey_data_114_station_habitat.csv") #STATION WITH HIBITAT AND WORKING_AREA

# -------------------------------
# 2. List of Remote Sensing Files
# -------------------------------
# old version only contains March-December
# rs_files = {
#     'Landsat': 'Landsat_FieldStations_TR.csv',
#     'Sentinel2': 'Sentinel2_FieldStations_TR.csv',
#     'MODIS': 'MODIS_FieldStations_TR.csv'
# }

# New version contain all year round
rs_files = {
    'Landsat': 'Landsat_FieldStations_TR_26.csv',
    'Sentinel2': 'Sentinel2_FieldStations_TR_26.csv',
    'MODIS': 'MODIS_FieldStations_TR_26.csv'
}
all_data_list = []

# -------------------------------
# 3. Process and Merge
# -------------------------------
# Updating the 'Habitat_f2' column
for sensor_name, file_name in rs_files.items():
    if os.path.exists(file_name):
        df = pd.read_csv(file_name)
        print(f"{sensor_name} DF shape: {df.shape}")
        df['Sensor'] = sensor_name

        # Merge Habitat_f2 from station_info
        df = pd.merge(
            df,
            station_info[['Station', 'Habitat_f2']],
            on='Station',
            how='left'
        )

        all_data_list.append(df)

    else:
        print(f"Warning: File {file_name} not found!")

full_df = pd.concat(all_data_list, ignore_index=True)
print(f"\nCombined DF shape: {full_df.shape}")
print("Columns:", full_df.columns.tolist())
# -------------------------------
# 4. Date Filtering (2017-2025)
# -------------------------------
full_df['date'] = pd.to_datetime(full_df['date'], errors='coerce')
full_df = full_df.dropna(subset=['date'])
full_df = full_df[(full_df['date'].dt.year >= 2017) & (full_df['date'].dt.year <= 2025)]

# Extract temporal features
full_df['Month_Num'] = full_df['date'].dt.month
full_df['Year_Month'] = full_df['date'].dt.to_period('M')

# Define seasons
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Fall'

full_df['Season'] = full_df['Month_Num'].apply(get_season)
full_df['Year_Season'] = full_df['date'].dt.year.astype(str) + '-' + full_df['Season']
```

**Updates(3/3/26):**

**Objectives:**

  - Train models to predict vegetation cover  
  - Calculate **feature importance** for each spectral index

```python
import pandas as pd
# --- Load the Data if Needed ---
veg_df = pd.read_csv('Trophic survey data analysis_all_2016-2023_habitat.csv') #the csv file was updated and inserted the date on 3/3/26

# --- Load Remote Sensing Spectral Index CSVs ---
rs_files = ['Landsat_FieldStations_TR_26.csv','Sentinel2_FieldStations_TR_26.csv','MODIS_FieldStations_TR_26.csv']
rs_dfs = [pd.read_csv(fp) for fp in rs_files]
# print(rs_dfs[0].columns.tolist())

# Updating the intergrating datasets
# ===========================
# Function 1: Temporal merge - Vegetation field data from 'GEE & Python Workflow.ipynb'
# ===========================
def merge_veg_with_rs(field_df: pd.DataFrame,
                      rs_df: pd.DataFrame,
                      rs_indices=None,
                      window_days: int = 8) -> pd.DataFrame:
    """
    Merge vegetation field data with RS indices using actual field collection dates.
    Matches closest RS image within a specified time window.

    Parameters
    ----------
    field_df : pd.DataFrame
        Must contain 'Station' and 'Date' (field sampling date).
    rs_df : pd.DataFrame
        Must contain 'Station' and 'date' (RS image date).
    rs_indices : list of str, optional
        List of RS indices to include.
    window_days : int
        Maximum allowed difference in days for temporal matching.

    Returns
    -------
    pd.DataFrame
        Merged dataset with closest RS match.
    """

    field_df = field_df.copy()
    rs_df = rs_df.copy()

    # -----------------------------
    # 1️ Convert dates to datetime
    # -----------------------------
    field_df['Date'] = pd.to_datetime(field_df['Date'])
    rs_df['date'] = pd.to_datetime(rs_df['date'])

    # -----------------------------
    # 2️ Define RS indices
    # -----------------------------
    if rs_indices is None:
        rs_indices = [i for i in
                      ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']
                      if i in rs_df.columns]

    # -----------------------------
    # 3️ Match closest RS date per field record
    # -----------------------------
    merged_records = []

    for _, row in field_df.iterrows():

        subset = rs_df[rs_df['Station'] == row['Station']].copy()

        if subset.empty:
            continue

        # Calculate absolute time difference
        subset['date_diff'] = (subset['date'] - row['Date']).abs()

        # Find closest RS date
        min_diff = subset['date_diff'].min()

        if min_diff <= pd.Timedelta(days=window_days):
            best = subset.loc[subset['date_diff'].idxmin()]

            merged_records.append({
                **row.to_dict(),
                **best[rs_indices].to_dict(),
                'RS_date': best['date'],
                'Date_diff_days': min_diff.days
            })

    merged_df = pd.DataFrame(merged_records)
    print(f"The size of the field-RS merged data set is {merged_df.shape}")
    return merged_df
```

```python
# --- Apply function for all datasets ---
landsat_veg_merged = merge_veg_with_rs(veg_df, rs_dfs[0])
sentinel2_veg_merged = merge_veg_with_rs(veg_df, rs_dfs[1])
modis_veg_merged = merge_veg_with_rs(veg_df, rs_dfs[2])

# # Save merged file
landsat_veg_merged.to_csv( "Merged_Veg_Landsat_26.csv", index=False)
sentinel2_veg_merged.to_csv("Merged_Veg_Sentinel2_26.csv", index=False)
modis_veg_merged.to_csv("Merged_Veg_Modis_26.csv", index=False)
```

```python
import os
import pandas as pd
from scipy.stats import pearsonr


def RS_F_alive_correlations_by_satellite_season(
        satellite_dfs,
        rs_indices,
        target='F_alive',
        season_col='Season',
        season_order=('Spring','Summer','Fall','Winter'),
        output_dir='RS_F_alive_summary'):

    """
    Calculate Pearson correlation between RS indices and F_alive
    by Satellite and Season (no habitat separation).
    """

    os.makedirs(output_dir, exist_ok=True)

    results = []

    for sat_name, df in satellite_dfs.items():

        if df is None or df.empty:
            print(f"{sat_name} dataframe empty.")
            continue

        # Keep required columns
        keep_cols = [season_col, target] + rs_indices
        keep_cols = [c for c in keep_cols if c in df.columns]

        df = df[keep_cols].dropna()

        if df.empty:
            print(f"{sat_name} has no valid data.")
            continue

        df[season_col] = pd.Categorical(
            df[season_col],
            categories=season_order,
            ordered=True
        )

        # -----------------------
        # ALL seasons combined
        # -----------------------
        for idx in rs_indices:

            if idx not in df.columns:
                continue

            if df[idx].nunique() < 3:
                continue

            r, p = pearsonr(df[idx], df[target])

            results.append({
                'Satellite': sat_name,
                'Season': 'All',
                'Index': idx,
                'Pearson_r': r,
                'R_squared': r**2,
                'p_value': p,
                'N': len(df)
            })

        # -----------------------
        # Seasonal correlations
        # -----------------------
        for season in season_order:

            df_season = df[df[season_col] == season]

            if len(df_season) < 5:
                continue

            for idx in rs_indices:

                if idx not in df_season.columns:
                    continue

                if df_season[idx].nunique() < 3:
                    continue

                r, p = pearsonr(df_season[idx], df_season[target])

                results.append({
                    'Satellite': sat_name,
                    'Season': season,
                    'Index': idx,
                    'Pearson_r': r,
                    'R_squared': r**2,
                    'p_value': p,
                    'N': len(df_season)
                })

    # -----------------------
    # Convert to dataframe
    # -----------------------
    corr_df = pd.DataFrame(results)

    if not corr_df.empty:
        corr_df[['Pearson_r','R_squared','p_value']] = \
            corr_df[['Pearson_r','R_squared','p_value']].round(3)

    # -----------------------
    # Export CSV
    # -----------------------
    output_file = os.path.join(
        output_dir,
        'RS_F_alive_correlations_by_satellite_season_26.csv'
    )

    corr_df.to_csv(output_file, index=False)

    print(f"Saved correlation summary: {output_file}")

    return corr_df
```

```python
satellite_dfs = {
    'Landsat': landsat_veg_merged,
    'Sentinel-2': sentinel2_veg_merged,
    'MODIS': modis_veg_merged
}

rs_indices = [
    'NDVI','EVI','SAVI','MSAVI',
    'NDMI','NBR','NBR2','NDWI','MNDWI'
]

corr_df = RS_F_alive_correlations_by_satellite_season(
    satellite_dfs,
    rs_indices
)

corr_df.head()
```

```python
import pandas as pd

# -----------------------------
# 1 Keep the "All" rows
# -----------------------------
df_all = corr_df[corr_df['Season'] == 'All']

# -----------------------------
# 2 Find best season per Satellite + Index
# -----------------------------
df_season = corr_df[corr_df['Season'] != 'All']

best_season = (
    df_season
    .sort_values('R_squared', ascending=False)
    .groupby(['Satellite','Index'])
    .first()
    .reset_index()
)

# -----------------------------
# 3 Combine All + Best season
# -----------------------------
summary_df = pd.concat([df_all, best_season], ignore_index=True)

# -----------------------------
# 4 Sort nicely
# -----------------------------
summary_df = summary_df.sort_values(
    ['Satellite','Index','Season']
)

summary_df = summary_df.reset_index(drop=True)

print(summary_df.head())

summary_file = os.path.join(
    output_dir,
    'RS_F_alive_corr_summary_best_season.csv'
)

summary_df.to_csv(summary_file, index=False)
print("Summary df was Saved:", summary_file)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
from scipy.stats import pearsonr

# ------------------------
# Configuration
# ------------------------
SATELLITE_NAME = 'Sentinel-2'  # or 'Landsat', 'Sentinel-2', 'MODIS'
rs_indices = ['NDVI', 'EVI', 'SAVI', 'MSAVI','NDMI', 'NBR', 'NBR2', 'NDWI', 'MNDWI']
season_order = ['Spring', 'Summer', 'Fall', 'Winter']

# Ensure corr_df is a copy to avoid SettingWithCopyWarning
df = satellite_dfs[SATELLITE_NAME]
df['Season'] = pd.Categorical(df['Season'], categories=season_order, ordered=True)

habitats = df['Habitat_f2'].unique()
palette = sns.color_palette('Set2', n_colors=len(season_order))
season_colors = dict(zip(season_order, palette))

# ------------------------
# Compute R² by habitat, RS index, and season
# ------------------------
summary_list = []

for habitat in habitats:
    df_hab = df[df['Habitat_f2'] == habitat]

    for idx in rs_indices:
        for season in season_order:
            df_season = df_hab[df_hab['Season'] == season][['F_alive', idx]].dropna()

            if len(df_season) < 3:
                r2 = np.nan
            else:
                r, _ = pearsonr(df_season[idx], df_season['F_alive'])
                r2 = r**2

            summary_list.append({
                'Habitat_f2': habitat,
                'RS_Index': idx,
                'Season': season,
                'R2': r2
            })

summary_df = pd.DataFrame(summary_list)
summary_df.sort_values(['Habitat_f2','RS_Index','Season'], inplace=True)

# ------------------------
# Plot combined heatmap figure
# ------------------------
n_habitats = len(habitats)
n_cols = 3
n_rows = int(np.ceil(n_habitats / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(10, 3 * n_rows))
axes = axes.flatten()

for ax, habitat in zip(axes, habitats):
    df_plot = summary_df[summary_df['Habitat_f2'] == habitat].pivot(
        index='RS_Index', columns='Season', values='R2'
    )
    df_plot = df_plot.reindex(index=rs_indices, columns=season_order)

    sns.heatmap(df_plot, annot=True, fmt=".2f", cmap='YlGnBu',
                cbar=ax==axes[0], ax=ax, cbar_kws={'label':'R²'})
    ax.set_title(habitat)
    ax.set_xlabel('Season')
    ax.set_ylabel('RS Index')

# Hide any unused axes
for i in range(len(habitats), len(axes)):
    fig.delaxes(axes[i])

fig.suptitle(f'R² between F_alive and RS indices ({SATELLITE_NAME}) by Habitat and Season',
             fontsize=16, y=1.02)
plt.tight_layout()
plt.show()

# ------------------------
# Optional: export combined figure
# ------------------------
output_dir = 'RS_F_alive_summary'
os.makedirs(output_dir, exist_ok=True)
fig_path = os.path.join(output_dir, f'{SATELLITE_NAME}_F_alive_R2.png')
fig.savefig(fig_path, dpi=300, bbox_inches='tight')
print(f"Saved combined heatmap figure: {fig_path}")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# ------------------------
# Configuration
# ------------------------
satellite_dfs = {
    'Landsat': landsat_veg_merged,
    'Sentinel-2': sentinel2_veg_merged,
    'MODIS': modis_veg_merged
}

rs_indices = [
    'NDVI', 'EVI', 'SAVI', 'MSAVI',
    'NDMI', 'NBR', 'NBR2', 'NDWI', 'MNDWI'
]

target = 'F_alive'
season_order = ['Spring', 'Summer', 'Fall', 'Winter']

# ------------------------
# Prepare results list
# ------------------------
results = []

for sat_name, df in satellite_dfs.items():
    
    # Keep only needed columns
    keep_cols = ['Habitat_f2', 'Season', target] + rs_indices
    df = df[keep_cols].dropna()
    
    # Ensure Season is categorical
    df['Season'] = pd.Categorical(df['Season'], categories=season_order, ordered=True)
    
    # Loop over habitats
    habitats = df['Habitat_f2'].unique()
    
    for habitat in habitats:
        df_hab = df[df['Habitat_f2'] == habitat]
        
        # ---------- Overall correlations ----------
        for idx in rs_indices:
            if df_hab[idx].nunique() < 3:
                continue
            r, p = pearsonr(df_hab[idx], df_hab[target])
            results.append({
                'Satellite': sat_name,
                'Habitat': habitat,
                'Index': idx,
                'Season': 'All',
                'Pearson_r': r,
                'R_squared': r**2,
                'p_value': p,
                'N': len(df_hab)
            })
        
        # ---------- Seasonal correlations ----------
        for season in season_order:
            df_season = df_hab[df_hab['Season'] == season]
            if len(df_season) < 5:
                continue
            for idx in rs_indices:
                if df_season[idx].nunique() < 3:
                    continue
                r, p = pearsonr(df_season[idx], df_season[target])
                results.append({
                    'Satellite': sat_name,
                    'Habitat': habitat,
                    'Index': idx,
                    'Season': season,
                    'Pearson_r': r,
                    'R_squared': r**2,
                    'p_value': p,
                    'N': len(df_season)
                })

# ------------------------
# Convert to DataFrame
# ------------------------
corr_df = pd.DataFrame(results)

# Optional: round numeric columns
corr_df[['Pearson_r','R_squared','p_value']] = corr_df[['Pearson_r','R_squared','p_value']].round(3)

# ------------------------
# Export CSV
# ------------------------
output_dir = 'RS_F_alive_summary'
os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, 'RS_F_alive_corr_by_satellite_season_habitat_f2.csv')
# output_file = 'RS_F_alive_correlations_by_satellite_season_habitat.csv'
corr_df.to_csv(output_file, index=False)
print(f"Saved correlation summary table: {output_file}")

# ------------------------
# Quick check
# ------------------------
corr_df.head(10)
```

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# -------------------------------
# Config
# -------------------------------
merged_files = {
    # 'Landsat': 'Merged_Veg_Landsat.csv',
    # 'Sentinel-2': 'Merged_Veg_Sentinel2.csv',
    # 'MODIS': 'Merged_Veg_Modis.csv'

    'Landsat': 'Merged_Veg_Landsat_26.csv',
    'Sentinel-2': 'Merged_Veg_Sentinel2_26.csv',
    'MODIS': 'Merged_Veg_Modis_26.csv'
}

rs_indices = ['NDVI', 'EVI', 'SAVI', 'MSAVI','NDMI', 'NBR', 'NBR2', 'NDWI', 'MNDWI']
target = 'F_alive'
season_order = ['Spring', 'Summer', 'Fall', 'Winter']
palette = sns.color_palette('Set2', n_colors=len(season_order))
season_colors = dict(zip(season_order, palette))

output_dir = 'Plot_RS_Corr'
os.makedirs(output_dir, exist_ok=True)

# -------------------------------
# Loop over sensors
# -------------------------------
for SATELLITE_NAME, file_name in merged_files.items():
    
    if not os.path.exists(file_name):
        print(f"Warning: {file_name} not found, skipping {SATELLITE_NAME}")
        continue

    plot_df = pd.read_csv(file_name)
    plot_df['Season'] = pd.Categorical(plot_df['Season'], categories=season_order, ordered=True)
    
    fig, axes = plt.subplots(3, 3, figsize=(14, 12), sharey=True)
    axes = axes.flatten()
    legend_handles = []
    legend_labels = []

    for ax, idx in zip(axes, rs_indices):
        # ------------------
        # Overall regression (ALL seasons) – black line
        # ------------------
        sns.regplot(
            data=plot_df,
            x=idx,
            y=target,
            scatter=False,
            ci=None,
            ax=ax,
            color='black',
            line_kws={'linewidth': 2.2, 'linestyle': '--'}
        )

        # Overall correlation
        r_all, _ = pearsonr(plot_df[idx], plot_df[target])
        r2_all = r_all ** 2
        ax.text(0.02, 0.98, f'All: R²={r2_all:.2f}', transform=ax.transAxes, fontsize=9, fontweight='bold', color='black', va='top')

        # ------------------
        # Season-specific lines
        # ------------------
        y_text = 0.88
        for season in season_order:
            df_season = plot_df[plot_df['Season'] == season]
            if len(df_season) < 3:
                continue

            sc = ax.scatter(df_season[idx], df_season[target], color=season_colors[season], alpha=0.6, s=25)
            sns.regplot(
                data=df_season,
                x=idx,
                y=target,
                scatter=False,
                ci=None,
                ax=ax,
                color=season_colors[season],
                line_kws={'linewidth': 1.6}
            )

            r, _ = pearsonr(df_season[idx], df_season[target])
            r2 = r ** 2
            ax.text(0.02, y_text, f'{season}: r²={r2:.2f}', transform=ax.transAxes, fontsize=8, color=season_colors[season], va='top')
            y_text -= 0.08

            # Collect legend once
            if idx == rs_indices[0]:
                legend_handles.append(sc)
                legend_labels.append(season)

        ax.set_title(idx, fontsize=11)
        ax.set_xlabel(idx)
        ax.grid(alpha=0.3)

    # Shared Y label
    fig.text(0.04, 0.5, target, va='center', rotation='vertical', fontsize=12)

    # Shared legend
    fig.legend(legend_handles, legend_labels, title='Season', loc='upper center', ncol=1, frameon=False, bbox_to_anchor=(0.95, 0.31))

    # Figure title with satellite name
    fig.suptitle(f'Seasonal and Overall Relationships between RS Indices and {target} ({SATELLITE_NAME})', fontsize=14, y=0.94)

    plt.tight_layout(rect=[0.04, 0.04, 1, 0.95])

    # Save figure per sensor
    fig_path = os.path.join(output_dir, f'Seasonal_Corr_{SATELLITE_NAME}_26.png')
    plt.savefig(fig_path, dpi=300)
    print(f"Saved figure: {fig_path}")
    plt.show()
    plt.close()
```

```python
# load the anecessery moudels
import os
print(os.getcwd())

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, roc_auc_score, accuracy_score
import seaborn as sns

# --- Load Field Data ---
# veg_df = pd.read_csv('Trophic survey data analysis_all_2016-2023_habitat_wa.csv')
veg_df = pd.read_csv('Trophic survey data analysis_all_2016-2023_habitat.csv') #updated on 3/3
print(f"Vegetation dataframe: {veg_df.shape} columns: {veg_df.columns.tolist()}")

# --- Load Field-RS Merged files  ---
merged_files = ['Merged_Veg_Landsat_26.csv',
                'Merged_Veg_Sentinel2_26.csv',
                'Merged_Veg_Modis_26.csv'
                # "Merged_Veg_Combined.csv"
               ]

# merged_dfs = [pd.read_csv(fp) for fp in merged_files]
landsat_veg_merged = pd.read_csv(merged_files[0])
sentinel2_veg_merged = pd.read_csv(merged_files[1])
modis_veg_merged = pd.read_csv(merged_files[2])
# combined_veg = pd.read_csv(merged_files[3])
```

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, mean_squared_error
from sklearn.model_selection import train_test_split, cross_val_score, KFold
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import textwrap
import os
import numpy as np

import matplotlib.pyplot as plt

# Global index color palette (similar to your vertical feature importance plot)
index_list = ['NDVI','EVI','SAVI','MSAVI','NBR','NBR2','NDMI','NDWI','MNDWI']
colors = plt.cm.tab10.colors  # 10 distinct colors from matplotlib
index_palette = {idx: colors[i % len(colors)] for i, idx in enumerate(index_list)}


def rf_vegetation_analysis(
    df, 
    target_col='F_alive',
    habitat_col='Habitat_f2', #Habitat or 'Habitat_f2'
    satellite_name='RS',
    save_dir='Plot_RF_Pred',
    by_habitat=False,
    n_estimators=500,
    random_state=42,
    n_splits=5
):

    if df.empty:
        print("Input dataframe is empty.")
        return None

    df = df.copy()
    os.makedirs(save_dir, exist_ok=True)

    feature_cols = [c for c in df.columns if c.upper() in
                    ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']]

    if not feature_cols:
        raise ValueError("No valid vegetation index columns found.")

    df = df.dropna(subset=[target_col])
    X = df[feature_cols]
    y = df[target_col]

    # Detect regression vs classification
    if y.nunique() <= 10 and y.dtype in ['object', 'category', 'bool', 'int']:
        model_type = 'classification'
        model = RandomForestClassifier(
            n_estimators=n_estimators,
            random_state=random_state,
            n_jobs=-1,
            oob_score=True,
            bootstrap=True
        )
        scoring = 'accuracy'
    else:
        model_type = 'regression'
        model = RandomForestRegressor(
            n_estimators=n_estimators,
            random_state=random_state,
            n_jobs=-1,
            oob_score=True,
            bootstrap=True
        )
        scoring = 'r2'

    # ==========================
    # Train/Test Split
    # ==========================
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=random_state
    )

    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # ==========================
    # Metrics
    # ==========================
    if model_type == 'regression':

        train_r2 = r2_score(y_train, y_train_pred)
        test_r2 = r2_score(y_test, y_test_pred)

        train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
        test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))

        oob_score = model.oob_score_

        kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
        cv_scores = cross_val_score(model, X_train, y_train,
                                    cv=kf, scoring='r2', n_jobs=-1)
        mean_cv = np.mean(cv_scores)

        print(f"\n{satellite_name} Random Forest (Regression):")
        print(f"  Train R²: {train_r2:.3f}")
        print(f"  Test R²: {test_r2:.3f}")
        print(f"  OOB R²: {oob_score:.3f}")
        print(f"  Train RMSE: {train_rmse:.3f}")
        print(f"  Test RMSE: {test_rmse:.3f}")
        print(f"  Mean 5-fold CV R²: {mean_cv:.3f}")

    else:
        train_r2 = accuracy_score(y_train, y_train_pred)
        test_r2 = accuracy_score(y_test, y_test_pred)
        oob_score = model.oob_score_

        kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
        cv_scores = cross_val_score(model, X_train, y_train,
                                    cv=kf, scoring='accuracy', n_jobs=-1)
        mean_cv = np.mean(cv_scores)

        print(f"\n{satellite_name} Random Forest (Classification):")
        print(f"  Train Accuracy: {train_r2:.3f}")
        print(f"  Test Accuracy: {test_r2:.3f}")
        print(f"  OOB Accuracy: {oob_score:.3f}")
        print(f"  Mean 5-fold CV Accuracy: {mean_cv:.3f}")

    metrics_dict = {
        'Satellite': satellite_name,
        'Model_Type': model_type,
        'Train_R2': train_r2,
        'Test_R2': test_r2,
        'OOB_R2': oob_score,
        'Train_RMSE': train_rmse,
        'Test_RMSE': test_rmse,
        'CV_R2_mean': mean_cv,
        'N_samples': len(df)
    }
    metrics_df = pd.DataFrame([metrics_dict])
    metrics_df.to_csv(f"{save_dir}/{satellite_name}_RF_ModelMetrics.csv",index=False)
    
    # ==========================
    # Predicted vs Observed Plot
    # ==========================          
    if model_type == 'regression':

        df_test = pd.DataFrame({
            'Observed': y_test, #y_train;y_test
            'Predicted': y_test_pred #y_train_pred;y_test_pred
        })

        df_train = pd.DataFrame({
            'Observed': y_train, #y_train;y_test
            'Predicted': y_train_pred #y_train_pred;y_test_pred
        })
        
        if 'Season' in df.columns:
            df_train['Season'] = df.loc[y_train.index, 'Season']
            df_test['Season'] = df.loc[y_test.index, 'Season']

        plt.figure(figsize=(5, 5))
        sns.scatterplot(data=df_train, #df_test
                        x='Observed',
                        y='Predicted',
                        hue='Season' if 'Season' in df_train else None,
                        alpha=0.7)
        plt.plot([df_train.Observed.min(), df_train.Observed.max()],
                 [df_train.Observed.min(), df_train.Observed.max()],
                 'k--')
        plt.title(f'{satellite_name} - Predicted vs Observed on train set') # on test set or train set
        plt.tight_layout()
        plt.savefig(f"{save_dir}/{satellite_name}_Pred_vs_Obs_train.png", dpi=300)
        plt.show()

        plt.figure(figsize=(5, 5))
        sns.scatterplot(data=df_test, #df_test
                        x='Observed',
                        y='Predicted',
                        hue='Season' if 'Season' in df_test else None,
                        alpha=0.7)
        plt.plot([df_test.Observed.min(), df_test.Observed.max()],
                 [df_test.Observed.min(), df_test.Observed.max()],
                 'k--')
        plt.title(f'{satellite_name} - Predicted vs Observed on test set') # on test set or train set
        plt.tight_layout()
        plt.savefig(f"{save_dir}/{satellite_name}_Pred_vs_Obs_test.png", dpi=300)
        plt.show()
        
        # ==========================
        # Residual Plot
        # ==========================
        df_test['Residual'] = df_test['Observed'] - df_test['Predicted']

        plt.figure(figsize=(6, 4))
        sns.scatterplot(x='Predicted', y='Residual',
                        data=df_test, alpha=0.7)
        plt.axhline(0, linestyle='--')
        plt.title(f'{satellite_name} - Residual Plot')
        plt.tight_layout()
        plt.savefig(f"{save_dir}/{satellite_name}_Residuals.png", dpi=300)
        plt.show()

    # ==========================
    # Feature Importance
    # ==========================
    importance_df = pd.DataFrame({
        'Index': feature_cols,
        'Importance': model.feature_importances_
    }).sort_values('Importance', ascending=False)

    plt.figure(figsize=(6, 4))
    sns.barplot(x='Importance', y='Index',
                data=importance_df,
                color='skyblue')
    plt.title(f'{satellite_name} - Feature Importance')
    plt.tight_layout()
    plt.savefig(f"{save_dir}/{satellite_name}_FeatureImportance.png", dpi=300)
    plt.show()

    # ==========================
    # Habitat-Specific Importance
    # ==========================
    habitat_importance_df = None
    if by_habitat and habitat_col and habitat_col in df.columns:

        habitat_importances = []

        for habitat, subset in df.groupby(habitat_col):
            if len(subset) < n_splits * 2:
                continue

            X_h = subset[feature_cols]
            y_h = subset[target_col]

            rf_h = RandomForestRegressor(
                n_estimators=n_estimators,
                random_state=random_state,
                n_jobs=-1
            )

            rf_h.fit(X_h, y_h)

            imp_h = pd.DataFrame({
                'Habitat': habitat,
                'Index': feature_cols,
                'Importance': rf_h.feature_importances_
            })

            habitat_importances.append(imp_h)

        habitat_importance_df = pd.concat(habitat_importances)

        # plt.figure(figsize=(8, 6))
        # sns.barplot(x='Importance', y='Index',
        #             hue='Habitat',
        #             data=habitat_importance_df)
        # plt.title(f'{satellite_name} - Feature Importance by Habitat')
        # plt.tight_layout()
        # plt.savefig(f"{save_dir}/{satellite_name}_FeatureImportance_ByHabitat.png", dpi=300)
        # plt.show()

        habitat_importance_df.to_csv(
            f"{save_dir}/{satellite_name}_HabitatImportance.csv",
            index=False
        )

    # # ==========================
    # # Vertical Habitat Importance Plot with consistent colors
    # # ==========================
    # plt.figure(figsize=(8, 5))
    # sns.barplot(
    # data=habitat_importance_df,
    # x='Habitat',
    # y='Importance',
    # hue='Index',
    # dodge=True,
    # palette=index_palette,
    # hue_order=index_list  # ensures indices follow the global palette order
    # )
    # plt.title(f'{satellite_name} - Feature Importance by Habitat', fontsize=12)
    # plt.xlabel('Habitat Type', fontsize=10)
    # plt.ylabel('Importance Rate', fontsize=11)
    
    # # Wrap labels
    # labels = [label.get_text() for label in ax.get_xticklabels()]
    # wrapped_labels = ['\n'.join(textwrap.wrap(label, width=12)) for label in labels]

    # ax.set_xticks(range(len(wrapped_labels)))
    # ax.set_xticklabels(wrapped_labels)

    # # plt.xticks(rotation=30)
    # # Option B — Put legend inside the plot (top-right corner)
    # plt.legend(title='RS Index', loc='upper right')
    # # plt.legend(title='RS Index', bbox_to_anchor=(1.05, 1), loc='upper right')
    # plt.tight_layout()

    # plt.savefig(
    #     f"{save_dir}/{satellite_name}_FeatureImportance_ByHabitat_Vertical.png",dpi=300)
    # plt.show()
    
    # ==========================
    # Return
    # ==========================
    df_out = df.copy()
    df_out['Predicted'] = model.predict(X)

    return importance_df, habitat_importance_df, df_out
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import textwrap
import os

# -----------------------------
# Input files
# -----------------------------
imp_file = [
    'Plot_RF_Pred/Landsat_HabitatImportance.csv',
    'Plot_RF_Pred/Sentinel-2_HabitatImportance.csv',
    'Plot_RF_Pred/MODIS_HabitatImportance.csv'
]

habitat_order = [
    "Reg",
    "Chamaephyte steppe",
    "Anabasis steppe",
    "Alfah-grass steppe",
    "Salty steppe",
    "Spreading area",
    "Wadi and gullies"
]

save_dir = "Plot_RF_Pred"
os.makedirs(save_dir, exist_ok=True)

# -----------------------------
# Load data
# -----------------------------
landsat_hab = pd.read_csv(imp_file[0])
sentinel_hab = pd.read_csv(imp_file[1])
modis_hab = pd.read_csv(imp_file[2])

landsat_hab["Satellite"] = "Landsat"
sentinel_hab["Satellite"] = "Sentinel-2"
modis_hab["Satellite"] = "MODIS"

combined_imp_df = pd.concat([landsat_hab, sentinel_hab, modis_hab], ignore_index=True)

# enforce habitat order
combined_imp_df["Habitat"] = pd.Categorical(
    combined_imp_df["Habitat"],
    categories=habitat_order,
    ordered=True
)

combined_imp_df.to_csv(f"{save_dir}/Combined_HabitatImportance_26.csv", index=False)

# -----------------------------
# Global index color palette
# -----------------------------
index_list = ['NDVI','EVI','SAVI','MSAVI','NBR','NBR2','NDMI','NDWI','MNDWI']
colors = plt.cm.tab10.colors
index_palette = {idx: colors[i % len(colors)] for i, idx in enumerate(index_list)}

# -----------------------------
# Function: vertical plot
# -----------------------------
def plot_vertical_importance(df, satellite_name):

    plt.figure(figsize=(8,5))

    ax = sns.barplot(
        data=df,
        x="Habitat",
        y="Importance",
        hue="Index",
        palette=index_palette,
        hue_order=index_list,
        order=habitat_order
    )

    plt.title(f"{satellite_name} - Feature Importance by Habitat")
    plt.xlabel("Habitat Type")
    plt.ylabel("Importance Rate")

    # wrap labels
    labels = [l.get_text() for l in ax.get_xticklabels()]
    wrapped = ['\n'.join(textwrap.wrap(label,12)) for label in labels]

    ax.set_xticks(range(len(wrapped)))
    ax.set_xticklabels(wrapped)

    plt.legend(title="RS Index", loc="upper right")
    plt.tight_layout()

    save_path = f"{save_dir}/{satellite_name}_FeatureImportance_ByHabitat_Vertical.png"
    plt.savefig(save_path, dpi=300)
    plt.show()

    print(f"Saved: {save_path}")

# -----------------------------
# Generate plots per satellite
# -----------------------------
for sat in ["Landsat","Sentinel-2","MODIS"]:

    df_sat = combined_imp_df[combined_imp_df["Satellite"]==sat]

    plot_vertical_importance(df_sat, sat)
```

```python
import os
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt

# Global index color palette (similar to your vertical feature importance plot)
index_list = ['NDVI','EVI','SAVI','MSAVI','NBR','NBR2','NDMI','NDWI','MNDWI']
colors = plt.cm.tab10.colors  # 10 distinct colors from matplotlib
index_palette = {idx: colors[i % len(colors)] for i, idx in enumerate(index_list)}

habitat_order = [
    "Reg",
    "Chamaephyte steppe",
    "Anabasis steppe",
    "Alfah-grass steppe",
    "Salty steppe",
    "Spreading area",
    "Wadi and gullies"
]

def plot_best_index_summary(df, save_dir, habitat_order):
    """
    Plot the top vegetation index per habitat per satellite with consistent colors
    and visible borders for each bar using global index_palette.
    """
    
    global index_palette  # use the global palette
    
    # Keep only the top index per habitat per satellite
    df_top = df.sort_values('Importance', ascending=False).groupby(['Habitat','Satellite']).head(1).reset_index(drop=True)
    df_top["Habitat"] = pd.Categorical(df_top["Habitat"],categories=habitat_order,ordered=True)
    df_top = df_top.sort_values("Habitat")

    plt.figure(figsize=(8, 5))
    ax = plt.gca()
    
    satellite_order = ["Landsat", "Sentinel-2", "MODIS"]
    habitat_order = df_top["Habitat"].unique()
    bar_width = 0.2
    
    for i, hab in enumerate(habitat_order):
        df_hab = df_top[df_top["Habitat"] == hab]
        for j, sat in enumerate(satellite_order):
            df_row = df_hab[df_hab["Satellite"] == sat]
            if df_row.empty:
                continue
            df_row = df_row.iloc[0]
            
            x = i + (j - 1) * bar_width  # side-by-side bars
            ax.bar(
                x,
                df_row["Importance"],
                width=bar_width,
                color=index_palette.get(df_row["Index"], 'grey'),  # fallback color
                edgecolor="black",
                linewidth=1.2
            )
            
            # Add satellite name on top of first habitat
            if i == 0:
                ax.text(
                    x,
                    df_row["Importance"],
                    sat,
                    ha='center',
                    va='bottom',
                    weight='bold',
                    fontsize=10
                )
    
    # X-axis labels
    # Wrap habitat labels
    wrapped = ["\n".join(textwrap.wrap(label, 14))for label in habitat_order]
    ax.set_xticks(range(len(wrapped)))
    ax.set_xticklabels(wrapped)
    
    # ax.set_xticks(range(len(habitat_order)))
    # ax.set_xticklabels(habitat_order, rotation=0, ha='center')
    ax.set_ylabel("Importance Rate")
    ax.set_xlabel("Habitat")
    ax.set_title("Top Vegetation Index per Habitat Across Satellites")
    
    # Legend for all indices using global palette
    legend_handles = [mpatches.Patch(color=index_palette[idx], label=idx) for idx in index_palette]
    ax.legend(handles=legend_handles, title="RS Index", loc='upper right')
    
    plt.tight_layout()
    os.makedirs(save_dir, exist_ok=True)
    save_path = os.path.join(save_dir, "BestIndex_Summary_Bars.png")
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"Summary plot saved to: {save_path}")
```

#### Discarded RF models

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, make_scorer
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import cross_val_score, KFold
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np

def rf_vegetation_analysis(df, target_col='F_alive',habitat_col=None,
                           satellite_name='RS',save_dir='Plot_RF_Pred',by_habitat=False,
                           n_estimators=500,random_state=42,n_splits=5):
    """
    Compute and plot Random Forest feature importance for vegetation indices.
    Optionally compute habitat-specific importance if habitat_col is provided.
    Automatically detects regression or classification model based on target variable type.
    Includes 5-fold cross-validation for more reliable model performance estimates.
    """

    if df.empty:
        print("Input dataframe is empty.")
        return None

    df = df.copy()
    os.makedirs(save_dir, exist_ok=True)

    # Select vegetation index columns
    feature_cols = [c for c in df.columns if c.upper() in
                    ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']]
    if not feature_cols:
        raise ValueError("No valid vegetation index columns found in dataframe.")

    df = df.dropna(subset=[target_col])
    X = df[feature_cols]
    y = df[target_col]

    # Detect model type automatically
    if y.nunique() <= 10 and y.dtype in ['object', 'category', 'bool', 'int']:
        model_type = 'classification'
        model = RandomForestClassifier(n_estimators=n_estimators,
                                       random_state=random_state, n_jobs=-1)
        scoring = 'accuracy'
    else:
        model_type = 'regression'
        model = RandomForestRegressor(n_estimators=n_estimators,
                                      random_state=random_state, n_jobs=-1)
        scoring = make_scorer(r2_score)

    # =====================
    # Model fitting & score
    # =====================
    model.fit(X, y)
    y_pred = model.predict(X)

    if model_type == 'regression':
        train_score = r2_score(y, y_pred)
    else:
        train_score = accuracy_score(y, y_pred)

    # 5-fold CV
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    cv_scores = cross_val_score(model, X, y, cv=kf, scoring=scoring, n_jobs=-1)
    mean_cv_score = np.mean(cv_scores)

    # Print model summary
    print(f"\n {satellite_name} Random Forest ({model_type.capitalize()}) Results:")
    if model_type == 'regression':
        print(f"  - Training R²: {train_score:.3f}")
        print(f"  - Mean 5-fold CV R²: {mean_cv_score:.3f}")
    else:
        print(f"  - Training Accuracy: {train_score:.3f}")
        print(f"  - Mean 5-fold CV Accuracy: {mean_cv_score:.3f}")

    # =====================
    # General Feature Importance
    # =====================
    importance_df = (
        pd.DataFrame({'Index': feature_cols,
                      'Importance': model.feature_importances_})
        .sort_values('Importance', ascending=False)
        .reset_index(drop=True)
    )

    plt.figure(figsize=(6, 4))
    sns.barplot(x='Importance', y='Index', data=importance_df,  color='skyblue')
    plt.title(f'{satellite_name} - Feature Importance ({model_type.capitalize()})', fontsize=12)
    plt.xlabel('Importance Rate', fontsize=11)
    plt.ylabel('Remote Sensing Index', fontsize=11)
    plt.tight_layout()
    plot_path = f"{save_dir}/{satellite_name}_FeatureImportance_General.png"
    plt.savefig(plot_path, dpi=300)
    plt.show()
    print(f" General feature importance plot saved to: {plot_path}")

    # ===================================
    # Habitat-specific feature importance
    # ===================================
    if by_habitat and habitat_col and habitat_col in df.columns:
        habitat_importances = []

        for habitat, subset in df.groupby(habitat_col):
            if len(subset) < n_splits * 2:
                continue

            X_h, y_h = subset[feature_cols], subset[target_col]

            if model_type == 'classification':
                rf_h = RandomForestClassifier(n_estimators=n_estimators,
                                              random_state=random_state, n_jobs=-1)
                scoring_h = 'accuracy'
            else:
                rf_h = RandomForestRegressor(n_estimators=n_estimators,
                                             random_state=random_state, n_jobs=-1)
                scoring_h = make_scorer(r2_score)

            rf_h.fit(X_h, y_h)
            y_pred_h = rf_h.predict(X_h)

            if model_type == 'regression':
                train_score_h = r2_score(y_h, y_pred_h)
            else:
                train_score_h = accuracy_score(y_h, y_pred_h)

            cv_scores_h = cross_val_score(rf_h, X_h, y_h, cv=n_splits, scoring=scoring_h)
            mean_cv_score_h = np.mean(cv_scores_h)

            imp_h = pd.DataFrame({
                'Habitat': habitat,
                'Index': feature_cols,
                'Importance': rf_h.feature_importances_,
                'Train_Score': train_score_h,
                'CV_Score': mean_cv_score_h
            })
            habitat_importances.append(imp_h)

        habitat_importance_df = pd.concat(habitat_importances, ignore_index=True)

        # Correct plot: use hue='Habitat' and proper dataframe
        plt.figure(figsize=(8, 6))
        sns.barplot(x='Importance', y='Index', hue='Habitat', data=habitat_importance_df, dodge=True)
        plt.title(f'{satellite_name} - Feature Importance by Habitat ({model_type.capitalize()})', fontsize=12)
        plt.xlabel('Importance Rate', fontsize=11)
        plt.ylabel('Remote Sensing Index', fontsize=11)
        plt.tight_layout()
        habitat_plot_path = f"{save_dir}/{satellite_name}_FeatureImportance_ByHabitat.png"
        plt.savefig(habitat_plot_path, dpi=300)
        plt.show()
        print(f" Habitat-specific feature importance plot saved to: {habitat_plot_path}")

        # return habitat_importance_df
        # ===============================
        # Save habitat importance results
        # ===============================
        csv_path = os.path.join(save_dir, f"{satellite_name}_HabitatImportance.csv")
        habitat_importance_df.to_csv(csv_path, index=False)
        print(f" Habitat-specific feature importance table saved to: {csv_path}\n")

        # Print first few rows as preview
        print(habitat_importance_df.head(10))

        return habitat_importance_df
        
    return importance_df
```

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, make_scorer
from sklearn.model_selection import cross_val_score, KFold
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np

def rf_vegetation_analysis(
    df, 
    target_col='F_alive',
    habitat_col=None,
    satellite_name='RS',
    save_dir='Plot_RF_Pred',
    by_habitat=False,
    n_estimators=500,
    random_state=42,
    n_splits=5
):
    """
    Random Forest analysis for vegetation indices predicting F_alive.
    
    Outputs:
    - General feature importance plot
    - Habitat-specific feature importance plot & CSV (optional)
    - Predicted vs Observed scatter plots per season
    - Returns: general importance DF, habitat importance DF (if by_habitat), predictions DF
    """
    if df.empty:
        print("Input dataframe is empty.")
        return None

    df = df.copy()
    os.makedirs(save_dir, exist_ok=True)

    # Select vegetation indices
    feature_cols = [c for c in df.columns if c.upper() in
                    ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']]
    if not feature_cols:
        raise ValueError("No valid vegetation index columns found in dataframe.")

    df = df.dropna(subset=[target_col])
    X = df[feature_cols]
    y = df[target_col]

    # Detect regression vs classification
    if y.nunique() <= 10 and y.dtype in ['object', 'category', 'bool', 'int']:
        model_type = 'classification'
        model = RandomForestClassifier(n_estimators=n_estimators,
                                       random_state=random_state, n_jobs=-1)
        scoring = 'accuracy'
    else:
        model_type = 'regression'
        model = RandomForestRegressor(n_estimators=n_estimators,
                                      random_state=random_state, n_jobs=-1)
        scoring = make_scorer(r2_score)

    # Fit model
    model.fit(X, y)
    y_pred = model.predict(X)

    if model_type == 'regression':
        train_score = r2_score(y, y_pred)
    else:
        train_score = accuracy_score(y, y_pred)

    # 5-fold CV
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    cv_scores = cross_val_score(model, X, y, cv=kf, scoring=scoring, n_jobs=-1)
    mean_cv_score = np.mean(cv_scores)

    print(f"\n{satellite_name} Random Forest ({model_type.capitalize()}):")
    if model_type == 'regression':
        print(f"  - Training R²: {train_score:.3f}")
        print(f"  - Mean 5-fold CV R²: {mean_cv_score:.3f}")
    else:
        print(f"  - Training Accuracy: {train_score:.3f}")
        print(f"  - Mean 5-fold CV Accuracy: {mean_cv_score:.3f}")

    # -------------------------
    # General feature importance
    # -------------------------
    importance_df = pd.DataFrame({
        'Index': feature_cols,
        'Importance': model.feature_importances_
    }).sort_values('Importance', ascending=False).reset_index(drop=True)

    plt.figure(figsize=(6, 4))
    sns.barplot(x='Importance', y='Index', data=importance_df, color='skyblue')
    plt.title(f'{satellite_name} - Feature Importance ({model_type.capitalize()})', fontsize=12)
    plt.xlabel('Importance Rate', fontsize=11)
    plt.ylabel('RS Index', fontsize=11)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/{satellite_name}_FeatureImportance_General.png", dpi=300)
    plt.show()

    # -------------------------
    # Predicted vs Observed plots
    # -------------------------
    if model_type == 'regression' and 'Season' in df.columns:
        plt.figure(figsize=(8, 6))
        sns.scatterplot(
            x=y, y=y_pred, hue=df['Season'], palette='Set2', alpha=0.7
        )
        plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--', lw=2)
        plt.xlabel('Observed F_alive')
        plt.ylabel('Predicted F_alive')
        plt.title(f'{satellite_name} - Predicted vs Observed F_alive')
        plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        plt.tight_layout()
        plt.savefig(f"{save_dir}/{satellite_name}_Predicted_vs_Observed.png", dpi=300)
        plt.show()

    # -------------------------
    # Habitat-specific importance
    # -------------------------
    habitat_importance_df = None
    if by_habitat and habitat_col and habitat_col in df.columns:
        habitat_importances = []
        for habitat, subset in df.groupby(habitat_col):
            if len(subset) < n_splits * 2:
                continue
            X_h, y_h = subset[feature_cols], subset[target_col]
            rf_h = RandomForestRegressor(n_estimators=n_estimators,
                                         random_state=random_state, n_jobs=-1) \
                   if model_type=='regression' else RandomForestClassifier(n_estimators=n_estimators,
                                                                          random_state=random_state, n_jobs=-1)
            rf_h.fit(X_h, y_h)
            y_pred_h = rf_h.predict(X_h)
            train_score_h = r2_score(y_h, y_pred_h) if model_type=='regression' else accuracy_score(y_h, y_pred_h)
            cv_scores_h = cross_val_score(rf_h, X_h, y_h, cv=n_splits, scoring=scoring)
            mean_cv_score_h = np.mean(cv_scores_h)
            imp_h = pd.DataFrame({
                'Habitat': habitat,
                'Index': feature_cols,
                'Importance': rf_h.feature_importances_,
                'Train_Score': train_score_h,
                'CV_Score': mean_cv_score_h
            })
            habitat_importances.append(imp_h)

        habitat_importance_df = pd.concat(habitat_importances, ignore_index=True)

        plt.figure(figsize=(8, 6))
        sns.barplot(x='Importance', y='Index', hue='Habitat', data=habitat_importance_df, dodge=True)
        plt.title(f'{satellite_name} - Feature Importance by Habitat', fontsize=12)
        plt.xlabel('Importance Rate')
        plt.ylabel('RS Index')
        plt.tight_layout()
        plt.savefig(f"{save_dir}/{satellite_name}_FeatureImportance_ByHabitat.png", dpi=300)
        plt.show()

        # Save CSV
        habitat_importance_df.to_csv(f"{save_dir}/{satellite_name}_HabitatImportance.csv", index=False)
        print(f"Habitat-specific importance saved to CSV: {save_dir}/{satellite_name}_HabitatImportance.csv")

    # -------------------------
    # Return results
    # -------------------------
    # Add predictions to dataframe
    df_out = df.copy()
    df_out['Predicted'] = y_pred

    return importance_df, habitat_importance_df, df_out
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def scatter_falive_vs_rf_by_habitat_and_satellite(
    df,
    value_col="Value",
    satellite_col="Satellite",
    index_col="Index",
    field_sat="Field",
    field_index="F_alive",
    rf_index="Predicted_RF",
    habitat_col="Habitat",
    figsize=(6,6),
    dpi=300,
    output_dir="Plot_RF_Falive"
):
    """
    Scatter plot: Observed F_alive (Field) vs RF predicted F_alive,
    faceted by Satellite and colored by Habitat.

    - x-axis: Observed F_alive (Field)
    - y-axis: RF predicted F_alive
    - One subplot per satellite (Landsat, Sentinel-2, MODIS)
    """

    os.makedirs(output_dir, exist_ok=True)

    # # --- Extract observed F_alive ---
    # df_field = df[(df[satellite_col]==field_sat) & (df[index_col]==field_index)]
    # if df_field.empty:
    #     raise ValueError("No observed F_alive (Field) data found.")

    satellites = ["Landsat", "Sentinel-2", "MODIS"]
    habitats = df[habitat_col].dropna().unique()
    palette = sns.color_palette("tab10", len(habitats))
    colors = dict(zip(habitats, palette))

    # --- Create subplots ---
    fig, axes = plt.subplots(
        nrows=1,
        ncols=len(satellites),
        figsize=(figsize[0]*len(satellites), figsize[1]),
        sharey=True
    )
    if len(satellites) == 1:
        axes = [axes]

    for ax, sat in zip(axes, satellites):
        df_rf = df[(df[satellite_col]==sat) & (df[index_col]==rf_index)]
        
        if df_rf.empty:
            ax.set_visible(False)
            continue

        plotted_any = False  # Flag to track if any points are plotted

        # print(f"The dimentions of rf_df for {sat} is {df_rf.size}")
        # print(f"The columns.tolist of rf_df for {sat} is {df_rf.columns.tolist}")
        
        for habitat in habitats:
            # df_h_field = df_field[df_field[habitat_col]==habitat]
            df_h_rf = df_rf[df_rf[habitat_col]==habitat]

            if df_h_rf.empty:
                continue

            # Use the **shortest length** to align observed and predicted
            n = len(df_h_rf)
            x = df_h_rf[field_index].values[:n]
            y = df_h_rf[value_col].values[:n]

            ax.scatter(
                x,
                y,
                alpha=0.7,
                color=colors[habitat],
                label=habitat
            )
            plotted_any = True

        if not plotted_any:
            ax.set_visible(False)
            continue

        # Diagonal 1:1 line
        all_vals = np.concatenate([
            df_rf[field_index].dropna().values,
            df_rf[value_col].dropna().values
        ])
        ax.plot([all_vals.min(), all_vals.max()],
                [all_vals.min(), all_vals.max()],
                'k--', lw=2)

        ax.set_xlabel("Observed F_alive (Field)")
        ax.set_ylabel("RF Predicted F_alive")
        ax.set_title(f"{sat} RF Predictions")
        ax.grid(True, alpha=0.3)
        ax.legend(frameon=False, fontsize=9)

    plt.tight_layout()
    outpath = os.path.join(output_dir, "Scatter_Falive_vs_RF_By_Habitat_And_Satellite.png")
    plt.savefig(outpath, dpi=dpi, bbox_inches="tight")
    plt.show()
    print(f"Saved: {outpath}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def scatter_falive_vs_rf_by_habitat(
    df,
    value_col="Value",
    satellite_col="Satellite",
    index_col="Index",
    field_sat="Field",
    field_index="F_alive",
    rf_index="Predicted_RF",
    habitat_col="Habitat",
    figsize=(6,6),
    dpi=300,
    output_dir="Plot_RF_Falive"
):
    """
    Scatter plot: Observed F_alive (Field) vs RF predicted F_alive (Landsat, Sentinel-2, MODIS)
    Points colored by Habitat.

    - x-axis: Observed F_alive (Field)
    - y-axis: RF predicted F_alive
    """

    os.makedirs(output_dir, exist_ok=True)

    # --- Extract observed F_alive ---
    df_field = df[(df[satellite_col]==field_sat) & (df[index_col]==field_index)]
    if df_field.empty:
        raise ValueError("No observed F_alive (Field) data found.")

    # --- RF predictions ---
    df_rf = df[(df[index_col]==rf_index) & (df[satellite_col] != field_sat)]
    if df_rf.empty:
        raise ValueError("No RF predictions found.")

    # --- Merge observed and RF predictions ---
    # Assuming each record can be matched by Year, Season, Station or another unique key
    # If not, just use RF predictions as y and field as x for scatter
    # Here, we'll just scatter all points together
    plt.figure(figsize=figsize)
    habitats = df_rf[habitat_col].dropna().unique()
    palette = sns.color_palette("tab10", len(habitats))
    colors = dict(zip(habitats, palette))

    for habitat in habitats:
        df_h = df_rf[df_rf[habitat_col]==habitat]
        if df_h.empty:
            continue
        # Find corresponding observed values
        df_h_field = df_field[df_field[habitat_col]==habitat]
        if df_h_field.empty:
            continue
        # Scatter predicted vs observed
        plt.scatter(
            df_h_field[value_col],
            df_h[value_col],
            alpha=0.7,
            color=colors[habitat],
            label=f"{habitat}"
        )

    # Diagonal line
    all_vals = np.concatenate([
        df_field[value_col].dropna().values,
        df_rf[value_col].dropna().values
    ])
    plt.plot([all_vals.min(), all_vals.max()],
             [all_vals.min(), all_vals.max()],
             'k--', lw=2)

    plt.xlabel("Observed F_alive (Field)")
    plt.ylabel("RF Predicted F_alive")
    plt.title("Observed vs RF Predicted F_alive by Habitat")
    plt.legend(frameon=False, fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    # --- Save ---
    outpath = os.path.join(output_dir, "Scatter_Falive_vs_RF_By_Habitat_26.png")
    plt.savefig(outpath, dpi=dpi, bbox_inches="tight")
    plt.show()
    print(f"Saved: {outpath}")
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

def scatter_falive_vs_rf_simple(
    df,
    satellite_name,
    falive_col="F_alive",
    value_col="Value",
    habitat_col="Habitat",
    figsize=(6, 4),
    dpi=300,
    output_dir="Plot_RF_Falive"
):
    """
    Simple scatter plot:
    - x-axis: F_alive (observed)
    - y-axis: RF predicted (Value)
    - color: Habitat
    """

    os.makedirs(output_dir, exist_ok=True)

    # --- basic sanity filter ---
    df = df[[falive_col, value_col, habitat_col]].dropna()
    if df.empty:
        raise ValueError("No valid data after filtering.")

    plt.figure(figsize=figsize)

    sns.scatterplot(
        data=df,
        x=falive_col,
        y=value_col,
        hue=habitat_col,
        palette="tab10",
        alpha=0.7,
        edgecolor=None
    )

    # --- 1:1 line ---
    vmin = min(df[falive_col].min(), df[value_col].min())
    vmax = max(df[falive_col].max(), df[value_col].max())
    plt.plot([vmin, vmax], [vmin, vmax], "k--", lw=1.5)

    plt.xlabel("Observed F_alive")
    plt.ylabel("RF Predicted F_alive")
    plt.title(f"{satellite_name}: Observed vs RF Predicted F_alive")

    plt.legend(
        title="Habitat",
        bbox_to_anchor=(1.05, 1),
        loc="upper left",
        frameon=False
    )

    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    outpath = os.path.join(
        output_dir,
        f"{satellite_name}_Scatter_Falive_vs_RF_ByHabitat.png"
    )
    plt.savefig(outpath, dpi=dpi, bbox_inches="tight")
    plt.show()

    print(f"Saved: {outpath}")
```

```python
# updates on 3/31, add habitat_2_ to the rs_files
import pandas as pd
import os

# Master CSV with updated habitat info
master_csv = "Trophic_survey_data_114_station_habitat.csv"

# List of RS CSV files to update
rs_files = [
    'Landsat_FieldStations_TR_26_17-25.csv',
    'Sentinel2_FieldStations_TR_26_17-25.csv',
    'MODIS_FieldStations_TR_26_17-25.csv'
]

# Output folder for updated CSVs
output_folder = "updated_RS_files"
os.makedirs(output_folder, exist_ok=True)

# Load master habitat data
master_df = pd.read_csv(master_csv)
master_df = master_df[['Station', 'Habitat_f2']]

for file in rs_files:
    # Load RS CSV
    rs_df = pd.read_csv(file)
    
    # Map new habitat values using station as key
    rs_df['Habitat_2_'] = rs_df['Station'].map(master_df.set_index('Station')['Habitat_f2'])
    
    # Save to new CSV
    new_file = os.path.join(output_folder, file)
    rs_df.to_csv(new_file, index=False)
    print(f"Saved updated file: {new_file}")
```

```python
from sklearn.ensemble import RandomForestRegressor

# RF for Landsat
landsat_df = pd.read_csv('Merged_Veg_Landsat_26.csv')  # RS-field integrated
feature_cols = ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']
target_col = 'F_alive'

X_train = landsat_df[feature_cols]
y_train = landsat_df[target_col]

ld_rf_model = RandomForestRegressor(n_estimators=500, random_state=42, n_jobs=-1)
ld_rf_model.fit(X_train, y_train)
```

```python
# RF for Sentinel-2
sentinel2_df = pd.read_csv('Merged_Veg_Sentinel2_26.csv')  # RS-field integrated
feature_cols = ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']
target_col = 'F_alive'

X_train = sentinel2_df[feature_cols]
y_train = sentinel2_df[target_col]

s2_rf_model = RandomForestRegressor(n_estimators=500, random_state=42, n_jobs=-1)
s2_rf_model.fit(X_train, y_train)
```

```python
# RF for MODIS
modis_df = pd.read_csv('Merged_Veg_Modis_26.csv')  # RS-field integrated
feature_cols = ['NDVI','EVI','SAVI','MSAVI','NDMI','NBR','NBR2','NDWI','MNDWI']
target_col = 'F_alive'

X_train = modis_df[feature_cols]
y_train = modis_df[target_col]

md_rf_model = RandomForestRegressor(n_estimators=500, random_state=42, n_jobs=-1)
md_rf_model.fit(X_train, y_train)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_all_habitats_combined_tight(
    df,
    field_col="F_alive",
    output_dir="Plot_RF_Pred",
    figsize=(8, 4),
    dpi=300,
    max_xticks=10  # max number of x-axis labels to show
):
    """
    Dual-axis plot (tight):
    - Left axis: F_alive (global range)
    - Right axis: RS Index (dynamic per habitat)
    - Tighter figure, adjusted fonts, reduced x-label clutter
    """

    os.makedirs(output_dir, exist_ok=True)

    # --- Configure seasons ---
    season_order = ["Spring", "Summer", "Fall", "Winter"]
    df["Season"] = pd.Categorical(df["Season"], categories=season_order, ordered=True)
    habitats = sorted(df["Habitat_f2"].str.strip().unique().tolist())

    # --- F_alive global y-limits ---
    f_vals = df[field_col].dropna().values
    if len(f_vals) == 0:
        raise ValueError(f"No valid values found for {field_col}.")
    
    f_low, f_high = np.nanpercentile(f_vals, [2, 98])
    if f_low == f_high:
        f_low, f_high = np.nanmin(f_vals), np.nanmax(f_vals)
    f_range = f_high - f_low
    if f_range == 0:
        f_range = 1
    f_ylim = (f_low - 0.1*f_range, f_high + 0.1*f_range)

    # --- Satellite colors ---
    satellites = sorted(df["Satellite"].dropna().unique().tolist())
    color_palette = sns.color_palette("tab10", n_colors=max(10, len(satellites)))
    satellite_colors = {sat: color_palette[i % len(color_palette)] for i, sat in enumerate(satellites)}

    for habitat in habitats:
        df_sel = df[df["Habitat_f2"].str.strip() == habitat].copy()
        if df_sel.empty:
            print(f"No data for habitat '{habitat}', skipping.")
            continue

        df_sel = df_sel.sort_values(["Year", "Season"])
        df_sel["Season_Year"] = df_sel["Year"].astype(str) + "-" + df_sel["Season"].astype(str)
        x_axis = df_sel["Season_Year"].unique().tolist()

        # --- F_alive values ---
        df_field = df_sel.groupby("Season_Year")[field_col].mean().reindex(x_axis)

        # --- Satellite-Index combinations ---
        combos = sorted(df_sel[["Satellite", "Index"]].drop_duplicates().values.tolist())

        # # --- RS axis limits ---
        # rs_vals_habitat = df_sel["Value"].dropna().values
        # if len(rs_vals_habitat) > 0:
        #     rs_low, rs_high = np.nanpercentile(rs_vals_habitat, [2, 100])
        #     if rs_low == rs_high:
        #         rs_low -= 0.01 * abs(rs_low) if rs_low != 0 else 0.01
        #         rs_high += 0.01 * abs(rs_high) if rs_high != 0 else 0.01
        #     rs_range = rs_high - rs_low
        #     rs_ylim = (rs_low - 0.01*rs_range, rs_high + 0.01*rs_range)
        # else:
        #     rs_ylim = (0, 1)

        # --- Figure ---
        fig, ax1 = plt.subplots(figsize=figsize)

        # LEFT AXIS: F_alive
        # ax1.plot(x_axis, df_field.values, marker="o", color="black", linewidth=1.5, label=field_col)
        ax1.set_ylabel(field_col, color="black", fontsize=10)
        ax1.tick_params(axis="y", labelcolor="black", labelsize=9)
        ax1.set_ylim(f_ylim)

        # --- x-axis: reduce clutter ---
        # ax1.tick_params(axis="x", rotation=45, labelsize=8)
        n_labels = len(x_axis)
        # if n_labels > max_xticks:
        #     step = max(1, n_labels // max_xticks)
        #     ax1.set_xticks(range(0, n_labels, step))
        #     ax1.set_xticklabels([x_axis[i] for i in range(0, n_labels, step)], rotation=45, ha="right")
        # else:
        #     ax1.set_xticks(range(n_labels))
        #     ax1.set_xticklabels(x_axis, rotation=45, ha="right")

        # ax1.grid(axis="x", linestyle="--", alpha=0.3)

        if n_labels > max_xticks:
            step = max(1, n_labels // max_xticks)
        else:
            step = 1

        tick_positions = np.arange(0, n_labels, step)
        tick_labels = [x_axis[i] for i in tick_positions]

        ax1.set_xticks(tick_positions)
        ax1.set_xticklabels(tick_labels, rotation=30, ha="right", fontsize=8)
        ax1.grid(axis="x", linestyle="--", alpha=0.3)
        
        
        # RIGHT AXIS: RS index
        ax2 = ax1.twinx()
        ax2.set_ylabel("RF predicted F_alive", fontsize=10)
        ax2.tick_params(axis="y", labelsize=9)
        # ax2.set_ylim(rs_ylim)

        for sat, idx in combos:
            df_rs = df_sel[(df_sel["Satellite"] == sat) & (df_sel["Index"] == idx)]
            if df_rs.empty or df_rs["Value"].dropna().empty:
                continue
            df_rs_plot = df_rs.groupby("Season_Year")["Value"].mean().reindex(x_axis)
            ax2.plot(x_axis, df_rs_plot.values, marker="o", linewidth=1.2,
                     color=satellite_colors.get(sat, "grey"), label=f"{sat}-{idx}")

        # Legends
        # ax1.legend(loc="upper left", fontsize=8)
        ax2.legend(loc="upper right", fontsize=8)

        plt.title(habitat, fontsize=12)
        plt.tight_layout()

        # SAVE
        fname = f"{habitat.replace(' ', '_')}_combined_tight.png"
        fpath = os.path.join(output_dir, fname)
        plt.savefig(fpath, dpi=dpi, bbox_inches="tight")
        plt.close()
        print(f"Saved: {fpath}")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def plot_observed_vs_rf_multiples(
    df,
    habitat,
    field_col="F_alive",
    value_col="Value",
    sensors=("Landsat", "Sentinel-2", "MODIS"),
    season_order=("Spring", "Summer", "Fall", "Winter"),
    max_xticks=8,
    figsize=(8, 8),
    dpi=300,
    output_dir="Plot_RF_Pred"
):
    """
    Small multiples (rows = sensors):
    - Observed F_alive: black dashed line (left axis)
    - RF predictions: colored solid lines (right axis)
    """

    os.makedirs(output_dir, exist_ok=True)

    # --- filter habitat ---
    df = df[df["Habitat_f2"].str.strip() == habitat].copy()
    if df.empty:
        raise ValueError(f"No data found for habitat: {habitat}")

    # --- categorical season ---
    df["Season"] = pd.Categorical(df["Season"], categories=season_order, ordered=True)
    df = df.sort_values(["Year", "Season"])
    df["Season_Year"] = df["Year"].astype(str) + "-" + df["Season"].astype(str)

    x_axis = df["Season_Year"].unique()
    n_labels = len(x_axis)

    # --- global F_alive limits ---
    f_vals = df[field_col].dropna().values
    f_low, f_high = np.nanpercentile(f_vals, [2, 98])
    f_range = max(f_high - f_low, 1e-6)
    f_ylim = (f_low - 0.1 * f_range, f_high + 0.1 * f_range)

    # --- colors ---
    index_palette = sns.color_palette("tab10")
    indices = sorted(df["Index"].unique())
    index_colors = dict(zip(indices, index_palette))

    # --- figure ---
    fig, axes = plt.subplots(
        nrows=len(sensors),
        ncols=1,
        figsize=figsize,
        sharex=True
    )

    if len(sensors) == 1:
        axes = [axes]

    for ax, sensor in zip(axes, sensors):

        df_s = df[df["Satellite"] == sensor]
        if df_s.empty:
            ax.set_visible(False)
            continue

        # ===== LEFT AXIS: observed F_alive =====
        ax.plot(
            x_axis,
            df_s.groupby("Season_Year")[field_col].mean().reindex(x_axis),
            color="black",
            linestyle="--",
            linewidth=1.3,
            label="Observed F_alive"
        )
        ax.set_ylim(f_ylim)
        ax.set_ylabel("Observed F_alive", fontsize=10)
        ax.tick_params(axis="y", labelsize=9)

        # ===== RIGHT AXIS: RF predictions =====
        ax2 = ax.twinx()
        ax2.set_ylabel("RF predicted F_alive", fontsize=10)
        ax2.tick_params(axis="y", labelsize=9)

        for idx in indices:
            df_idx = df_s[df_s["Index"] == idx]
            if df_idx.empty:
                continue

            y = df_idx.groupby("Season_Year")[value_col].mean().reindex(x_axis)

            ax2.plot(
                x_axis,
                y,
                color=index_colors[idx],
                linewidth=1.6,
                marker="o",
                label=idx
            )

        ax.set_title(sensor, fontsize=11, loc="left")
        ax.grid(axis="x", linestyle="--", alpha=0.3)

    # --- x ticks (reduced) ---
    step = max(1, n_labels // max_xticks)
    ticks = np.arange(0, n_labels, step)
    axes[-1].set_xticks(ticks)
    axes[-1].set_xticklabels(
        [x_axis[i] for i in ticks],
        rotation=30,
        ha="right",
        fontsize=9
    )

    # --- shared legend ---
    handles_obs = [
        plt.Line2D([0], [0], color="black", linestyle="--", linewidth=1.3)
    ]
    labels_obs = ["Observed F_alive"]

    handles_rf = [
        plt.Line2D([0], [0], color=index_colors[i], linewidth=1.6)
        for i in indices
    ]
    labels_rf = indices

    fig.legend(
        handles_obs + handles_rf,
        labels_obs + labels_rf,
        # ncol=1,
        loc="upper right",
        bbox_to_anchor=(0.915, 0.87),
        frameon=False,
        fontsize=9
    )

    fig.suptitle(
        f"Observed vs RF-predicted F_alive across sensors ({habitat})",
        fontsize=13,
        y=0.92
    )

    plt.tight_layout(rect=[0, 0, 1, 0.95])

    # --- save ---
    fname = f"{habitat.replace(' ', '_')}_Observed_vs_RF_small_multiples.png"
    fpath = os.path.join(output_dir, fname)
    plt.savefig(fpath, dpi=dpi, bbox_inches="tight")
    plt.close()

    print(f"Saved: {fpath}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_falive_vs_rf(df,
                season_order=("Spring", "Summer", "Fall", "Winter"),
                value_col="Value",
                satellite_col="Satellite",
                index_col="Index",
                field_sat="Field",
                field_index="F_alive",
                rf_index="Predicted_RF",
                figsize=(6,4),
                dpi=300,
                output_dir="Plot_RF_Falive"):
    """
    Plot observed F_alive (Field) and RF predicted F_alive (Landsat, Sentinel-2, MODIS)
    on a single axis.
    
    Four lines:
    - Observed F_alive (Field)
    - Predicted_RF (Landsat)
    - Predicted_RF (Sentinel-2)
    - Predicted_RF (MODIS)
    """

    os.makedirs(output_dir, exist_ok=True)

    # --- preprocess seasons ---
    df["Season"] = pd.Categorical(df["Season"], categories=season_order, ordered=True)
    df = df.sort_values(["Year", "Season"])
    df["Season_Year"] = df["Year"].astype(str) + "-" + df["Season"].astype(str)
    x_axis = sorted(df["Season_Year"].unique())

    # --- Figure ---
    fig, ax = plt.subplots(figsize=figsize)

    # ---- Observed F_alive (Field) ----
    df_field = df[(df[satellite_col]==field_sat) & (df[index_col]==field_index)]
    if df_field.empty:
        raise ValueError("No observed F_alive (Field) data found.")
    series_field = df_field.groupby("Season_Year")[value_col].mean().reindex(x_axis)
    ax.plot(x_axis, series_field, color="black", linestyle="--", linewidth=1.5, label="Observed F_alive (Field)")

    # ---- RF predictions ----
    rf_sats = ["Landsat", "Sentinel-2", "MODIS"]
    palette = sns.color_palette("tab10", len(rf_sats))
    colors = dict(zip(rf_sats, palette))

    for sat in rf_sats:
        df_rf = df[(df[satellite_col]==sat) & (df[index_col]==rf_index)]
        if df_rf.empty:
            continue
        series_rf = df_rf.groupby("Season_Year")[value_col].mean().reindex(x_axis)
        ax.plot(x_axis, series_rf, color=colors[sat], linewidth=1.8, marker="o", label=f"{sat} RF predicted")

    # ---- Y limits ----
    all_vals = np.concatenate([
        series_field.dropna().values
    ] + [
        df[(df[satellite_col]==sat) & (df[index_col]==rf_index)][value_col].dropna().values
        for sat in rf_sats
    ])
    f_low, f_high = np.nanpercentile(all_vals, [2, 98])
    pad = 0.05 * (f_high - f_low)
    ax.set_ylim(f_low - pad, f_high + pad)

    # ---- X ticks ----
    max_xticks = 8
    step = max(1, len(x_axis) // max_xticks)
    ax.set_xticks(range(0, len(x_axis), step))
    ax.set_xticklabels([x_axis[i] for i in ax.get_xticks()], rotation=30, ha="right")

    ax.set_xlabel("Season-Year")
    ax.set_ylabel("Fractional alive vegetation cover (F_alive)")
    ax.grid(True, alpha=0.3)
    ax.legend(frameon=False, fontsize=10)
    plt.tight_layout()

    # --- Save ---
    outpath = os.path.join(output_dir, "Observed_vs_RF.png")
    plt.savefig(outpath, dpi=dpi, bbox_inches="tight")
    plt.show()
    print(f"Saved: {outpath}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_falive_by_habitat(df,
                            habitat_col="Habitat_f2",
                            season_order=("Spring", "Summer", "Fall", "Winter"),
                            value_col="Value",
                            satellite_col="Satellite",
                            index_col="Index",
                            field_sat="Field",
                            field_index="F_alive",
                            rf_index="Predicted_RF",
                            figsize=(6,4),
                            dpi=300,
                            output_dir="Plot_RF_Falive"):
    """
    Plot observed F_alive (Field) and RF predicted F_alive (Landsat, Sentinel-2, MODIS)
    for each habitat separately.

    Four lines per plot:
    - Observed F_alive (Field)
    - Predicted_RF (Landsat)
    - Predicted_RF (Sentinel-2)
    - Predicted_RF (MODIS)
    """

    os.makedirs(output_dir, exist_ok=True)

    habitats = df[habitat_col].unique()
    print(f"Found habitats: {habitats}")

    for habitat in habitats:
        df_h = df[df[habitat_col] == habitat].copy()
        if df_h.empty:
            print(f"Skipping habitat '{habitat}': no data")
            continue

        # --- preprocess seasons ---
        df_h["Season"] = pd.Categorical(df_h["Season"], categories=season_order, ordered=True)
        df_h = df_h.sort_values(["Year", "Season"])
        df_h["Season_Year"] = df_h["Year"].astype(str) + "-" + df_h["Season"].astype(str)
        x_axis = sorted(df_h["Season_Year"].unique())

        fig, ax = plt.subplots(figsize=figsize)

        # ---- Observed F_alive (Field) ----
        df_field = df_h[(df_h[satellite_col]==field_sat) & (df_h[index_col]==field_index)]
        if df_field.empty:
            print(f"Skipping habitat '{habitat}': no observed F_alive (Field)")
            continue
        series_field = df_field.groupby("Season_Year")[value_col].mean().reindex(x_axis)
        ax.plot(x_axis, series_field, color="black", linestyle="--", linewidth=1.5, label="Observed F_alive (Field)")

        # ---- RF predictions ----
        rf_sats = ["Landsat", "Sentinel-2", "MODIS"]
        palette = sns.color_palette("tab10", len(rf_sats))
        colors = dict(zip(rf_sats, palette))

        for sat in rf_sats:
            df_rf = df_h[(df_h[satellite_col]==sat) & (df_h[index_col]==rf_index)]
            if df_rf.empty:
                continue
            series_rf = df_rf.groupby("Season_Year")[value_col].mean().reindex(x_axis)
            ax.plot(x_axis, series_rf, color=colors[sat], linewidth=1.8, marker="o", label=f"{sat} RF predicted")

        # ---- Y limits ----
        all_vals = np.concatenate([
            series_field.dropna().values
        ] + [
            df_h[(df_h[satellite_col]==sat) & (df_h[index_col]==rf_index)][value_col].dropna().values
            for sat in rf_sats
        ])
        f_low, f_high = np.nanpercentile(all_vals, [2, 98])
        pad = 0.05 * (f_high - f_low)
        # ax.set_ylim(f_low - pad, f_high + pad)
        ax.set_ylim(0,0.16)

        # ---- X ticks ----
        max_xticks = 8
        step = max(1, len(x_axis) // max_xticks)
        ax.set_xticks(range(0, len(x_axis), step))
        ax.set_xticklabels([x_axis[i] for i in ax.get_xticks()], rotation=30, ha="right")

        ax.set_xlabel("Season-Year")
        ax.set_ylabel("Fractional alive vegetation cover (F_alive)")
        ax.set_title(f"Habitat: {habitat}", fontsize=13)
        ax.grid(True, alpha=0.3)
        ax.legend(frameon=False, fontsize=10)
        plt.tight_layout()

        # --- Save ---
        outpath = os.path.join(output_dir, f"{habitat.replace(' ', '_')}_Observed_vs_RF.png")
        plt.savefig(outpath, dpi=dpi, bbox_inches="tight")
        plt.close()
        print(f"Saved: {outpath}")
```

### 4.2.2 Long-term trends: **Option A** Use all indices to predict vegetation cover and then extract trends

Pros:

Captures all aspects of vegetation dynamics, including greenness, moisture, and disturbance.

Random Forest inherently handles correlated predictors, so including all indices usually won’t “break” the model.

Predicted trends will reflect a holistic combination of indices, closer to what the model learned from field data.

Cons:

Interpretation of which index drives the trend becomes less clear.

May slightly dilute the effect of the most important indices, though generally minor if RF properly weighted them.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from pymannkendall import original_test as mk_test

def plot_predicted_vegetation_trends(
    satellite_map,
    smooth_windows=[16,5,8],  # rolling windows for Landsat, Sentinel-2, MODIS
    frac=0.1,
    outdir='Plot_RF_Pred_Trend',
    csv_outfile='F_alive_trend_summary.csv'
):
    """
    Plot temporal trends of predicted vegetation cover (F_alive_Pred) for habitats and satellites.
    Computes LOWESS smoothing, rolling mean, Mann-Kendall test, slope, and pseudo-R².
    Also creates a summary table and saves as CSV.

    Args:
        satellite_map (dict): {'Landsat': df1, 'Sentinel-2': df2, 'MODIS': df3}, 
                              each df must contain ['date','Habitat','F_alive_Pred']
        smooth_windows (list): rolling window days for [Landsat, Sentinel-2, MODIS]
        frac (float): LOWESS fraction
        outdir (str): directory to save plots
        csv_outfile (str): path for summary CSV
    """
    os.makedirs(outdir, exist_ok=True)
    summary_list = []

    sat_names = list(satellite_map.keys())
    sat_window_map = dict(zip(sat_names, smooth_windows))

    for sat, df in satellite_map.items():
        print(f"\nProcessing satellite: {sat}")
        if 'F_alive_Pred' not in df.columns:
            print(f"{sat} missing 'F_alive_Pred', skipping.")
            continue

        df = df.copy()
        df['date'] = pd.to_datetime(df['date'], errors='coerce')
        df = df.dropna(subset=['date', 'F_alive_Pred'])
        
        habitats = df['Habitat_2_'].dropna().unique()
        for habitat in habitats:
            df_hab = df[df['Habitat_2_'] == habitat].copy()
            if df_hab.empty:
                print(f"No data for {habitat} in {sat}, skipping.")
                continue

            df_hab = df_hab.sort_values('date')

            # Outlier removal (IQR)
            Q1, Q3 = df_hab['F_alive_Pred'].quantile([0.25, 0.75])
            IQR = Q3 - Q1
            lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
            df_hab = df_hab[(df_hab['F_alive_Pred'] >= lower) & (df_hab['F_alive_Pred'] <= upper)]

            if df_hab.empty:
                print(f"All values removed as outliers for {habitat} in {sat}")
                continue

            # Rolling mean
            smooth_window = sat_window_map.get(sat, 8)
            df_hab['smoothed'] = df_hab['F_alive_Pred'].rolling(
                window=smooth_window, min_periods=1, center=True
            ).mean()

            # LOWESS trend
            x_days = (df_hab['date'] - df_hab['date'].min()).dt.days.values
            df_hab['lowess_trend'] = lowess(
                df_hab['F_alive_Pred'], x_days, frac=frac, return_sorted=False
            )

            # Metrics
            slope = ((df_hab['lowess_trend'].iloc[-1] - df_hab['lowess_trend'].iloc[0]) /
                     (x_days[-1] - x_days[0])) * 365
            pseudo_r2 = np.corrcoef(df_hab['F_alive_Pred'], df_hab['lowess_trend'])[0, 1] ** 2

            try:
                mk_result = mk_test(df_hab['F_alive_Pred'].values)
                mk_tau, mk_p, mk_trend = mk_result.Tau, mk_result.p, mk_result.trend
            except Exception:
                mk_tau, mk_p, mk_trend = np.nan, np.nan, 'No Trend'

            # Append summary
            summary_list.append({
                'Habitat': habitat,
                'Satellite': sat,
                'Slope_per_year': slope,
                'Pseudo_R2': pseudo_r2,
                'MK_tau': mk_tau,
                'MK_p': mk_p,
                'MK_trend': mk_trend,
                'N_points': len(df_hab)
            })

            # Plot
            plt.figure(figsize=(12,5))
            plt.scatter(df_hab['date'], df_hab['F_alive_Pred'], color='gray', s=12, alpha=0.5, label='Predicted')
            plt.plot(df_hab['date'], df_hab['smoothed'], color='black', lw=2, label=f'{smooth_window}-day mean')
            plt.plot(df_hab['date'], df_hab['lowess_trend'], color='steelblue', ls='--', lw=2.2, label=f'LOWESS (frac={frac})')
            plt.title(f"{habitat} - {sat} Predicted Vegetation", fontsize=14)
            plt.xlabel('Date')
            plt.ylabel('F_alive_Pred')
            plt.grid(alpha=0.3, linestyle='--')
            plt.legend(fontsize=9, loc='lower right')
            plt.text(0.02, 0.85,
                     f"Slope ≈ {slope:.4f}/yr\nPseudo-R² = {pseudo_r2:.2f}\nMK τ = {mk_tau:.2f}, p={mk_p:.3f}\nTrend: {mk_trend}",
                     transform=plt.gca().transAxes, fontsize=9, color='red',
                     bbox=dict(facecolor='white', alpha=0.6, edgecolor='none'))
            outfile = os.path.join(outdir, f"{habitat}_{sat}_F_alive_trend.png")
            plt.tight_layout()
            plt.savefig(outfile, dpi=300, bbox_inches='tight')
            plt.close()
            print(f"Saved: {outfile}")

    # Save summary CSV
    summary_df = pd.DataFrame(summary_list)
    summary_csv_path = os.path.join(outdir, csv_outfile)
    summary_df.to_csv(summary_csv_path, index=False)
    print(f"\nSummary table saved to: {summary_csv_path}")
    return summary_df
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from pymannkendall import original_test as mk_test


def plot_monthly_predicted_trends(
    satellite_map,
    value_col='F_alive_Pred',
    frac=0.3,
    outdir='Plot_RF_Pred_Trend',
    csv_outfile='Monthly_trend_summary.csv'
):
    """
    Plot long-term trends using MONTHLY aggregated predicted vegetation cover.
    Applies LOWESS smoothing and Mann–Kendall trend test on aggregated series.

    Args:
        satellite_map (dict): {'Landsat': df1, 'Sentinel-2': df2, 'MODIS': df3}
        value_col (str): column containing predicted vegetation cover
        frac (float): LOWESS smoothing fraction
        outdir (str): output directory
        csv_outfile (str): summary CSV filename
    """

    os.makedirs(outdir, exist_ok=True)
    summary = []

    for sat, df in satellite_map.items():
        print(f"\nProcessing {sat}")

        df = df.copy()
        df['date'] = pd.to_datetime(df['date'], errors='coerce')
        df = df.dropna(subset=['date', value_col, 'Habitat_2_'])

        for habitat in df['Habitat_2_'].unique():

            df_h = df[df['Habitat_2_'] == habitat].copy()
            if df_h.empty:
                continue

            # ---- Monthly aggregation ----
            df_h['month'] = df_h['date'].dt.to_period('M')
            df_m = (
                df_h
                .groupby('month')[value_col]
                .mean()
                .reset_index()
            )
            df_m['date'] = df_m['month'].dt.to_timestamp()

            if len(df_m) < 12:
                print(f"  Skipping {habitat} ({sat}): insufficient monthly data")
                continue

            # ---- LOWESS ----
            x_days = (df_m['date'] - df_m['date'].min()).dt.days.values
            df_m['lowess'] = lowess(
                df_m[value_col],
                x_days,
                frac=frac,
                return_sorted=False
            )

            # ---- Trend metrics ----
            slope = (
                (df_m['lowess'].iloc[-1] - df_m['lowess'].iloc[0]) /
                (x_days[-1] - x_days[0])
            ) * 365

            r2 = np.corrcoef(df_m[value_col], df_m['lowess'])[0, 1] ** 2

            try:
                mk = mk_test(df_m[value_col].values)
                mk_tau, mk_p, mk_trend = mk.Tau, mk.p, mk.trend
            except Exception:
                mk_tau, mk_p, mk_trend = np.nan, np.nan, 'No Trend'

            summary.append({
                'Habitat': habitat,
                'Satellite': sat,
                'Slope_per_year': slope,
                'Pseudo_R2': r2,
                'MK_tau': mk_tau,
                'MK_p': mk_p,
                'MK_trend': mk_trend,
                'N_months': len(df_m)
            })

            # ---- Plot ----
            plt.figure(figsize=(7, 3))
            plt.scatter(
                df_m['date'], df_m[value_col],
                s=35, color='gray', alpha=0.7, label='Monthly mean'
            )
            plt.plot(
                df_m['date'], df_m['lowess'],
                lw=2.5, ls='--', color='steelblue',
                label=f'LOWESS (frac={frac})'
            )

            plt.title(f"{habitat} – {sat}", fontsize=14)
            plt.xlabel('Date')
            plt.ylabel('Predicted vegetation cover')
            plt.grid(alpha=0.3, linestyle='--')
            plt.legend(loc='upper left', bbox_to_anchor=(0.65, 0.95))

            plt.text(
                0.02, 0.95,
                f"Slope ≈ {slope:.4f} / yr\n"
                f"Pseudo-R² = {r2:.2f}\n"
                f"MK τ = {mk_tau:.2f}, p = {mk_p:.3f}\n"
                f"Trend: {mk_trend}",
                transform=plt.gca().transAxes,
                fontsize=9,
                va='top',
                bbox=dict(facecolor='white', alpha=0.7, edgecolor='none')
            )

            outfile = os.path.join(
                outdir, f"{habitat}_{sat}_monthly_trend.png"
            )
            plt.tight_layout()
            plt.savefig(outfile, dpi=300)
            plt.close()

            print(f"  Saved: {outfile}")

    # ---- Save summary table ----
    summary_df = pd.DataFrame(summary)
    summary_path = os.path.join(outdir, csv_outfile)
    summary_df.to_csv(summary_path, index=False)

    print(f"\nSummary table saved to: {summary_path}")
    return summary_df
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from pymannkendall import original_test as mk_test


def plot_monthly_trends_by_satellite(
    satellite_map,
    value_col='F_alive_Pred',
    frac=0.3,
    outdir='plots_monthly_trends',
    csv_outfile='Monthly_trend_summary.csv'
):
    """
    Plot MONTHLY aggregated LOWESS trends for all habitats in ONE figure per satellite.
    Computes slope, pseudo-R², and Mann–Kendall trend for each habitat.

    Args:
        satellite_map (dict): {'Landsat': df1, 'Sentinel-2': df2, 'MODIS': df3}
        value_col (str): predicted vegetation column
        frac (float): LOWESS fraction
        outdir (str): output directory
        csv_outfile (str): summary CSV filename
    """

    os.makedirs(outdir, exist_ok=True)
    summary = []

    for sat, df in satellite_map.items():
        print(f"\nProcessing {sat}")

        df = df.copy()
        df['date'] = pd.to_datetime(df['date'], errors='coerce')
        df = df.dropna(subset=['date', value_col, 'Habitat_2_'])

        habitats = sorted(df['Habitat_2_'].unique())
        cmap = plt.get_cmap('tab10')
        colors = dict(zip(habitats, cmap.colors[:len(habitats)]))

        plt.figure(figsize=(14, 6))
        ax = plt.gca()

        y_min, y_max = np.inf, -np.inf
        text_y = 0.95

        for i, habitat in enumerate(habitats):
            df_h = df[df['Habitat_2_'] == habitat].copy()
            if df_h.empty:
                continue

            # ---- Monthly aggregation ----
            df_h['month'] = df_h['date'].dt.to_period('M')
            df_m = (
                df_h
                .groupby('month')[value_col]
                .mean()
                .reset_index()
            )
            df_m['date'] = df_m['month'].dt.to_timestamp()

            if len(df_m) < 12:
                continue

            # ---- LOWESS ----
            x_days = (df_m['date'] - df_m['date'].min()).dt.days.values
            df_m['lowess'] = lowess(
                df_m[value_col],
                x_days,
                frac=frac,
                return_sorted=False
            )

            # ---- Metrics ----
            slope = (
                (df_m['lowess'].iloc[-1] - df_m['lowess'].iloc[0]) /
                (x_days[-1] - x_days[0])
            ) * 365

            r2 = np.corrcoef(df_m[value_col], df_m['lowess'])[0, 1] ** 2

            try:
                mk = mk_test(df_m[value_col].values)
                mk_tau, mk_p, mk_trend = mk.Tau, mk.p, mk.trend
            except Exception:
                mk_tau, mk_p, mk_trend = np.nan, np.nan, 'No Trend'

            summary.append({
                'Habitat': habitat,
                'Satellite': sat,
                'Slope_per_year': slope,
                'Pseudo_R2': r2,
                'MK_tau': mk_tau,
                'MK_p': mk_p,
                'MK_trend': mk_trend,
                'N_months': len(df_m)
            })

            # ---- Plot ----
            ax.plot(
                df_m['date'],
                df_m['lowess'],
                lw=2.5,
                color=colors[habitat],
                label=habitat
            )

            y_min = min(y_min, df_m['lowess'].min())
            y_max = max(y_max, df_m['lowess'].max())

            # ---- Text metrics (stacked, fixed spacing) ----
            ax.text(
                0.01,
                text_y,
                f"{habitat}: slope={slope:+.4f}/yr, τ={mk_tau:.2f}, p={mk_p:.3f}",
                transform=ax.transAxes,
                fontsize=9,
                color=colors[habitat],
                va='top'
            )
            text_y -= 0.055

        # ---- Figure formatting ----
        ax.set_title(
            f"{sat} – Monthly Predicted Vegetation Trends by Habitat",
            fontsize=15
        )
        ax.set_xlabel('Date')
        ax.set_ylabel('Predicted vegetation cover')
        ax.grid(alpha=0.3, linestyle='--')

        ax.set_ylim(
            y_min - 0.05 * abs(y_min),
            y_max + 0.05 * abs(y_max)
        )

        # ---- Legend outside ----
        ax.legend(
            title='Habitat',
            loc='center left',
            bbox_to_anchor=(1.02, 0.5),
            frameon=False
        )

        plt.tight_layout(rect=[0, 0, 0.82, 1])

        outfile = os.path.join(
            outdir, f"{sat}_monthly_trends_all_habitats.png"
        )
        plt.savefig(outfile, dpi=300, bbox_inches='tight')
        plt.close()

        print(f"Saved: {outfile}")

    # ---- Save summary table ----
    summary_df = pd.DataFrame(summary)
    summary_path = os.path.join(outdir, csv_outfile)
    summary_df.to_csv(summary_path, index=False)
    print(f"\nSummary table saved to: {summary_path}")

    return summary_df
```

### [!Discarded!] 4.2.3 Long-term trends: **Option B** Use the top ranked indices to predict the vegetation cover trends
Use only the top-ranked indices (from RF importance) to predict trends

Pros:

Focuses on the most informative signals, reducing noise from weaker or redundant indices.

Easier to interpret trends: you know which indices are driving predictions.

Can reduce multicollinearity if indices are correlated.

Cons:

May lose complementary information captured by other indices (e.g., moisture or disturbance indices not top-ranked but still relevant).

Trends may be slightly biased toward what the top indices measure (e.g., mostly greenness).

Seasonal or habitat-specific patterns captured by lower-ranked indices might be ignored.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from pymannkendall import original_test as mk_test

def plot_selected_rs_trends(
    satellite_map,
    selected_indices_dict,
    smooth_windows=[16,5,8],  # rolling windows for Landsat, Sentinel-2, MODIS
    frac=0.1,
    outdir='plots_trends_v2'
):
    """
    Plot temporal patterns for habitats using the selected RS index per satellite.
    Applies LOWESS smoothing, rolling mean, Mann-Kendall test, slope, and pseudo-R².

    Args:
        satellite_map (dict): {'Landsat': df1, 'Sentinel-2': df2, 'MODIS': df3}
        selected_indices_dict (dict): {'Habitat': {'Landsat':'NDMI', ...}, ...}
        smooth_windows (list): rolling window days for [Landsat, Sentinel-2, MODIS]
        frac (float): LOWESS fraction
        outdir (str): directory to save plots
    """
    os.makedirs(outdir, exist_ok=True)
    
    sat_names = list(satellite_map.keys())
    sat_window_map = dict(zip(sat_names, smooth_windows))
    
    for habitat, sat_index_dict in selected_indices_dict.items():
        print(f"\nProcessing habitat: {habitat}")
        for sat, idx in sat_index_dict.items():
            if sat not in satellite_map:
                print(f"Satellite {sat} data not found, skipping.")
                continue

            rs_df = satellite_map[sat].copy()
            rs_df['date'] = pd.to_datetime(rs_df['date'], errors='coerce')
            rs_df = rs_df.dropna(subset=['date'])
            
            # Filter by habitat
            df_sel = rs_df.loc[rs_df['Habitat'] == habitat].copy()
            if df_sel.empty or idx not in df_sel.columns:
                print(f"No data for {sat}-{idx} in {habitat}, skipping.")
                continue

            # Aggregate by date
            df_avg = df_sel.groupby('date')[idx].mean().reset_index()
            df_avg = df_avg.replace([np.inf, -np.inf], np.nan).dropna(subset=[idx])
            if df_avg.empty or df_avg[idx].nunique() < 4:
                print(f"Insufficient valid data for {sat}-{idx} in {habitat}, skipping.")
                continue

            # Outlier removal (IQR method)
            Q1, Q3 = df_avg[idx].quantile([0.25, 0.75])
            IQR = Q3 - Q1
            lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
            df_clean = df_avg[(df_avg[idx] >= lower) & (df_avg[idx] <= upper)].copy()
            if df_clean.empty:
                print(f"All values removed as outliers for {sat}-{idx} in {habitat}")
                continue

            df_clean = df_clean.sort_values('date')
            
            # Rolling mean
            smooth_window = sat_window_map.get(sat, 8)
            df_clean['smoothed'] = df_clean[idx].rolling(
                window=smooth_window, min_periods=1, center=True
            ).mean()

            # LOWESS trend
            x_days = (df_clean['date'] - df_clean['date'].min()).dt.days.values
            df_clean['lowess_trend'] = lowess(
                df_clean[idx], x_days, frac=frac, return_sorted=False
            )

            # Metrics
            slope = ((df_clean['lowess_trend'].iloc[-1] - df_clean['lowess_trend'].iloc[0]) /
                     (x_days[-1] - x_days[0])) * 365
            pseudo_r2 = np.corrcoef(df_clean[idx], df_clean['lowess_trend'])[0, 1] ** 2

            # Mann-Kendall test (catch constant series)
            try:
                mk_result = mk_test(df_clean[idx].values)
                mk_tau, mk_p, mk_trend = mk_result.Tau, mk_result.p, mk_result.trend
            except Exception as e:
                mk_tau, mk_p, mk_trend = np.nan, np.nan, 'No Trend'
            
            # Plot
            plt.figure(figsize=(12,5))
            plt.scatter(df_clean['date'], df_clean[idx], color='gray', s=12, alpha=0.5, label='Raw data')
            plt.plot(df_clean['date'], df_clean['smoothed'], color='black', lw=2, label=f'{smooth_window}-day mean')
            plt.plot(df_clean['date'], df_clean['lowess_trend'], color='steelblue', ls='--', lw=2.2, label=f'LOWESS (frac={frac})')
            plt.title(f"{habitat} - {sat}-{idx}", fontsize=14)
            plt.xlabel('Date')
            plt.ylabel(idx)
            plt.grid(alpha=0.3, linestyle='--')
            plt.legend(fontsize=9, loc='lower right')

            # Metrics text
            plt.text(0.02, 0.85,
                     f"Slope ≈ {slope:.4f}/yr\nPseudo-R² = {pseudo_r2:.2f}\nMK τ = {mk_tau:.2f}, p={mk_p:.3f}\nTrend: {mk_trend}",
                     transform=plt.gca().transAxes, fontsize=9, color='red',
                     bbox=dict(facecolor='white', alpha=0.6, edgecolor='none'))

            # Save figure
            outfile = os.path.join(outdir, f"{habitat}_{sat}_{idx}_trend.png")
            plt.tight_layout()
            plt.savefig(outfile, dpi=300, bbox_inches='tight')
            plt.close()
            print(f"Saved: {outfile}")
```

```python
import pandas as pd

# Load your summary table
trend_df = pd.read_csv('plots_trends_v2/RS_trend_summary.csv') 

# Define index types
greenness = ['NDVI','EVI','SAVI','MSAVI']
burn = ['NBR','NBR2']
moisture = ['NDMI','NDWI','MNDWI']

def standardize_slope(row):
    """Make slope positive if it indicates better vegetation"""
    if row['Index'] in greenness:
        return row['Slope_per_year']
    elif row['Index'] in burn:
        return row['Slope_per_year']  # NBR positive = recovering vegetation, leave as is
    elif row['Index'] in moisture:
        return row['Slope_per_year']  # NDMI, NDWI: higher = wetter/healthier vegetation, leave as is
    else:
        return row['Slope_per_year']

# Apply standardization
trend_df['Slope_standardized'] = trend_df.apply(standardize_slope, axis=1)

# Add index type column
def index_type(idx):
    if idx in greenness:
        return 'Greenness'
    elif idx in burn:
        return 'Burn/Disturbance'
    elif idx in moisture:
        return 'Moisture/Water'
    else:
        return 'Other'

trend_df['Index_Type'] = trend_df['Index'].apply(index_type)

# Optional: reorder columns for clarity
cols = ['Habitat','Satellite','Index','Index_Type','Slope_per_year','Slope_standardized',
        'Pseudo_R2','MK_tau','MK_p','MK_trend','N_points']
trend_df = trend_df[cols]

# Save standardized trend table
out_csv = 'plots_trends_v2/trend_summary_standardized.csv'
trend_df.to_csv(out_csv, index=False)
print(f"Standardized trend summary saved to: {out_csv}")

# Quick preview
print(trend_df.head(10))
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from pymannkendall import original_test as mk_test

def plot_selected_rs_trends(
    satellite_map,
    selected_indices_dict,
    smooth_windows=[16,5,8],  # rolling windows for Landsat, Sentinel-2, MODIS
    frac=0.1,
    outdir='plots_trends_v2'
):
    """
    Plot temporal patterns for habitats using the selected RS index per satellite.
    Applies LOWESS smoothing, rolling mean, Mann-Kendall test, slope, and pseudo-R².

    Args:
        satellite_map (dict): {'Landsat': df1, 'Sentinel-2': df2, 'MODIS': df3}
        selected_indices_dict (dict): {'Habitat': {'Landsat':'NDMI', ...}, ...}
        smooth_windows (list): rolling window days for [Landsat, Sentinel-2, MODIS]
        frac (float): LOWESS fraction
        outdir (str): directory to save plots
    """
    os.makedirs(outdir, exist_ok=True)
    
    sat_names = list(satellite_map.keys())
    sat_window_map = dict(zip(sat_names, smooth_windows))
    
    for habitat, sat_index_dict in selected_indices_dict.items():
        print(f"\nProcessing habitat: {habitat}")
        for sat, idx in sat_index_dict.items():
            if sat not in satellite_map:
                print(f"Satellite {sat} data not found, skipping.")
                continue

            rs_df = satellite_map[sat].copy()
            rs_df['date'] = pd.to_datetime(rs_df['date'], errors='coerce')
            rs_df = rs_df.dropna(subset=['date'])
            
            # Filter by habitat
            df_sel = rs_df.loc[rs_df['Habitat'] == habitat].copy()
            if df_sel.empty or idx not in df_sel.columns:
                print(f"No data for {sat}-{idx} in {habitat}, skipping.")
                continue

            # Aggregate by date
            df_avg = df_sel.groupby('date')[idx].mean().reset_index()
            df_avg = df_avg.replace([np.inf, -np.inf], np.nan).dropna(subset=[idx])
            if df_avg.empty or df_avg[idx].nunique() < 4:
                print(f"Insufficient valid data for {sat}-{idx} in {habitat}, skipping.")
                continue

            # Outlier removal (IQR method)
            Q1, Q3 = df_avg[idx].quantile([0.25, 0.75])
            IQR = Q3 - Q1
            lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
            df_clean = df_avg[(df_avg[idx] >= lower) & (df_avg[idx] <= upper)].copy()
            if df_clean.empty:
                print(f"All values removed as outliers for {sat}-{idx} in {habitat}")
                continue

            df_clean = df_clean.sort_values('date')
            
            # Rolling mean
            smooth_window = sat_window_map.get(sat, 8)
            df_clean['smoothed'] = df_clean[idx].rolling(
                window=smooth_window, min_periods=1, center=True
            ).mean()

            # LOWESS trend
            x_days = (df_clean['date'] - df_clean['date'].min()).dt.days.values
            df_clean['lowess_trend'] = lowess(
                df_clean[idx], x_days, frac=frac, return_sorted=False
            )

            # Metrics
            slope = ((df_clean['lowess_trend'].iloc[-1] - df_clean['lowess_trend'].iloc[0]) /
                     (x_days[-1] - x_days[0])) * 365
            pseudo_r2 = np.corrcoef(df_clean[idx], df_clean['lowess_trend'])[0, 1] ** 2

            # Mann-Kendall test (catch constant series)
            try:
                mk_result = mk_test(df_clean[idx].values)
                mk_tau, mk_p, mk_trend = mk_result.Tau, mk_result.p, mk_result.trend
            except Exception as e:
                mk_tau, mk_p, mk_trend = np.nan, np.nan, 'No Trend'
            
            # Plot
            plt.figure(figsize=(12,5))
            plt.scatter(df_clean['date'], df_clean[idx], color='gray', s=12, alpha=0.5, label='Raw data')
            plt.plot(df_clean['date'], df_clean['smoothed'], color='black', lw=2, label=f'{smooth_window}-day mean')
            plt.plot(df_clean['date'], df_clean['lowess_trend'], color='steelblue', ls='--', lw=2.2, label=f'LOWESS (frac={frac})')
            plt.title(f"{habitat} - {sat}-{idx}", fontsize=14)
            plt.xlabel('Date')
            plt.ylabel(idx)
            plt.grid(alpha=0.3, linestyle='--')
            plt.legend(fontsize=9, loc='lower right')

            # Metrics text
            plt.text(0.02, 0.85,
                     f"Slope ≈ {slope:.4f}/yr\nPseudo-R² = {pseudo_r2:.2f}\nMK τ = {mk_tau:.2f}, p={mk_p:.3f}\nTrend: {mk_trend}",
                     transform=plt.gca().transAxes, fontsize=9, color='red',
                     bbox=dict(facecolor='white', alpha=0.6, edgecolor='none'))

            # Save figure
            outfile = os.path.join(outdir, f"{habitat}_{sat}_{idx}_trend.png")
            plt.tight_layout()
            plt.savefig(outfile, dpi=300, bbox_inches='tight')
            plt.close()
            print(f"Saved: {outfile}")
```

```python
import pandas as pd

# Load your summary table
trend_df = pd.read_csv('plots_trends_v2/RS_trend_summary.csv') 

# Define index types
greenness = ['NDVI','EVI','SAVI','MSAVI']
burn = ['NBR','NBR2']
moisture = ['NDMI','NDWI','MNDWI']

def standardize_slope(row):
    """Make slope positive if it indicates better vegetation"""
    if row['Index'] in greenness:
        return row['Slope_per_year']
    elif row['Index'] in burn:
        return row['Slope_per_year']  # NBR positive = recovering vegetation, leave as is
    elif row['Index'] in moisture:
        return row['Slope_per_year']  # NDMI, NDWI: higher = wetter/healthier vegetation, leave as is
    else:
        return row['Slope_per_year']

# Apply standardization
trend_df['Slope_standardized'] = trend_df.apply(standardize_slope, axis=1)

# Add index type column
def index_type(idx):
    if idx in greenness:
        return 'Greenness'
    elif idx in burn:
        return 'Burn/Disturbance'
    elif idx in moisture:
        return 'Moisture/Water'
    else:
        return 'Other'

trend_df['Index_Type'] = trend_df['Index'].apply(index_type)

# Optional: reorder columns for clarity
cols = ['Habitat','Satellite','Index','Index_Type','Slope_per_year','Slope_standardized',
        'Pseudo_R2','MK_tau','MK_p','MK_trend','N_points']
trend_df = trend_df[cols]

# Save standardized trend table
out_csv = 'plots_trends_v2/trend_summary_standardized.csv'
trend_df.to_csv(out_csv, index=False)
print(f"Standardized trend summary saved to: {out_csv}")

# Quick preview
print(trend_df.head(10))
```

### 5.3 Anomaly detection

#### e. Full pipeline script
- Temp anomaly = monthly temp − long-term (2010-2024) mean of that month
- Rainfall anomaly = monthly rainfall − long-term (2010-2024) mean of that month

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# Aggregate climate monthly by Habitat
def aggregate_climate_monthly(df):

    df = df.copy()
    df['date'] = pd.to_datetime(df['date'])

    df_monthly = (
        df.groupby(['Habitat', pd.Grouper(key='date', freq='MS')])
        .agg(
            temp_mean=('temp_C', 'mean'),
            precip_total=('precip_mm', 'sum')
        )
        .reset_index()
    )

    df_monthly['month'] = df_monthly['date'].dt.month

    return df_monthly

# Compute monthly climatology baseline
def compute_climatology(df_monthly):

    climatology = (
        df_monthly.groupby(['Habitat', 'month'])
        .agg(
            temp_clim=('temp_mean', 'mean'),
            precip_clim=('precip_total', 'mean')
        )
        .reset_index()
    )

    return climatology

# Compute anomalies
def compute_anomalies(df_monthly, climatology):

    df = df_monthly.merge(
        climatology,
        on=['Habitat', 'month'],
        how='left'
    )

    df['temp_anomaly'] = df['temp_mean'] - df['temp_clim']
    df['precip_anomaly'] = df['precip_total'] - df['precip_clim']

    return df

# # Prepare vegetation monthly
# def aggregate_veg_monthly(df):

#     df = df.copy()
#     df['date'] = pd.to_datetime(df['date'])

#     df_monthly = (
#         df.groupby(['Habitat', 'Satellite', pd.Grouper(key='date', freq='MS')])
#         .agg(veg_pred=('veg_pred', 'mean'))
#         .reset_index()
#     )

#     return df_monthly

def aggregate_veg_monthly(df, value_col='F_alive_Pred'):
    """
    Aggregate RS vegetation predictions to monthly means.

    Returns dataframe ready for climate merge.
    """

    df = df.copy()

    # -------------------
    # Clean dates
    # -------------------
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df = df.dropna(subset=['date', value_col, 'Habitat'])

    # -------------------
    # Create monthly period (BEST for merging)
    # -------------------
    df['month_period'] = df['date'].dt.to_period('M')

    # -------------------
    # Aggregate monthly
    # -------------------
    group_cols = ['Habitat', 'month_period']

    # keep Satellite if exists
    if 'Satellite' in df.columns:
        group_cols.append('Satellite')

    df_monthly = (
        df.groupby(group_cols)[value_col]
        .mean()
        .reset_index()
        .rename(columns={value_col: 'veg_pred'})
    )

    # Convert period back to timestamp (month start)
    df_monthly['date'] = df_monthly['month_period'].dt.to_timestamp()

    return df_monthly

# Merge climate + vegetation
def merge_climate_veg(clim_df, veg_df):

    merged = veg_df.merge(
        clim_df,
        on=['Habitat', 'date'],
        how='inner'
    )

    return merged

# Lag correlation function
from scipy.stats import pearsonr
import pandas as pd

def compare_climate_vs_anomaly(df, habitat, satellite, max_lag=6):

    df_h = df[
        (df['Habitat'] == habitat) &
        (df['Satellite'] == satellite)
    ].copy()

    if len(df_h) < 12:
        return None

    df_h = df_h.sort_values('date').reset_index(drop=True)

    results = []

    # -----------------------
    # Loop lags
    # -----------------------
    for lag in range(max_lag + 1):

        df_h[f'temp_lag{lag}'] = df_h['temp_mean'].shift(lag)
        df_h[f'precip_lag{lag}'] = df_h['precip_total'].shift(lag)
        df_h[f'temp_anom_lag{lag}'] = df_h['temp_anomaly'].shift(lag)
        df_h[f'precip_anom_lag{lag}'] = df_h['precip_anomaly'].shift(lag)

        df_lag = df_h.dropna()

        if len(df_lag) < 10:
            continue

        r_temp, _ = pearsonr(df_lag['veg_pred'], df_lag[f'temp_lag{lag}'])
        r_precip, _ = pearsonr(df_lag['veg_pred'], df_lag[f'precip_lag{lag}'])
        r_temp_anom, _ = pearsonr(df_lag['veg_pred'], df_lag[f'temp_anom_lag{lag}'])
        r_precip_anom, _ = pearsonr(df_lag['veg_pred'], df_lag[f'precip_anom_lag{lag}'])

        results.append({
            'Habitat': habitat,
            'Satellite': satellite,
            'Lag': lag,
            'Temp_r': r_temp,
            'Precip_r': r_precip,
            'Temp_anom_r': r_temp_anom,
            'Precip_anom_r': r_precip_anom,
            'N': len(df_lag)
        })

    if len(results) == 0:
        return None

    lag_df = pd.DataFrame(results)

    # -----------------------
    # Identify best lags
    # -----------------------
    best_temp_row = lag_df.loc[lag_df['Temp_r'].abs().idxmax()]
    best_precip_row = lag_df.loc[lag_df['Precip_r'].abs().idxmax()]
    best_temp_anom_row = lag_df.loc[lag_df['Temp_anom_r'].abs().idxmax()]
    best_precip_anom_row = lag_df.loc[lag_df['Precip_anom_r'].abs().idxmax()]

    # -----------------------
    # Summary table
    # -----------------------
    summary = {
        'Habitat': habitat,
        'Satellite': satellite,

        'Best_temp_lag': int(best_temp_row['Lag']),
        'Best_temp_r': best_temp_row['Temp_r'],

        'Best_precip_lag': int(best_precip_row['Lag']),
        'Best_precip_r': best_precip_row['Precip_r'],

        'Best_temp_anom_lag': int(best_temp_anom_row['Lag']),
        'Best_temp_anom_r': best_temp_anom_row['Temp_anom_r'],

        'Best_precip_anom_lag': int(best_precip_anom_row['Lag']),
        'Best_precip_anom_r': best_precip_anom_row['Precip_anom_r'],

        'N_months': len(df_h)
    }

    return lag_df, summary
```

```python
# Step 1 Load the data
satellite_map = {
    'Landsat': pd.read_csv('RF_Prediction/rf_pred_landsat.csv'),
    'Sentinel-2': pd.read_csv('RF_Prediction/rf_pred_sentinel2.csv'),
    'MODIS': pd.read_csv('RF_Prediction/rf_pred_modis.csv')
}

tc_clim_areas = pd.read_csv("Climate_TerraClimate_StudyAreas_Monthly_2000_2024.csv") #1980 2024
tc_clim_stations = pd.read_csv("Climate_TerraClimate_Stations_Monthly_2000_2024.csv") #1980 2024

climate_df = tc_clim_stations.copy()

if 'date' not in climate_df.columns:
    climate_df['date'] = pd.to_datetime(
        dict(year=climate_df['year'], month=climate_df['month'], day=1)
    )
    
# Step 2 Aggregate climate
clim_monthly = aggregate_climate_monthly(climate_df)
climatology = compute_climatology(clim_monthly)
clim_anom = compute_anomalies(clim_monthly, climatology)

# Step 3 Merge Climate with Veg predictions by satellites
merged_all = []

for sat, veg_df in satellite_map.items():

    veg_df = veg_df.copy()
    veg_df['Satellite'] = sat  # IMPORTANT

    veg_monthly = aggregate_veg_monthly(veg_df)

    rf_clim_df = merge_climate_veg(clim_anom, veg_monthly)

    merged_all.append(rf_clim_df)

rf_clim_df = pd.concat(merged_all, ignore_index=True)

# Step 4 Run lag analysis
all_lag_tables = []
summaries = []

for sat in rf_clim_df['Satellite'].unique():
    for habitat in rf_clim_df['Habitat'].unique():

        res = compare_climate_vs_anomaly(rf_clim_df, habitat, sat)

        if res is not None:
            lag_df, summary = res
            all_lag_tables.append(lag_df)
            summaries.append(summary)

lag_results_df = pd.concat(all_lag_tables, ignore_index=True)
summary_df = pd.DataFrame(summaries)

# Step 5 Export results
lag_results_df.to_csv("Lag_Correlation_All.csv", index=False)
summary_df.to_csv("Lag_Summary_BestLag.csv", index=False)

print(summary_df.round(3))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def plot_best_corr_heatmap(best_corr_df,
                           value_col='Best_Corr',
                           row_col='Habitat',
                           col_col='Variable',
                           sat_col='Satellite',
                           figsize=(10,6),
                           cmap='RdBu_r'):
    """
    Plot heatmap of best correlations.

    Parameters
    ----------
    best_corr_df : DataFrame
        Must contain columns: Habitat, Satellite, Variable, Best_Corr
    """

    # Create combined row label (Habitat + Satellite)
    best_corr_df['Row'] = best_corr_df[row_col] + ' | ' + best_corr_df[sat_col]

    # Pivot table → ensures 4 columns
    heatmap_df = best_corr_df.pivot(index='Row',
                                    columns=col_col,
                                    values=value_col)

    # Sort columns to fixed order
    desired_order = ['Raw_Precip', 'Anom_Precip',
                     'Raw_Temp', 'Anom_Temp']

    heatmap_df = heatmap_df.reindex(columns=desired_order)

    # Plot
    plt.figure(figsize=figsize)

    sns.heatmap(
        heatmap_df,
        annot=True,
        fmt=".2f",
        cmap=cmap,
        center=0,
        linewidths=0.5,
        linecolor='lightgray',
        cbar_kws={'label': 'Correlation'}
    )

    plt.title("Best Correlation Heatmap", fontsize=14, weight='bold')
    plt.xlabel("Climate Variable")
    plt.ylabel("Habitat | Satellite")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def plot_correlation_heatmap(summary_df,
                             outpath=None,
                             sort_rows=True,
                             cmap="RdBu_r",
                             vmin=-1,
                             vmax=1):

    # ---------------------------------
    # Prepare matrix
    # ---------------------------------
    heat_df = summary_df.copy()

    heat_df["Temp"] = heat_df["Best_temp_r"]
    heat_df["Precip"] = heat_df["Best_precip_r"]
    heat_df["Temp_anom"] = heat_df["Best_temp_anom_r"]
    heat_df["Precip_anom"] = heat_df["Best_precip_anom_r"]

    heat_df["HabSat"] = heat_df["Habitat"] + " | " + heat_df["Satellite"]

    matrix = heat_df.set_index("HabSat")[[
        "Temp", "Precip", "Temp_anom", "Precip_anom"
    ]]

    # ---------------------------------
    # Optional sorting by strongest signal
    # ---------------------------------
    if sort_rows:
        matrix["max_abs"] = matrix.abs().max(axis=1)
        matrix = matrix.sort_values("max_abs", ascending=False)
        matrix = matrix.drop(columns="max_abs")

    # ---------------------------------
    # Plot
    # ---------------------------------
    plt.figure(figsize=(10, 7))

    im = plt.imshow(
        matrix.values,
        aspect='auto',
        cmap=cmap,
        vmin=vmin,
        vmax=vmax
    )

    # Colorbar
    cbar = plt.colorbar(im)
    cbar.set_label("Correlation (r)", fontsize=11)

    # Axis ticks
    plt.xticks(range(len(matrix.columns)), matrix.columns, fontsize=11)
    plt.yticks(range(len(matrix.index)), matrix.index, fontsize=10)

    # Gridlines
    plt.grid(which='major', color='lightgray', linestyle='-', linewidth=0.5)

    # Annotate values
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            value = matrix.iloc[i, j]
            plt.text(j, i, f"{value:.2f}",
                     ha='center', va='center',
                     fontsize=9,
                     color='black')

    # Titles
    plt.title("Climate–Vegetation Correlation Strength",
              fontsize=14, weight='bold')
    plt.xlabel("Climate Variable", fontsize=12)
    plt.ylabel("Habitat | Satellite", fontsize=12)

    plt.tight_layout()

    # Save
    if outpath:
        plt.savefig(outpath, dpi=300, bbox_inches="tight")

    plt.show()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import pearsonr

def plot_veg_vs_climate_bestlag(
    df,
    summary_table_path,
    habitat,
    satellite,
    climate_var="precip",   # "precip" or "temp"
    use_anomaly=True,
    veg_col="veg_pred",
    outdir="Plots_Climate_Lag",
    smooth_window=None      # e.g., 3 for smoothing
):

    os.makedirs(outdir, exist_ok=True)

    # -------------------------
    # Load summary table
    # -------------------------
    summary_df = pd.read_csv(summary_table_path)

    match = summary_df[
        (summary_df["Habitat"] == habitat) &
        (summary_df["Satellite"] == satellite)
    ]

    if match.empty:
        print(f"No summary found for {habitat}-{satellite}")
        return

    # -------------------------
    # Select lag column
    # -------------------------
    lag_lookup = {
        ("precip", True): "Best_precip_anom_lag",
        ("precip", False): "Best_precip_lag",
        ("temp", True): "Best_temp_anom_lag",
        ("temp", False): "Best_temp_lag"
    }

    lag_col = lag_lookup[(climate_var, use_anomaly)]

    if lag_col not in match.columns:
        print(f"{lag_col} not found in summary table")
        return

    lag_months = int(match[lag_col].iloc[0])
    print(f"Using lag = {lag_months} months")

    # -------------------------
    # Filter data
    # -------------------------
    df_h = df[
        (df["Habitat"] == habitat) &
        (df["Satellite"] == satellite)
    ].copy()

    if df_h.empty:
        print("No data available after filtering")
        return

    df_h["date"] = pd.to_datetime(df_h["date"])
    df_h = df_h.sort_values("date")

    # -------------------------
    # Climate column selection
    # -------------------------
    climate_lookup = {
        ("precip", True): "precip_anomaly",
        ("precip", False): "precip_total",
        ("temp", True): "temp_anomaly",
        ("temp", False): "temp_mean"
    }

    climate_col = climate_lookup[(climate_var, use_anomaly)]

    if climate_col not in df_h.columns:
        print(f"{climate_col} not found in dataframe")
        return

    # -------------------------
    # Monthly aggregation
    # -------------------------
    df_h["month"] = df_h["date"].dt.to_period("M")

    df_m = (
        df_h.groupby("month")
        .agg(
            veg_mean=(veg_col, "mean"),
            climate_mean=(climate_col, "mean")
        )
        .reset_index()
    )

    df_m["date"] = df_m["month"].dt.to_timestamp()

    # Optional smoothing
    if smooth_window:
        df_m["veg_mean"] = df_m["veg_mean"].rolling(smooth_window, center=True).mean()
        df_m["climate_mean"] = df_m["climate_mean"].rolling(smooth_window, center=True).mean()

    # -------------------------
    # Apply lag
    # -------------------------
    df_m["climate_lag"] = df_m["climate_mean"].shift(lag_months)
    df_m = df_m.dropna()

    if len(df_m) < 10:
        print("Not enough data after lag")
        return

    # -------------------------
    # Correlation
    # -------------------------
    r, p = pearsonr(df_m["veg_mean"], df_m["climate_lag"])

    # -------------------------
    # Plot
    # -------------------------
    fig, ax1 = plt.subplots(figsize=(10,4))

    # Vegetation line
    ax1.plot(
        df_m["date"],
        df_m["veg_mean"],
        lw=2.5,
        marker="o",
        label="Vegetation"
    )

    ax1.set_ylabel("Predicted vegetation cover")
    ax1.grid(alpha=0.3)

    # Climate axis
    ax2 = ax1.twinx()

    ax2.bar(
        df_m["date"],
        df_m["climate_lag"],
        alpha=0.4,
        label=f"{climate_var} ({'anom' if use_anomaly else 'raw'}) lag={lag_months}"
    )

    ax2.set_ylabel(
        f"{climate_var} {'anomaly' if use_anomaly else 'value'}"
    )

    # -------------------------
    # Title with correlation
    # -------------------------
    ax1.set_title(
        f"{habitat} | {satellite}\n"
        f"Vegetation vs {climate_var} ({'anomaly' if use_anomaly else 'raw'}) "
        f"(lag={lag_months}, r={r:.2f}, p={p:.3f})",
        fontsize=12
    )

    # Legends
    ax1.legend(loc="upper left")
    ax2.legend(loc="upper right")

    # -------------------------
    # Save
    # -------------------------
    outfile = os.path.join(
        outdir,
        f"{habitat}_{satellite}_{climate_var}_{'anom' if use_anomaly else 'raw'}_lag{lag_months}.png"
    )

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.close()

    print(f"Saved to {outfile}")
```

```python
import pandas as pd
import numpy as np
import marineHeatWaves as mhw

# =========================
# 1️ LOAD DATA
# =========================
# Expected columns:
# date| region| precip_mm | temp_C 
# Station| Habitat| date | precip_mm | temp_C 

tc_clim_areas = pd.read_csv("Climate_TerraClimate_StudyAreas_Monthly_2000_2024.csv") #1980 2024
tc_clim_stations = pd.read_csv("Climate_TerraClimate_Stations_Monthly_2000_2024.csv") #1980 2024

df = tc_clim_stations

# Ensure datetime
df['date'] = pd.to_datetime(df['date'])

# Extract time components
df['Year'] = df['date'].dt.year
df['Month'] = df['date'].dt.month

# =========================
# 2️ AGGREGATE BY HABITAT
# =========================
habitat_monthly = (
    df.groupby(['Habitat', 'Year', 'Month'])
      .agg(
          temp_mean=('temp_C', 'mean'),
          precip_total=('precip_mm', 'sum')
      )
      .reset_index()
)

# Create a monthly timestamp
habitat_monthly['date'] = pd.to_datetime(
    habitat_monthly[['Year', 'Month']].assign(day=15)
)

# Sort
habitat_monthly = habitat_monthly.sort_values(['Habitat', 'date'])

# =========================
# 3️ FUNCTION — RUN MHW
# =========================
def detect_temp_anomalies(df_hab):

    df_hab = df_hab.sort_values('date').copy()

    # Convert to ordinal time (required by marineHeatWaves)
    t = df_hab['date'].map(pd.Timestamp.toordinal).values
    temp = df_hab['temp_mean'].values

    # Detect anomalies + heatwaves
    events, clim = mhw.detect(
        t,
        temp,
        climatologyPeriod=[2001, 2024], # or 2010 - 2024
        pctile=90,
        minDuration=3
    )

    # Attach climatology outputs
    df_hab['seas'] = clim['seas']
    df_hab['thresh'] = clim['thresh']
    df_hab['temp_anomaly'] = temp - clim['seas']

    return df_hab, events


# =========================
# 4️ APPLY BY HABITAT
# =========================
results = []
all_events = {}

for habitat, subdf in habitat_monthly.groupby('Habitat'):
    processed, events = detect_temp_anomalies(subdf)
    results.append(processed)
    all_events[habitat] = events

temp_anomaly_df = pd.concat(results, ignore_index=True)

# =========================
# 5️ RAINFALL ANOMALIES
# =========================
# Calculate monthly climatology baseline
rain_clim = (
    temp_anomaly_df.groupby(['Habitat', 'Month'])['precip_total']
    .mean()
    .reset_index()
    .rename(columns={'precip_total': 'precip_clim'})
)

# Merge baseline
temp_anomaly_df = temp_anomaly_df.merge(
    rain_clim,
    on=['Habitat', 'Month'],
    how='left'
)

# Calculate anomaly
temp_anomaly_df['precip_anomaly'] = (
    temp_anomaly_df['precip_total'] - temp_anomaly_df['precip_clim']
)

# =========================
# 6️ FINAL OUTPUT
# =========================
print(temp_anomaly_df.head())

# Save results
temp_anomaly_df.to_csv("habitat_climate_anomalies.csv", index=False)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_climatology_curve(df, baseline_start="1991-01-01", baseline_end="2020-12-31"):

    df = df.copy()
    df["date"] = pd.to_datetime(df["date"])
    df["month"] = df["date"].dt.month

    baseline = df[(df["date"] >= baseline_start) & (df["date"] <= baseline_end)]

    clim = baseline.groupby("month")["temp_C"].mean()
    std = baseline.groupby("month")["temp_C"].std()

    plt.figure(figsize=(8,5))
    plt.plot(clim.index, clim.values, marker="o", label="Climatology")
    plt.fill_between(clim.index, clim-std, clim+std, alpha=0.2, label="±1 SD")

    plt.xticks(range(1,13))
    plt.xlabel("Month")
    plt.ylabel("Temperature (°C)")
    plt.title("Monthly Climatology (1991–2020)")
    plt.legend()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def compare_monthly_climatology(
    df,
    baseline_start="1991-01-01",
    baseline_end="2020-12-31",
    recent_start="2016-01-01",
    recent_end="2023-12-31",
    plot=True
):
    """
    Compare monthly climatology between baseline and recent period.

    Parameters
    ----------
    df : DataFrame
        Must contain columns ['date', 'temp_C']
    baseline_start, baseline_end : str
        Baseline period
    recent_start, recent_end : str
        Comparison period
    plot : bool
        Whether to plot results

    Returns
    -------
    clim_compare : DataFrame
        Monthly climatology comparison table
    """

    data = df.copy()

    # --- Ensure datetime ---
    data["date"] = pd.to_datetime(data["date"], errors="coerce")
    data = data.dropna(subset=["date", "temp_C"])

    # --- Add month ---
    data["month"] = data["date"].dt.month

    # --- Subset periods ---
    baseline = data[(data["date"] >= baseline_start) & (data["date"] <= baseline_end)]
    recent = data[(data["date"] >= recent_start) & (data["date"] <= recent_end)]

    # --- Check coverage ---
    if baseline.empty:
        raise ValueError("Baseline period has no data.")
    if recent.empty:
        raise ValueError("Recent period has no data.")

    # --- Monthly means ---
    clim_base = baseline.groupby("month")["temp_C"].mean()
    clim_recent = recent.groupby("month")["temp_C"].mean()

    # --- Combine ---
    clim_compare = pd.DataFrame({
        "baseline_mean": clim_base,
        "recent_mean": clim_recent
    })

    # --- Difference ---
    clim_compare["bias_recent_minus_baseline"] = (
        clim_compare["recent_mean"] - clim_compare["baseline_mean"]
    )

    # --- Plot ---
    if plot:
        plt.figure(figsize=(8,5))
        plt.plot(clim_compare.index, clim_compare["baseline_mean"],
                 marker="o", label="Baseline (1991–2020)")
        plt.plot(clim_compare.index, clim_compare["recent_mean"],
                 marker="o", label="Recent (2016–2023)")

        plt.xticks(range(1,13))
        plt.xlabel("Month")
        plt.ylabel("Temperature (°C)")
        plt.title("Monthly Climatology Comparison")
        plt.legend()
        plt.show()

        # Bias plot
        plt.figure(figsize=(8,4))
        plt.bar(clim_compare.index, clim_compare["bias_recent_minus_baseline"])
        plt.axhline(0)
        plt.xticks(range(1,13))
        plt.xlabel("Month")
        plt.ylabel("Temperature Difference (°C)")
        plt.title("Warming Bias by Month")
        plt.show()

    return clim_compare
```

#### b.  manual anomaly: Designed function

```python
import pandas as pd

def detect_monthly_heatwaves(clim_df, min_duration=2, pctile=90):
    """
    Detect monthly heatwave-like events from dataframe containing
    date and temperature columns.

    Parameters
    ----------
    clim_df : DataFrame
        Must contain columns ['date', 'temp_C']
    min_duration : int
        Minimum consecutive months
    pctile : int
        Percentile threshold

    Returns
    -------
    events_df : DataFrame
    mask : Series
    """

    # --- Copy to avoid modifying original ---
    df = clim_df.copy()

    # --- Ensure datetime ---
    df['date'] = pd.to_datetime(df['date'])

    # --- Sort by date ---
    df = df.sort_values('date')

    # --- Set index ---
    df = df.set_index('date')

    # --- Keep temperature ---
    df = df[['temp_C']].rename(columns={'temp_C': 'temp'})

    # --- Add calendar month ---
    df['month'] = df.index.month

    # --- Climatology ---
    clim = df.groupby('month')['temp'].mean()
    thresh = df.groupby('month')['temp'].quantile(pctile / 100)

    df['clim'] = df['month'].map(clim)
    df['thresh'] = df['month'].map(thresh)

    # --- Anomaly ---
    df['anom'] = df['temp'] - df['clim']

    # --- Exceedance ---
    df['exceed'] = df['temp'] > df['thresh']

    # --- Consecutive runs ---
    df['event_id'] = (df['exceed'] != df['exceed'].shift()).cumsum()

    groups = df[df['exceed']].groupby('event_id')

    events = []

    for eid, g in groups:
        if len(g) >= min_duration:
            events.append({
                'event_id': eid,
                'start': g.index.min(),
                'end': g.index.max(),
                'duration': len(g),
                'mean_intensity': g['anom'].mean(),
                'max_intensity': g['anom'].max(),
                'cum_intensity': g['anom'].sum()
            })

    events_df = pd.DataFrame(events)

    # --- Event mask ---
    if not events_df.empty:
        valid_ids = events_df['event_id'].values
        mask = df['event_id'].isin(valid_ids) & df['exceed']
    else:
        mask = pd.Series(False, index=df.index)

    return events_df, mask
```

```python
import pandas as pd

def detect_monthly_heatwaves_with_baseline(
    clim_df,
    baseline_start="2000-01-01",
    baseline_end="2020-12-31",
    min_duration=2,
    pctile=90
):

    df = clim_df.copy()

    df["date"] = pd.to_datetime(df["date"])
    df = df.sort_values("date")
    df = df.set_index("date")

    df["month"] = df.index.month

    # --- Baseline subset ---
    baseline = df.loc[baseline_start:baseline_end]

    # --- Climatology ---
    clim = baseline.groupby("month")["temp_C"].mean()
    thresh = baseline.groupby("month")["temp_C"].quantile(pctile/100)

    df["clim"] = df["month"].map(clim)
    df["thresh"] = df["month"].map(thresh)

    # --- Anomaly ---
    df["anom"] = df["temp_C"] - df["clim"]

    # --- Exceedance ---
    df["exceed"] = df["temp_C"] > df["thresh"]

    # --- Consecutive runs ---
    df["event_id"] = (df["exceed"] != df["exceed"].shift()).cumsum()

    groups = df[df["exceed"]].groupby("event_id")

    events = []

    for eid, g in groups:
        if len(g) >= min_duration:
            events.append({
                "event_id": eid,
                "start": g.index.min(),
                "end": g.index.max(),
                "duration": len(g),
                "mean_intensity": g["anom"].mean(),
                "max_intensity": g["anom"].max(),
                "cum_intensity": g["anom"].sum()
            })

    events_df = pd.DataFrame(events)

    mask = df["event_id"].isin(events_df["event_id"]) & df["exceed"] if not events_df.empty else pd.Series(False, index=df.index)

    return events_df, mask, df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ================================
# SETTINGS
# ================================
BASELINE_START = 2000
BASELINE_END = 2020

ANALYSIS_START = 2016
ANALYSIS_END = 2023

PERCENTILE = 0.90

# ================================
# PREPARE DATA
# ================================
df = tc_clim_stations.copy()

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date")

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["temp"] = df["temp_C"]

# ================================
# BASELINE CLIMATOLOGY
# ================================
baseline_df = df[
    (df["year"] >= BASELINE_START) &
    (df["year"] <= BASELINE_END)
]

climatology = (
    baseline_df.groupby(["Habitat","month"])["temp"]
    .mean()
    .rename("baseline")
)

threshold = (
    baseline_df.groupby(["Habitat","month"])["temp"]
    .quantile(PERCENTILE)
    .rename("threshold")
)

# Merge back
df = df.merge(climatology, on=["Habitat","month"], how="left")
df = df.merge(threshold, on=["Habitat","month"], how="left")

# ================================
# FILTER ANALYSIS PERIOD
# ================================
df_analysis = df[
    (df["year"] >= ANALYSIS_START) &
    (df["year"] <= ANALYSIS_END)
].copy()

# ================================
# ANOMALY CALCULATION
# ================================
df_analysis["anomaly"] = df_analysis["temp"] - df_analysis["baseline"]

df_analysis["moderate"] = (
    (df_analysis["temp"] > df_analysis["baseline"]) &
    (df_analysis["temp"] < df_analysis["threshold"])
)

df_analysis["extreme"] = df_analysis["temp"] >= df_analysis["threshold"]

# ================================
# SUMMARY TABLE
# ================================
summary = (
    df_analysis.groupby("Habitat")
    .agg(
        mean_anomaly=("anomaly","mean"),
        max_anomaly=("anomaly","max"),
        extreme_months=("extreme","sum"),
        moderate_months=("moderate","sum")
    )
    .reset_index()
)

print("\n===== HABITAT SUMMARY =====")
print(summary)

# ================================
# PLOT EACH HABITAT
# ================================
habitats = df_analysis["Habitat"].unique()

for hab in habitats:

    sub = df_analysis[df_analysis["Habitat"] == hab]

    plt.figure(figsize=(12,5))

    plt.plot(sub["date"], sub["temp"], label="Observed")
    plt.plot(sub["date"], sub["baseline"], linestyle="--", label="Baseline")
    plt.plot(sub["date"], sub["threshold"], linestyle=":", label="Threshold")

    plt.fill_between(
        sub["date"],
        sub["temp"],
        sub["baseline"],
        where=sub["moderate"],
        alpha=0.3,
        label="Moderate anomaly"
    )

    plt.fill_between(
        sub["date"],
        sub["temp"],
        sub["threshold"],
        where=sub["extreme"],
        alpha=0.5,
        label="Extreme anomaly"
    )

    plt.title(f"Temperature Anomalies (2016–2023) — {hab}")
    plt.xlabel("Year")
    plt.ylabel("Temperature (°C)")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ================================
# SETTINGS
# ================================
BASELINE_YEARS = (2010, 2024)   # 15-year baseline
ANALYSIS_YEARS = (2016, 2024)

GROUP_COL = "Habitat" 

# ================================
# PREPARE DATA
# ================================
df = tc_clim_stations.copy()

df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year

# ================================
# MONTHLY HABITAT MEAN
# ================================
monthly_habitat = (
    df.groupby([GROUP_COL, "year", df["date"].dt.month])["precip_mm"]
    .mean()   # mean across stations
    .reset_index(name="monthly_mean_precip")
)

# ================================
# ANNUAL TOTAL RAINFALL
# ================================
annual_rain = (
    monthly_habitat.groupby([GROUP_COL, "year"])["monthly_mean_precip"]
    .sum()   # sum 12 months
    .reset_index()
    .rename(columns={"monthly_mean_precip": "annual_rain"})
)

# ================================
# BASELINE CLIMATOLOGY
# ================================
baseline = annual_rain[
    (annual_rain["year"] >= BASELINE_YEARS[0]) &
    (annual_rain["year"] <= BASELINE_YEARS[1])
]

baseline_mean = (
    baseline.groupby(GROUP_COL)["annual_rain"]
    .mean()
    .rename("baseline_rain")
    .reset_index()
)

# Merge baseline
annual_rain = annual_rain.merge(baseline_mean, on=GROUP_COL, how="left")

# ================================
# CALCULATE ANOMALY
# ================================
annual_rain["rain_anomaly"] = (
    annual_rain["annual_rain"] - annual_rain["baseline_rain"]
)

# ================================
# FILTER ANALYSIS PERIOD
# ================================
rain_analysis = annual_rain[
    (annual_rain["year"] >= ANALYSIS_YEARS[0]) &
    (annual_rain["year"] <= ANALYSIS_YEARS[1])
]

print("\n===== Rainfall anomaly sample =====")
print(rain_analysis.head())

# ================================
# SUMMARY TABLE
# ================================
summary = (
    rain_analysis.groupby(GROUP_COL)
    .agg(
        mean_anomaly=("rain_anomaly","mean"),
        max_deficit=("rain_anomaly","min"),
        max_surplus=("rain_anomaly","max"),
        sd=("rain_anomaly","std")
    )
    .reset_index()
)

print("\n===== SUMMARY =====")
print(summary)

# ================================
# OUTPUT FOLDER
# ================================
output_dir = "climate_anomaly_plots"
os.makedirs(output_dir, exist_ok=True)

# ================================
# LOOP THROUGH SITES
# ================================
sites = rain_analysis[GROUP_COL].unique()

for site in sites:

    sub = rain_analysis[rain_analysis[GROUP_COL] == site]

    plt.figure(figsize=(6,3))

    plt.bar(sub["year"], sub["rain_anomaly"])
    plt.axhline(0, linestyle="--")

    plt.title(f"Annual Rainfall Anomaly — {site}")
    plt.ylabel("Anomaly (mm)")
    plt.xlabel("Year")

    plt.tight_layout()

    # Clean filename (remove spaces)
    safe_name = str(site).replace(" ", "_")

    # Save figure
    plt.savefig(
        f"{output_dir}/rain_anomaly_{safe_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()   # VERY IMPORTANT to avoid memory issues
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ================================
# SETTINGS
# ================================
BASELINE_YEARS = (2010, 2024)
ANALYSIS_YEARS = (2016, 2024)
GROUP_COL = "Habitat"

# ================================
# PREPARE DATA
# ================================
df = tc_clim_stations.copy()

df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

# ================================
# MONTHLY HABITAT MEAN TEMP
# ================================
monthly_temp = (
    df.groupby([GROUP_COL, "year", "month"])["temp_C"]
    .mean()   # mean across stations
    .reset_index(name="monthly_temp")
)

# Create date column
monthly_temp["date"] = pd.to_datetime(
    dict(year=monthly_temp["year"], month=monthly_temp["month"], day=1)
)

# ================================
# MONTHLY CLIMATOLOGY (baseline)
# ================================
baseline = monthly_temp[
    (monthly_temp["year"] >= BASELINE_YEARS[0]) &
    (monthly_temp["year"] <= BASELINE_YEARS[1])
]

monthly_climatology = (
    baseline.groupby([GROUP_COL, "month"])["monthly_temp"]
    .mean()
    .reset_index(name="baseline_temp")
)

# Merge climatology
monthly_temp = monthly_temp.merge(
    monthly_climatology,
    on=[GROUP_COL, "month"],
    how="left"
)

# ================================
# CALCULATE MONTHLY ANOMALY
# ================================
monthly_temp["temp_anomaly"] = (
    monthly_temp["monthly_temp"] -
    monthly_temp["baseline_temp"]
)

# ================================
# FILTER ANALYSIS PERIOD
# ================================
temp_analysis = monthly_temp[
    (monthly_temp["year"] >= ANALYSIS_YEARS[0]) &
    (monthly_temp["year"] <= ANALYSIS_YEARS[1])
]

# ================================
# OUTPUT FOLDER
# ================================
output_dir = "climate_anomaly_plots"
os.makedirs(output_dir, exist_ok=True)

# ================================
# PLOT PER HABITAT
# ================================
sites = temp_analysis[GROUP_COL].unique()

for site in sites:

    sub = temp_analysis[temp_analysis[GROUP_COL] == site].sort_values("date")

    fig, ax = plt.subplots(figsize=(9,4))

    # -------------------------
    # Plot observed temperature
    # -------------------------
    ax.plot(
        sub["date"],
        sub["monthly_temp"],
        color="black",
        linewidth=1.5,
        marker="o",
        label="Observed Temperature"
    )

    # -------------------------
    # Plot baseline climatology
    # -------------------------
    ax.plot(
        sub["date"],
        sub["baseline_temp"],
        linestyle="--",
        color="gray",
        linewidth=1,
        label="Baseline (2010–2024)"
    )

    # -------------------------
    # Red/Blue anomaly shading
    # -------------------------
    ax.fill_between(
        sub["date"],
        sub["monthly_temp"],
        sub["baseline_temp"],
        where=(sub["temp_anomaly"] >= 0),
        interpolate=True,
        alpha=0.4,
        label="Positive Anomaly"
    )

    ax.fill_between(
        sub["date"],
        sub["monthly_temp"],
        sub["baseline_temp"],
        where=(sub["temp_anomaly"] < 0),
        interpolate=True,
        alpha=0.4,
        label="Negative Anomaly"
    )

    # Improve colors manually (after fill to control legend cleanly)
    for collection in ax.collections:
        if len(collection.get_paths()) > 0:
            if collection.get_label() == "Positive Anomaly":
                collection.set_facecolor("red")
            elif collection.get_label() == "Negative Anomaly":
                collection.set_facecolor("blue")

    # -------------------------
    # Labels & formatting
    # -------------------------
    ax.set_title(f"Monthly Temperature Anomaly — {site}")
    ax.set_ylabel("Temperature (°C)")
    ax.set_xlabel("Year")

    ax.grid(alpha=0.3)

    ax.legend(loc="upper left", frameon=False)

    plt.tight_layout()

    safe_name = str(site).replace(" ", "_")

    plt.savefig(
        f"{output_dir}/monthly_temp_anomaly_shaded_{safe_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
```

```python
# STEP 1 — Build climate anomaly dataset
# add monthly temperature anomaly
# add monthly precipitation anomaly
# add heatwave flag

import pandas as pd
import numpy as np

# Create anomalies
# ================================
# SETTINGS
# ================================
BASELINE_YEARS = (2010, 2024)

df = tc_clim_stations.copy()

df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

# ================================
# BASELINE CLIMATOLOGY
# ================================
baseline = df[
    (df["year"] >= BASELINE_YEARS[0]) &
    (df["year"] <= BASELINE_YEARS[1])
]

clim = (
    baseline.groupby(["Habitat","month"])
    .agg(
        temp_clim=("temp_C","mean"),
        precip_clim=("precip_mm","mean")
    )
    .reset_index()
)

df = df.merge(clim, on=["Habitat","month"], how="left")

# ================================
# ANOMALIES
# ================================
df["temp_anomaly"] = df["temp_C"] - df["temp_clim"]
df["precip_anomaly"] = df["precip_mm"] - df["precip_clim"]

# STEP 2 — Create heatwave flag
# Percentile threshold per habitat
thresh = (
    baseline.groupby("Habitat")["temp_C"]
    .quantile(0.9)
    .rename("temp_thresh")
    .reset_index()
)

df = df.merge(thresh, on="Habitat", how="left")

df["heatwave_flag"] = (df["temp_C"] > df["temp_thresh"]).astype(int)

# Climate dataframe
df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

df_clim_ano = df

df_clim_ano.to_csv("Climate_TerraClimate_Anomaly_StudyAreas.csv",index=False)
# # STEP 3 — Merge with vegetation dataset
# df_all = pd.merge(
#     veg_df,
#     df,
#     on=["date","Habitat"],
#     how="inner"
# )
```

```python
import pandas as pd
import numpy as np
# =====================================================
# 1. Load and prepare data
# =====================================================
clim_areas = pd.read_csv("Climate_Areas_Monthly_1980_Present.csv")

clim_areas["date"] = pd.to_datetime(clim_areas["date"])
clim_areas["year"] = clim_areas["date"].dt.year
clim_areas["month"] = clim_areas["date"].dt.month

# =====================================================
# 2. Assign climatological seasons
# =====================================================
def month_to_season(month):
    if month in [12, 1, 2]:
        return "Winter (DJF)"
    elif month in [3, 4, 5]:
        return "Spring (MAM)"
    elif month in [6, 7, 8]:
        return "Summer (JJA)"
    else:
        return "Autumn (SON)"

clim_areas["season"] = clim_areas["month"].apply(month_to_season)

# =====================================================
# 3. Seasonal mean temperature (long-term)
#    dataset × region × season
# =====================================================
seasonal_temp = (
    clim_areas
    .groupby(["dataset", "region", "season"])["temp_C"]
    .mean()
    .reset_index()
)

print("\n=== Long-term Seasonal Mean Temperature (°C) ===")
print(seasonal_temp)

# =====================================================
# 4. Extract winter & summer ranges (for reporting)
# =====================================================
winter_summer = seasonal_temp[
    seasonal_temp["season"].isin(["Winter (DJF)", "Summer (JJA)"])
]

summary_ranges = (
    winter_summer
    .groupby(["dataset", "season"])["temp_C"]
    .agg(["min", "max"])
    .reset_index()
    .rename(columns={
        "min": "min_temp_C",
        "max": "max_temp_C"
    })
)

print("\n=== Winter & Summer Temperature Ranges (°C) ===")
print(summary_ranges)

# =====================================================
# 5. OPTIONAL: save outputs
# =====================================================
seasonal_temp.to_csv(
    "Seasonal_Mean_Temperature_by_Dataset_Region.csv",
    index=False
)

summary_ranges.to_csv(
    "Winter_Summer_Temperature_Ranges_by_Dataset.csv",
    index=False
)

# =====================================================
# 2. Long-term AVERAGE ANNUAL PRECIPITATION
#    (mm/year, per dataset × region)
# =====================================================

annual_precip_results = []

for ds in clim_areas["dataset"].unique():
    for reg in clim_areas["region"].unique():

        df = clim_areas[
            (clim_areas["dataset"] == ds) &
            (clim_areas["region"] == reg)
        ]

        if df.empty:
            continue

        # Monthly → Annual sum
        annual_precip = (
            df.groupby("year")["precip_mm"]
              .sum()
        )

        # Long-term mean of annual totals
        mean_annual_precip = annual_precip.mean()

        annual_precip_results.append({
            "dataset": ds,
            "region": reg,
            "mean_annual_precip_mm": mean_annual_precip,
            "n_years": annual_precip.shape[0]
        })

mean_annual_precip_df = pd.DataFrame(annual_precip_results)

print("\n=== Long-term Average Annual Precipitation (mm/year) ===")
# print(mean_annual_precip_df)


# =====================================================
# 3. Long-term MONTHLY MEAN TEMPERATURE (climatology)
#    (°C, per dataset × region)
# =====================================================

monthly_temp_results = []

for ds in clim_areas["dataset"].unique():
    for reg in clim_areas["region"].unique():

        df = clim_areas[
            (clim_areas["dataset"] == ds) &
            (clim_areas["region"] == reg)
        ]

        if df.empty:
            continue

        # Monthly climatology
        monthly_temp = (
            df.groupby("month")["temp_C"]
              .mean()
              .reset_index()
        )

        monthly_temp["dataset"] = ds
        monthly_temp["region"] = reg

        monthly_temp_results.append(monthly_temp)

monthly_temp_df = pd.concat(monthly_temp_results, ignore_index=True)

print("\n=== Long-term Monthly Mean Temperature (°C) ===")
# print(monthly_temp_df)


# =====================================================
# 4. OPTIONAL: save results
# =====================================================

mean_annual_precip_df.to_csv(
    "Mean_Annual_Precip_by_Dataset_Region.csv",
    index=False
)

monthly_temp_df.to_csv(
    "Monthly_Temperature_Climatology_by_Dataset_Region.csv",
    index=False
)

# =====================================================
# 6. Annual precipitation per year (for wet/dry year identification)
# =====================================================
annual_precip_per_year = []

for ds in clim_areas["dataset"].unique():
    for reg in clim_areas["region"].unique():

        df = clim_areas[
            (clim_areas["dataset"] == ds) &
            (clim_areas["region"] == reg)
        ]

        if df.empty:
            continue

        # Monthly → Annual sum
        annual_precip = df.groupby("year")["precip_mm"].sum().reset_index()
        annual_precip["dataset"] = ds
        annual_precip["region"] = reg

        annual_precip_per_year.append(annual_precip)

annual_precip_yearly_df = pd.concat(annual_precip_per_year, ignore_index=True)

print("\n=== Annual Precipitation per Year (mm/year) ===")
# print(annual_precip_yearly_df.head())

# Save yearly precipitation for further analysis
annual_precip_yearly_df.to_csv(
    "Annual_Precipitation_by_Year_Dataset_Region.csv",
    index=False
)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

def plot_annual_precipitation_trends(
    df,
    dataset=None,
    regions=None,
    start_year=None,
    end_year=None,
    show_mean=True,   # now means "show zero baseline"
    figsize=(12, 6),
    outdir='Plot_Clim'
):
    """
    Plot annual precipitation anomalies by dataset and region.
    The mean annual precipitation is shifted to y = 0 (baseline).

    Parameters
    ----------
    df : pandas.DataFrame
        Must contain columns: ['year','precip_mm','dataset','region']
    """

    sns.set(style="whitegrid")

    df_plot = df.copy()

    if dataset is not None:
        df_plot = df_plot[df_plot["dataset"] == dataset]
    if regions is not None:
        df_plot = df_plot[df_plot["region"].isin(regions)]
    if start_year is not None:
        df_plot = df_plot[df_plot["year"] >= start_year]
    if end_year is not None:
        df_plot = df_plot[df_plot["year"] <= end_year]

    if df_plot.empty:
        raise ValueError("No data available for the selected dataset/regions/year range")

    regions_unique = df_plot["region"].unique()
    n_regions = len(regions_unique)

    fig, axes = plt.subplots(
        n_regions, 1,
        figsize=(figsize[0], figsize[1] * n_regions),
        sharex=True
    )

    if n_regions == 1:
        axes = [axes]

    for ax, region in zip(axes, regions_unique):
        df_region = (
            df_plot[df_plot["region"] == region]
            .sort_values("year")
            .copy()
        )

        # ---- shift mean to baseline ----
        mean_precip = df_region["precip_mm"].mean()
        df_region["precip_anomaly"] = df_region["precip_mm"] - mean_precip

        # Plot anomalies
        ax.bar(
            df_region["year"],
            df_region["precip_anomaly"],
            color="skyblue",
            edgecolor="black",
            label="Annual precipitation anomaly"
        )

        # Zero line = mean precipitation
        if show_mean:
            ax.axhline(
                0,
                color="red",
                linestyle="--",
                linewidth=2,
                label=f"Mean ({mean_precip:.1f} mm)"
            )

        ax.set_ylabel("Precipitation anomaly (mm)")
        ax.set_title(
            f"{dataset if dataset else 'All Datasets'} – Annual Precipitation Anomaly ({region})"
        )

        ax.legend(frameon=False)

    plt.xlabel("Year")
    plt.tight_layout()
    outfile = os.path.join(outdir, f"Climte_annaul_preci_trends.png")
    plt.savefig(outfile, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os

def plot_monthly_temp_precip_faceted_bar(
    df,
    dataset="ERA5",
    start_year=None,
    regions=None,
    show_temp=True,
    show_precip=True,
    outdir='Plot_Clim'
):
    """
    Faceted monthly climate trends by region.
    Temperature is the main y-axis (line), precipitation as dual-axis (bars).

    Parameters
    ----------
    df : pandas.DataFrame
        Must contain columns: ['dataset','region','date','precip_mm','temp2m_C']
    dataset : str
        One of ['ERA5','CHIRPS','TerraClimate']
    start_year : int or None
        Filter data starting from this year
    regions : list or None
        Subset of regions to plot (None = all)
    show_temp : bool
        Whether to plot temperature
    show_precip : bool
        Whether to plot precipitation
    """
    sns.set(style="whitegrid")

    # Filter dataset
    df = df[df["dataset"] == dataset].copy()
    df["date"] = pd.to_datetime(df["date"])
    
    if start_year is not None:
        df = df[df['date'].dt.year >= start_year]
    if regions is not None:
        df = df[df["region"].isin(regions)]
    if df.empty:
        raise ValueError(f"No data available for dataset={dataset} with current filters")

    regions_unique = df["region"].unique()
    n_regions = len(regions_unique)
    fig, axes = plt.subplots(n_regions, 1, figsize=(14, 4.5 * n_regions), sharex=True)
    if n_regions == 1:
        axes = [axes]

    # Loop through regions
    for ax, region in zip(axes, regions_unique):
        subset = df[df["region"] == region].sort_values("date")

        # Temperature as main y-axis (line)
        if show_temp:
            ax.plot(subset["date"], subset["temp_C"], marker="o", linestyle="-", color="black", linewidth=2, label="Temperature (°C)")
            ax.set_ylabel("Temperature (°C)", color="black")
            ax.tick_params(axis="y", labelcolor="black")

        # Precipitation as dual-axis (bars)
        if show_precip:
            ax2 = ax.twinx()
            ax2.bar(subset["date"], subset["precip_mm"], color="dimgray", width=20, alpha=0.8, label="Precipitation (mm)")
            ax2.set_ylabel("Precipitation (mm)", color="dimgray")
            ax2.tick_params(axis="y", labelcolor="dimgray")

        # Combine legends
        h1, l1 = ax.get_legend_handles_labels()
        h2, l2 = ax2.get_legend_handles_labels() if show_precip else ([], [])
        ax.legend(h1 + h2, l1 + l2, loc="upper left")

        ax.set_title(f"{dataset} – Monthly Climate Trends ({region})", fontsize=14)

    plt.xlabel("Date")
    plt.tight_layout()
    # Save figure to file
    save_fp = os.path.join(outdir,f"{dataset}_{start_year}_Monthly_Climatology.png")
    plt.savefig(save_fp, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import math
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

def plot_monthly_temp_precip_faceted_habitat(
    df,
    dataset="ERA5",
    start_year=None,
    habitats=None,
    show_temp=True,
    show_precip=True,
    outdir='Plot_Clim'
):
    """
    Plot monthly time series of temperature and precipitation by habitat.
    Temperature is main y-axis (line), precipitation is dual-axis (bars).
    Each month in each year is plotted, values averaged across stations within each habitat.

    Parameters
    ----------
    df : pandas.DataFrame
        Must contain columns: ['dataset','Habitat','Station','date','precip_mm','temp_C']
    dataset : str
        One of ['ERA5','CHIRPS','TerraClimate']
    start_year : int or None
        Filter data starting from this year
    habitats : list or None
        Subset of habitats to plot (None = all)
    show_temp : bool
        Whether to plot temperature
    show_precip : bool
        Whether to plot precipitation
    """
    sns.set(style="whitegrid")
    
    # Filter dataset
    df = df[df["dataset"] == dataset].copy()
    df["date"] = pd.to_datetime(df["date"])
    
    if start_year is not None:
        df = df[df['date'].dt.year >= start_year]
    if habitats is not None:
        df = df[df["Habitat"].isin(habitats)]
    if df.empty:
        raise ValueError(f"No data available for dataset={dataset} with current filters")
    
    # Average by habitat for each month
    monthly_df = (
        df.groupby(['Habitat','date'])
          .agg(temp_C=('temp_C','mean'),
               precip_mm=('precip_mm','mean'))
          .reset_index()
    )
    
    habitats_unique = monthly_df["Habitat"].unique()
    n_habitats = len(habitats_unique)
    
    # Facet layout: 2 columns
    cols = 2
    rows = math.ceil(n_habitats / cols)
    fig, axes = plt.subplots(nrows=rows, ncols=cols, figsize=(14, 3.5*rows), sharex=True)
    axes = axes.flatten()
    
    # Determine y-axis limits
    temp_min, temp_max = monthly_df["temp_C"].min(), monthly_df["temp_C"].max()
    precip_max = monthly_df["precip_mm"].max() * 1.1
    
    for ax, habitat in zip(axes, habitats_unique):
        df_h = monthly_df[monthly_df["Habitat"] == habitat].sort_values("date")
        
        # Temperature (line)
        if show_temp:
            ax.plot(df_h["date"], df_h["temp_C"], color='black', marker='o', linewidth=1.5, label='Temperature (°C)')
            ax.set_ylabel("Temperature (°C)")
            ax.set_ylim(temp_min-1, temp_max+1)
        
        # Precipitation (bars)
        if show_precip:
            ax2 = ax.twinx()
            ax2.bar(df_h["date"], df_h["precip_mm"], width=25, color='dimgray', alpha=0.8, label='Precipitation (mm)')
            ax2.set_ylabel("Precipitation (mm)", color='dimgray')
            ax2.set_ylim(0, precip_max)
        
        ax.set_title(f"{habitat} – {dataset} Monthly Trends")
    
    # Hide empty axes
    for ax in axes[n_habitats:]:
        ax.set_visible(False)
    
    # Single legend
    legend_elements = [
        Line2D([0],[0], color='black', marker='o', label='Temperature', linewidth=2),
        Patch(facecolor='gray', edgecolor='black', alpha=0.6, label='Precipitation')
    ]
    fig.legend(handles=legend_elements, loc='upper right', ncol=2, bbox_to_anchor=(0.945, 0.915))
    
    plt.tight_layout(rect=[0,0,1,0.95])
    save_fp = os.path.join(outdir,f"{dataset}_{start_year if start_year else 'all'}_Monthly_Trends_By_Habitat.png")
    plt.savefig(save_fp, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
def aggregate_rf_monthly(df, value_col='F_alive_Pred'):
    """
    Aggregate RS vegetation predictions to monthly means.

    Returns dataframe ready for climate merge.
    """

    df = df.copy()

    # -------------------
    # Clean dates
    # -------------------
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df = df.dropna(subset=['date', value_col, 'Habitat'])

    # -------------------
    # Create monthly period (BEST for merging)
    # -------------------
    df['month_period'] = df['date'].dt.to_period('M')

    # -------------------
    # Aggregate monthly
    # -------------------
    group_cols = ['Habitat', 'month_period']

    # keep Satellite if exists
    if 'Satellite' in df.columns:
        group_cols.append('Satellite')

    df_monthly = (
        df.groupby(group_cols)[value_col]
        .mean()
        .reset_index()
        .rename(columns={value_col: 'veg_pred'})
    )

    # Convert period back to timestamp (month start)
    df_monthly['date'] = df_monthly['month_period'].dt.to_timestamp()

    return df_monthly
```

#### b. Lag plotting: Orgianal climate vs. Anomaly

```python
import numpy as np
import pandas as pd
from scipy.stats import pearsonr

def climate_lag_analysis_v2(
    df,
    habitat,
    satellite,
    veg_col='veg_pred',
    precip_col='precip_anomaly',
    temp_col='temp_anomaly',
    heatwave_col='heatwave_flag',
    max_lag=6
):

    # -------------------
    # Filter data
    # -------------------
    df_h = df[
        (df['Habitat'] == habitat) &
        (df['Satellite'] == satellite)
    ].copy()

    if len(df_h) < 12:
        return None

    df_h['date'] = pd.to_datetime(df_h['date'])
    df_h['month'] = df_h['date'].dt.to_period('M')

    # -------------------
    # Monthly aggregation
    # -------------------
    df_m = (
        df_h.groupby('month')
        .agg(
            veg_mean=(veg_col, 'mean'),
            precip_anom=(precip_col, 'mean'),
            temp_anom=(temp_col, 'mean'),
            heatwave_freq=(heatwave_col, 'sum')
        )
        .reset_index()
    )

    df_m['date'] = df_m['month'].dt.to_timestamp()

    results = []

    # -------------------
    # Lag correlations
    # -------------------
    for lag in range(max_lag + 1):

        df_m[f'precip_lag{lag}'] = df_m['precip_anom'].shift(lag)
        df_m[f'temp_lag{lag}'] = df_m['temp_anom'].shift(lag)

        df_lag = df_m[['veg_mean', f'precip_lag{lag}', f'temp_lag{lag}']].dropna()

        if len(df_lag) < 10:
            continue

        r_p, p_p = pearsonr(df_lag['veg_mean'], df_lag[f'precip_lag{lag}'])
        r_t, p_t = pearsonr(df_lag['veg_mean'], df_lag[f'temp_lag{lag}'])

        results.append({
            'Habitat': habitat,
            'Satellite': satellite,
            'Lag': lag,
            'Precip_r': r_p,
            'Precip_p': p_p,
            'Temp_r': r_t,
            'Temp_p': p_t,
            'N': len(df_lag)
        })

    if len(results) == 0:
        return None

    results_df = pd.DataFrame(results)

    # -------------------
    # Best lags
    # -------------------
    best_precip = results_df.iloc[results_df['Precip_r'].abs().idxmax()]
    best_temp = results_df.iloc[results_df['Temp_r'].abs().idxmax()]

    # -------------------
    # Heatwave correlation
    # -------------------
    df_hw = df_m[['veg_mean', 'heatwave_freq']].dropna()

    if len(df_hw) > 10:
        r_hw, p_hw = pearsonr(df_hw['veg_mean'], df_hw['heatwave_freq'])
    else:
        r_hw, p_hw = np.nan, np.nan

    summary = {
        "Habitat": habitat,
        "Satellite": satellite,
        "Best_precip_lag": int(best_precip['Lag']),
        "Best_precip_r": best_precip['Precip_r'],
        "Best_temp_lag": int(best_temp['Lag']),
        "Best_temp_r": best_temp['Temp_r'],
        "Heatwave_r": r_hw,
        "Heatwave_p": p_hw,
        "Veg_mean": df_m['veg_mean'].mean(),
        "N_months": len(df_m)
    }

    return results_df, summary
```

```python
all_lag_tables = []
summaries = []

for sat in rf_clim_df['Satellite'].unique():
    for habitat in rf_clim_df['Habitat'].unique():

        res = climate_lag_analysis_v2(rf_clim_df, habitat, sat)

        if res is not None:
            lag_df, summary = res
            all_lag_tables.append(lag_df)
            summaries.append(summary)

lag_results_df = pd.concat(all_lag_tables, ignore_index=True)
summary_df = pd.DataFrame(summaries)

lag_results_df.to_csv("Lag_Correlation_All_heatwave.csv", index=False)
summary_df.to_csv("Lag_Summary_BestLag_heatwave.csv", index=False)

print(summary_df.round(3))
```

```python
from scipy.stats import pearsonr
import pandas as pd
import numpy as np
from pingouin import partial_corr  # for partial correlation

def lagged_climate_summary_extended(df, habitat, satellite,
                                    veg_col="veg_pred",
                                    precip_col="precip_mm",
                                    temp_col="temp_C",
                                    precip_anom_col="precip_anomaly",
                                    temp_anom_col="temp_anomaly",
                                    max_lag=6,
                                    ci_level=0.95):
    """
    Computes absolute & anomaly correlations, confidence intervals, partial correlations
    for a single Habitat × Satellite group.
    Returns: (results_df, summary_dict)
    """

    # -------------------
    # Filter dataset
    # -------------------
    df_h = df[(df['Habitat'] == habitat) & (df['Satellite'] == satellite)].copy()
    if len(df_h) < 12:
        return None

    # -------------------
    # Function to compute CI for Pearson r
    # -------------------
    def pearson_ci(r, n, ci=0.95):
        if n <= 3 or np.isnan(r):
            return (np.nan, np.nan)
        from math import atanh, tanh
        from scipy.stats import norm

        # Clip r to avoid math domain error at exactly 1 or -1
        r = np.clip(r, -0.999999, 0.999999)

        z = atanh(r)
        se = 1 / np.sqrt(n - 3)
        z_crit = norm.ppf(1 - (1 - ci)/2)
        lower = tanh(z - z_crit*se)
        upper = tanh(z + z_crit*se)
        return lower, upper

    # -------------------
    # Scan precipitation lags
    # -------------------
    best_r_abs = -np.inf
    best_p_abs = np.nan
    best_lag_abs = None

    best_r_anom = -np.inf
    best_p_anom = np.nan
    best_lag_anom = None

    for lag in range(0, max_lag + 1):
        df_h[f"precip_lag{lag}"] = df_h[precip_col].shift(lag)
        df_h[f"precip_anom_lag{lag}"] = df_h[precip_anom_col].shift(lag)

        # Absolute precipitation
        df_tmp_abs = df_h[[veg_col, f"precip_lag{lag}"]].dropna()
        if len(df_tmp_abs) >= 12 and df_tmp_abs[veg_col].nunique() > 1:
            r, p = pearsonr(df_tmp_abs[veg_col], df_tmp_abs[f"precip_lag{lag}"])
            if abs(r) > abs(best_r_abs):
                best_r_abs = r
                best_p_abs = p
                best_lag_abs = lag

        # Anomaly precipitation
        df_tmp_anom = df_h[[veg_col, f"precip_anom_lag{lag}"]].dropna()
        if len(df_tmp_anom) >= 12 and df_tmp_anom[veg_col].nunique() > 1:
            r, p = pearsonr(df_tmp_anom[veg_col], df_tmp_anom[f"precip_anom_lag{lag}"])
            if abs(r) > abs(best_r_anom):
                best_r_anom = r
                best_p_anom = p
                best_lag_anom = lag

    # -------------------
    # Temperature correlations (absolute)
    # -------------------
    df_temp_abs = df_h[[veg_col, temp_col, precip_col]].dropna()
    if len(df_temp_abs) >= 12 and df_temp_abs[veg_col].nunique() > 1:
        temp_r_abs, temp_p_abs = pearsonr(df_temp_abs[veg_col], df_temp_abs[temp_col])
        try:
            temp_partial_abs = partial_corr(df=df_temp_abs, x=veg_col, y=temp_col,
                                            covar=precip_col, method='pearson')
            temp_r_partial_abs = temp_partial_abs['r'].values[0]
            temp_p_partial_abs = temp_partial_abs['p-val'].values[0]
        except:
            temp_r_partial_abs = temp_p_partial_abs = np.nan
    else:
        temp_r_abs = temp_p_abs = temp_r_partial_abs = temp_p_partial_abs = np.nan

    # Temperature correlations (anomaly)
    df_temp_anom = df_h[[veg_col, temp_anom_col, precip_anom_col]].dropna()
    if len(df_temp_anom) >= 12 and df_temp_anom[veg_col].nunique() > 1:
        temp_r_anom, temp_p_anom = pearsonr(df_temp_anom[veg_col], df_temp_anom[temp_anom_col])
        try:
            temp_partial_anom = partial_corr(df=df_temp_anom, x=veg_col, y=temp_anom_col,
                                             covar=precip_anom_col, method='pearson')
            temp_r_partial_anom = temp_partial_anom['r'].values[0]
            temp_p_partial_anom = temp_partial_anom['p-val'].values[0]
        except:
            temp_r_partial_anom = temp_p_partial_anom = np.nan
    else:
        temp_r_anom = temp_p_anom = temp_r_partial_anom = temp_p_partial_anom = np.nan

    # -------------------
    # Confidence intervals
    # -------------------
    ci_abs = pearson_ci(best_r_abs, len(df_tmp_abs))
    ci_anom = pearson_ci(best_r_anom, len(df_tmp_anom))
    ci_temp_abs = pearson_ci(temp_r_abs, len(df_temp_abs))
    ci_temp_anom = pearson_ci(temp_r_anom, len(df_temp_anom))

    # -------------------
    # Build results table (long format)
    # -------------------
    results_df = pd.DataFrame({
        "Habitat": [habitat]*4,
        "Satellite": [satellite]*4,
        "Variable": [precip_col, temp_col, precip_anom_col, temp_anom_col],
        "Type": ["Absolute", "Absolute", "Anomaly", "Anomaly"],
        "Best_lag": [best_lag_abs, np.nan, best_lag_anom, np.nan],
        "R_value": [best_r_abs, temp_r_abs, best_r_anom, temp_r_anom],
        "P_value": [best_p_abs, temp_p_abs, best_p_anom, temp_p_anom],
        "Partial_R": [np.nan, temp_r_partial_abs, np.nan, temp_r_partial_anom],
        "Partial_P": [np.nan, temp_p_partial_abs, np.nan, temp_p_partial_anom],
        "R_CI": [ci_abs, ci_temp_abs, ci_anom, ci_temp_anom]
    })

    # -------------------
    # Summary dict
    # -------------------
    mean_abs_corr = np.nanmean([abs(best_r_abs), abs(temp_r_abs)])
    mean_anom_corr = np.nanmean([abs(best_r_anom), abs(temp_r_anom)])
    summary = {
        "Habitat": habitat,
        "Satellite": satellite,
        "Mean_abs_corr": mean_abs_corr,
        "Mean_anom_corr": mean_anom_corr,
        "Stronger_driver": "Anomaly" if mean_anom_corr > mean_abs_corr else "Absolute",
        "N_obs": len(df_h)
    }

    return results_df, summary
```

```python
all_tables = []
summaries = []

for sat in rf_clim_df['Satellite'].dropna().unique():
    for habitat in rf_clim_df['Habitat'].dropna().unique():
        res = lagged_climate_summary_extended(rf_clim_df, habitat, sat)
        if res is not None:
            table_df, summary = res
            all_tables.append(table_df)
            summaries.append(summary)

results_df = pd.concat(all_tables, ignore_index=True)
summary_df = pd.DataFrame(summaries)

# Save
results_df.to_csv("Climate_vs_Anomaly_Correlations.csv", index=False)
summary_df.to_csv("Climate_vs_Anomaly_Summary.csv", index=False)

print(results_df.round(3))
```

```python
import numpy as np
from scipy.stats import pearsonr
import pandas as pd

def climate_lag_analysis(
    df,
    habitat,
    satellite,
    veg_col='veg_pred',
    precip_col='precip_mm',
    temp_col='temp_C',
    max_lag=6
):

    # -------------------
    # Filter data
    # -------------------
    df_h = df[
        (df['Habitat'] == habitat) &
        (df['Satellite'] == satellite)
    ].copy()

    if len(df_h) < 12:
        return None

    df_h['date'] = pd.to_datetime(df_h['date'])
    df_h['month'] = df_h['date'].dt.to_period('M')

    # -------------------
    # Monthly aggregation
    # -------------------
    df_m = (
        df_h.groupby('month')
        .agg(
            veg_mean=(veg_col, 'mean'),
            precip_mean=(precip_col, 'mean'),
            temp_mean=(temp_col, 'mean')
        )
        .reset_index()
    )

    df_m['date'] = df_m['month'].dt.to_timestamp()

    results = []

    # -------------------
    # Lag correlations
    # -------------------
    for lag in range(max_lag + 1):

        df_m[f'precip_lag{lag}'] = df_m['precip_mean'].shift(lag)

        df_lag = df_m[['veg_mean', f'precip_lag{lag}']].dropna()

        if len(df_lag) < 10:
            continue

        r, p = pearsonr(df_lag['veg_mean'], df_lag[f'precip_lag{lag}'])

        results.append({
            'Habitat': habitat,
            'Satellite': satellite,
            'Lag_months': lag,
            'Correlation_r': r,
            'p_value': p,
            'N': len(df_lag)
        })

    if len(results) == 0:
        return None

    results_df = pd.DataFrame(results)

    # -------------------
    # Best lag
    # -------------------
    best_row = results_df.iloc[results_df['Correlation_r'].abs().idxmax()]

    # -------------------
    # Temperature correlation
    # -------------------
    df_temp = df_m[['veg_mean', 'temp_mean']].dropna()

    if len(df_temp) > 10:
        r_temp, p_temp = pearsonr(df_temp['veg_mean'], df_temp['temp_mean'])
    else:
        r_temp, p_temp = np.nan, np.nan

    summary = {
        "Habitat": habitat,
        "Satellite": satellite,
        "Best_lag_months": int(best_row['Lag_months']),
        "Best_corr_r": best_row['Correlation_r'],
        "Best_corr_p": best_row['p_value'],
        "Temp_corr_r": r_temp,
        "Temp_corr_p": p_temp,
        "Veg_mean": df_m['veg_mean'].mean(),
        "Precip_mean": df_m['precip_mean'].mean(),
        "N_months": len(df_m)
    }

    return results_df, summary
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

def plot_veg_vs_precip_lag(
    df,
    lag_table_path,
    habitat,
    satellite,
    veg_col='veg_pred',
    precip_col='precip_mm',
    lag_months=None,        # if None → auto lookup
    precip_roll=None,
    outdir="Plot_RF_Clim_Lag"
):

    os.makedirs(outdir, exist_ok=True)

    # ---------------------------------
    # Load lag summary table
    # ---------------------------------
    lag_df = pd.read_csv(lag_table_path)

    # Auto fetch lag if not provided
    if lag_months is None:
        match = lag_df[
            (lag_df['Habitat'] == habitat) &
            (lag_df['Satellite'] == satellite)
        ]

        if match.empty:
            print(f"No lag found for {habitat}-{satellite}, using lag=0")
            lag_months = 0
        else:
            lag_months = int(match['Best_lag_months'].iloc[0])

    # ---------------------------------
    # Filter data
    # ---------------------------------
    df_h = df[
        (df['Habitat'] == habitat) &
        (df['Satellite'] == satellite)
    ].copy()

    df_h['date'] = pd.to_datetime(df_h['date'])
    df_h = df_h.sort_values('date')

    if len(df_h) < 12:
        print(f"Skipping {habitat}-{satellite}: insufficient data")
        return

    # ---------------------------------
    # Monthly aggregation
    # ---------------------------------
    df_h['month'] = df_h['date'].dt.to_period('M')

    df_m = (
        df_h.groupby('month')
        .agg(
            veg_mean=(veg_col, 'mean'),
            precip_mean=(precip_col, 'mean')
        )
        .reset_index()
    )

    df_m['date'] = df_m['month'].dt.to_timestamp()

    # ---------------------------------
    # Optional rolling precipitation
    # ---------------------------------
    if precip_roll is not None:
        df_m['precip_mean'] = (
            df_m['precip_mean']
            .rolling(precip_roll, min_periods=1)
            .sum()
        )

    # ---------------------------------
    # Apply lag
    # ---------------------------------
    df_m['precip_lag'] = df_m['precip_mean'].shift(lag_months)
    df_m = df_m.dropna()

    if len(df_m) == 0:
        print(f"No data after lag for {habitat}-{satellite}")
        return

    # ---------------------------------
    # Auto y limits
    # ---------------------------------
    veg_max = df_m['veg_mean'].max() * 1.2
    precip_max = df_m['precip_lag'].max() * 1.2

    # ---------------------------------
    # Plot
    # ---------------------------------
    fig, ax1 = plt.subplots(figsize=(8,4))

    # Vegetation
    line1, = ax1.plot(
        df_m['date'],
        df_m['veg_mean'],
        marker='o',
        color='forestgreen',
        lw=2,
        label='Predicted vegetation'
    )

    ax1.set_ylabel("Predicted alive vegetation cover (F_alive)", color='forestgreen')
    ax1.tick_params(axis='y', labelcolor='forestgreen')
    # ax1.set_ylim(0, veg_max)
    ax1.set_ylim(0, 0.16)

    # Precipitation
    ax2 = ax1.twinx()

    bars = ax2.bar(
        df_m['date'],
        df_m['precip_lag'],
        width=25,
        alpha=0.5,
        color='steelblue',
        label=f'Precipitation (lag={lag_months} months)'
    )

    ax2.set_ylabel("Precipitation (mm)", color='steelblue')
    ax2.tick_params(axis='y', labelcolor='steelblue')
    # ax2.set_ylim(0, precip_max)
    ax2.set_ylim(0, 160)

    # Title
    ax1.set_title(
        f"{habitat} – {satellite}\n"
        f"Vegetation vs Precipitation (Best lag = {lag_months} months)"
    )

    ax1.grid(alpha=0.3, linestyle='--')

    # Legend
    ax1.legend([line1, bars], ['Predicted vegetation', f'Precipitation (lag={lag_months})'])

    # Save
    outfile = os.path.join(
        outdir,
        f"{habitat}_{satellite}_BestLag{lag_months}.png"
    )

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.close()

    print(f"Saved: {outfile}")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

def plot_veg_vs_precip_lag(
    df,
    habitat,
    satellite,
    veg_col='veg_pred',
    precip_col='precip_mm',
    lag_months=3,
    precip_roll=None,   # optional accumulation window
    outdir="Plot_RF_Clim_Lag"
):

    os.makedirs(outdir, exist_ok=True)

    # ---------------------------
    # Filter data
    # ---------------------------
    df_h = df[
        (df['Habitat'] == habitat) &
        (df['Satellite'] == satellite)
    ].copy()

    df_h['date'] = pd.to_datetime(df_h['date'])
    df_h = df_h.sort_values('date')

    if len(df_h) < 12:
        print(f"Skipping {habitat}-{satellite}: insufficient data")
        return

    # ---------------------------
    # Monthly averaging
    # ---------------------------
    df_h['month'] = df_h['date'].dt.to_period('M')

    df_m = (
        df_h.groupby('month')
        .agg(
            veg_mean=(veg_col, 'mean'),
            precip_mean=(precip_col, 'mean')
        )
        .reset_index()
    )

    df_m['date'] = df_m['month'].dt.to_timestamp()

    # ---------------------------
    # Optional precipitation accumulation
    # ---------------------------
    if precip_roll is not None:
        df_m['precip_mean'] = (
            df_m['precip_mean']
            .rolling(precip_roll, min_periods=1)
            .sum()
        )
    precip_max = df_m["precip_mean"].max() * 1.1
    
    # ---------------------------
    # Apply time lag
    # Shift precipitation backward
    # (Rain earlier influences later vegetation)
    # ---------------------------
    df_m['precip_lag'] = df_m['precip_mean'].shift(lag_months)
    
    
    df_m = df_m.dropna()

    # ---------------------------
    # Plot
    # ---------------------------
    fig, ax1 = plt.subplots(figsize=(8,4))

    # Vegetation (points + line)
    ax1.plot(
        df_m['date'],
        df_m['veg_mean'],
        marker='o',
        color='forestgreen',
        lw=2,
        label='Predicted vegetation'
    )
    ax1.set_ylabel("Predicted alive vegetation cover (F_alive)", color='forestgreen')
    ax1.tick_params(axis='y', labelcolor='forestgreen')
    ax1.set_ylim(0, 0.16)
    
    # Precipitation bars
    ax2 = ax1.twinx()

    ax2.bar(
        df_m['date'],
        df_m['precip_lag'],
        width=25,
        alpha=0.5,
        color='steelblue',
        label=f'Precipitation (lag={lag_months} months)'
    )
    ax2.set_ylim(0, 160)
    # ax2.set_ylim(0, precip_max)
    ax2.set_ylabel("Precipitation (mm)", color='steelblue')
    ax2.tick_params(axis='y', labelcolor='steelblue')

    # Title
    ax1.set_title(
        f"{habitat} – {satellite}\n"
        f"Vegetation vs Precipitation ({lag_months}-month lag)"
    )

    ax1.grid(alpha=0.3, linestyle='--')

    # Combined legend
    lines = ax1.get_lines()
    patches = [ax2.patches[0]] if len(ax2.patches) > 0 else []

    labels = ['Predicted vegetation', f'Precipitation (lag={lag_months})']

    ax1.legend(lines + patches, labels, loc='upper left')

    # Save
    outfile = os.path.join(
        outdir,
        f"{habitat}_{satellite}_Veg_vs_Precip_Lag{lag_months}_2.png"
    )

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.close()

    print(f"Saved: {outfile}")
```

```python
# quantify temporal sampling density
# compute actual temporal resolution statistics (median revisit time per area)
# can help to decide weekly vs biweekly vs monthly based on real data distribution

import pandas as pd

def prepare_df(df):
    df = df.copy()
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date"])
    df["year"] = df["date"].dt.year
    df["month"] = df["date"].dt.month
    return df

pre_landsat = prepare_df(pre_landsat)
pre_sentinel2 = prepare_df(pre_sentinel2)
pre_modis = prepare_df(pre_modis)
pre_all = prepare_df(pre_all)

# counts per year
def count_per_year(df, name):
    out = df.groupby("year").size().reset_index(name="n_obs")
    out["dataset"] = name
    return out

year_counts = pd.concat([
    count_per_year(pre_landsat, "Landsat"),
    count_per_year(pre_sentinel2, "Sentinel2"),
    count_per_year(pre_modis, "MODIS"),
    count_per_year(pre_all, "ALL")
])

print(year_counts)

# counts per year × month
def count_per_month(df, name):
    out = (
        df.groupby(["year", "month"])
        .size()
        .reset_index(name="n_obs")
    )
    out["dataset"] = name
    return out

month_counts = pd.concat([
    count_per_month(pre_landsat, "Landsat"),
    count_per_month(pre_sentinel2, "Sentinel2"),
    count_per_month(pre_modis, "MODIS"),
    count_per_month(pre_all, "ALL")
])

print(month_counts)

# normalization
def count_unique_dates(df, name):
    out = (
        df.groupby(["year", "month"])["date"]
        .nunique()
        .reset_index(name="n_unique_dates")
    )
    out["dataset"] = name
    return out

month_dates = pd.concat([
    count_unique_dates(pre_landsat, "Landsat"),
    count_unique_dates(pre_sentinel2, "Sentinel2"),
    count_unique_dates(pre_modis, "MODIS"),
    count_unique_dates(pre_all, "ALL")
])

print(month_dates)
month_dates.to_csv('RF_prediction_image_monthly_count.csv', index=False))
```

## Investigate the temporal intervalse
- Okay, I know how to evalute the temporal intervals
- calcualte the unique date per year, count the uniqu day
- and count the uniquk week per year
- then compare this two values, will see which one have higher amount of aviblae data

### Is weekly aggregation justified?
- Temporal density metrics (per year)
- n_days -> number of unique observation days
- n_weeks -> number of unique weeks with ≥1 observation
- coverage_week -> proportion of weeks with data
- obs_per_week -> average observations per active week

### Detect irregularity / gaps
- Sampling regularity (important)
- median revisit interval (days)
- std of revisit interval

### Decision rule (practical)
- if coverage_week > 0.6 -> weekly OK
- if 0.3–0.6 -> biweekly
- if < 0.3 -> monthly

```python
import pandas as pd
import numpy as np

def temporal_metrics(df, name):
    df = df.copy()
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date"])
    
    df["year"] = df["date"].dt.year
    df["week"] = df["date"].dt.to_period("W")
    
    results = []
    
    for year, g in df.groupby("year"):
        unique_days = g["date"].nunique()
        unique_weeks = g["week"].nunique()
        
        # coverage (max 52 weeks/year)
        coverage = unique_weeks / 52.0
        
        # revisit interval
        dates_sorted = g["date"].drop_duplicates().sort_values()
        revisit = dates_sorted.diff().dt.days.dropna()
        
        median_gap = revisit.median() if len(revisit) > 0 else np.nan
        std_gap = revisit.std() if len(revisit) > 0 else np.nan
        
        # observations per active week
        obs_per_week = unique_days / unique_weeks if unique_weeks > 0 else np.nan
        
        results.append({
            "dataset": name,
            "year": year,
            "n_days": unique_days,
            "n_weeks": unique_weeks,
            "coverage_week": coverage,
            "obs_per_week": obs_per_week,
            "median_gap_days": median_gap,
            "std_gap_days": std_gap
        })
    
    return pd.DataFrame(results)
```

```python
metrics = pd.concat([
    temporal_metrics(pre_landsat, "Landsat"),
    temporal_metrics(pre_sentinel2, "Sentinel2"),
    temporal_metrics(pre_modis, "MODIS"),
    temporal_metrics(pre_all, "ALL")
])

print(metrics.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Count unique dates per week ---
def weekly_counts(df):
    df = df.copy()
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date"])
    
    # ISO week number
    df["week"] = df["date"].dt.isocalendar().week.astype(int)
    
    # Count unique dates per week
    counts = (
        df.groupby("week")["date"]
        .nunique()
        .reindex(np.arange(1, 53), fill_value=0)
    )
    
    return counts

# --- Build weekly matrix ---
def build_weekly_matrix(datasets):
    mat = []
    labels = []
    for name, df in datasets.items():
        wk = weekly_counts(df)
        mat.append(wk.values)
        labels.append(name)
    return np.array(mat), labels

# --- Count unique dates per month ---
def monthly_counts(df):
    df = df.copy()
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date"])
    
    df["month"] = df["date"].dt.month
    
    counts = (
        df.groupby("month")["date"]
        .nunique()
        .reindex(np.arange(1, 13), fill_value=0)
    )
    
    return counts

# --- Build monthly matrix ---
def build_monthly_matrix(datasets):
    mat = []
    labels = []
    for name, df in datasets.items():
        mo = monthly_counts(df)
        mat.append(mo.values)
        labels.append(name)
    return np.array(mat), labels

# --- Plot heatmap ---
def plot_heatmap_counts(mat, labels, x_labels, title):
    plt.figure(figsize=(14, 4))
    im = plt.imshow(mat, aspect="auto", cmap="viridis")
    plt.colorbar(im, label="Unique acquisition dates")
    plt.yticks(np.arange(len(labels)), labels)
    plt.xticks(np.arange(len(x_labels)), x_labels)
    plt.xlabel("Time")
    plt.title(title)
    plt.tight_layout()
    plt.show()
```

```python
# --- run functions ---
datasets = {
    "Landsat": pre_landsat,
    "Sentinel-2": pre_sentinel2,
    "MODIS": pre_modis
}

# Check data
for name, df in datasets.items():
    print(name, df.shape, df["date"].dtype, df["date"].isna().sum())

# Weekly heatmap
mat_w, labels = build_weekly_matrix(datasets)
print(mat_w.shape)  # should be (3, 52)
plot_heatmap_counts(
    mat_w,
    labels,
    x_labels=np.arange(1, 53),
    title="Weekly Acquisition Date Counts"
)

# Monthly heatmap
mat_m, labels = build_monthly_matrix(datasets)
print(mat_m.shape)  # should be (3, 12)
plot_heatmap_counts(
    mat_m,
    labels,
    x_labels=np.arange(1, 13),
    title="Monthly Acquisition Date Counts"
)
```
